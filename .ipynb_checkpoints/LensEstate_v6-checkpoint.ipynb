<h1 style=" color: #426d84; font-size: 32px; font-weight: bold;">
    🧭  Introduction
</h1>

<p> The real estate sector represents one of the most economically and socially significant markets, directly influencing investment decisions, urban development, and household wealth. In emerging markets, however, real estate ecosystems often lack transparency, standardized information, and analytical infrastructure. 
    With the rapid expansion of digital platforms and social media, large volumes of property-related data are generated daily, yet much of this information remains unstructured and underutilized. Leveraging advanced data science and artificial intelligence techniques offers an opportunity to transform fragmented market signals into reliable insights, enabling smarter valuation, forecasting, and decision-making processes. </p>

<h2 style="color: #7b98a6;" > 🛠️ 1 Market Context :</h2> 
<p>The real estate market in Tunisia operates within a fragmented and partially informal ecosystem where property information is dispersed across multiple digital platforms, social media channels, and offline networks. Listings are often inconsistent, incomplete, or unstructured, making it difficult to obtain a unified and reliable view of the market. This lack of standardization limits transparency and complicates comparative analysis across regions and property types.</p>
<h2 style="color: #7b98a6;" > 🛠️  2 Problem Statement :</h2>
<p>Despite the growing availability of online listings, the market still lacks structured analytical mechanisms capable of transforming raw property data into actionable insights. Pricing decisions are frequently subjective, valuation practices vary significantly, and regional market dynamics remain insufficiently understood. The absence of predictive tools and intelligent systems prevents stakeholders from making fully informed investment, pricing, and planning decisions.</p>
<h2 style="color: #7b98a6;" > 🛠️ 3 Project Motivation :</h2>
<p>As digital transformation accelerates across industries, the real estate sector presents a strong opportunity for intelligent automation and data-driven innovation. Buyers seek fair and transparent pricing, sellers require objective positioning strategies, investors demand forward-looking indicators, and policymakers need regional intelligence to support sustainable urban development. Addressing these needs requires more than simple data aggregation — it calls for advanced analytical capabilities that can interpret, predict, and explain market behavior.</p>
<h2 style="color: #7b98a6;" > 🛠️ Vision and Objectives :</h2>

<p>This project proposes the development of an integrated AI-powered real estate intelligence framework designed to transform heterogeneous property data into structured, explainable, and actionable knowledge. By combining automated data acquisition, multi-modal valuation models, regional trend forecasting, conversational decision support, and intelligent property visualization, the system aims to enhance transparency, reduce uncertainty, and democratize access to real estate market insights.</p>

<h2 style="color: #426d84;"> 📌 1 — Methodology: Team Data Science Process (TDSP)</h2>

<p>
The Team Data Science Process (TDSP) provides a lifecycle-based approach for developing data-driven solutions. It emphasizes iterative development, collaboration, validation, and deployment readiness. TDSP ensures that analytical development remains continuously aligned with business value.
</p>

<h3>🔹 TDSP Core Phases</h3>

<ul>
<li><b>1. Business Understanding</b> — Define strategic objectives, identify stakeholders, analyze the market context, and translate business challenges into measurable Data Science Objectives.</li>

<li><b>2. Data Acquisition & Understanding</b> — Collect heterogeneous real estate data from multiple platforms, assess data quality, explore distributions, detect inconsistencies, and design the data schema.</li>

<li><b>3. Modeling</b> — Develop, train, validate, and evaluate multi-modal valuation models, forecasting agents, conversational AI systems, and spatial reconstruction pipelines.</li>

<li><b>4. Deployment</b> — Operationalize the models into an integrated intelligence framework with automated workflows, APIs, and user interfaces.</li>
</ul>

<p>
TDSP is inherently iterative, meaning insights from modeling may lead to refined data pipelines or adjusted business assumptions. This feedback loop strengthens the robustness and real-world relevance of the final system.
</p>

<h2 style="color: #426d84;"> 🏗️ 3 — System Architecture (Detailed Overview)</h2>

<p>
The proposed architecture is designed as a modular, agent-based, and scalable AI ecosystem structured around two major layers: the <b>Data Layer</b> and the <b>Intelligence & Analytics Layer</b>. The architecture follows a pipeline-oriented yet feedback-enabled design, allowing continuous learning, data enrichment, and model refinement.
</p>

<p>
Each component of the architecture corresponds directly to one or more Data Science Objectives (DSOs), ensuring that business strategy is tightly aligned with technical implementation.
</p>



<h3 style="color:#5c8399;"> 🔹 1. Data Layer — Intelligent Data Infrastructure</h3>

<p>
The Data Layer is responsible for transforming fragmented, heterogeneous, and partially unstructured real estate information into a standardized, high-quality analytical dataset. It operates through a coordinated multi-agent system supervised by a <b>Master Orchestrator Agent</b>.
</p>

<h4>🔸 Master Orchestrator Agent</h4>
<p>
This agent acts as the central coordinator of the data ecosystem. It schedules scraping tasks, monitors data quality checks, triggers cleaning workflows, and ensures synchronization between data sources. It also handles logging, version control of datasets, and error management.
</p>

<h4>🔸 Listing Scraper Agent</h4>
<p>
This autonomous agent collects property listings from digital platforms (e.g., Tayara, Mubawab) and potentially social media channels. It extracts:
</p>
<ul>
<li>Structured attributes (price, location, surface, property type)</li>
<li>Unstructured textual descriptions</li>
<li>Property images</li>
<li>Timestamps and listing metadata</li>
</ul>

<p>
The agent is designed to handle pagination, duplicate detection, anti-bot constraints, and dynamic content extraction.
</p>

<h4>🔸 Economic & Regional Data Agent</h4>
<p>
This agent retrieves macroeconomic, demographic, and infrastructural indicators such as:
</p>
<ul>
<li>Regional income levels</li>
<li>Population density</li>
<li>Interest rates</li>
<li>Construction activity</li>
<li>Urban development signals</li>
</ul>

<p>
These features are critical for DSO3 (regional trend forecasting) and enhance valuation context modeling.
</p>

<h4>🔸 Data Cleaning & Standardization Agent</h4>
<p>
Once raw data is collected, this agent performs:
</p>
<ul>
<li>Missing value handling</li>
<li>Schema normalization</li>
<li>Currency and unit standardization</li>
<li>Deduplication across platforms</li>
<li>Outlier detection</li>
</ul>

<p>
The output is a centralized structured database that feeds the Intelligence Layer.
</p>



<h3 style="color:#5c8399;"> 🔹 2. Intelligence & Analytics Layer — AI Decision Core</h3>

<p>
This layer transforms structured data into actionable intelligence. It consists of specialized analytical modules, each corresponding to a strategic capability.
</p>



<h4>🔸 Multi-Modal Valuation Engine (DSO2)</h4>

<p>
This module integrates three complementary modeling pipelines:
</p>

<ul>
<li><b>Structured Modeling Pipeline:</b> Gradient boosting / regression models trained on numerical and categorical attributes.</li>
<li><b>NLP Pipeline:</b> Language models analyze property descriptions to detect qualitative signals such as luxury level, renovation status, urgency bias, or promotional language.</li>
<li><b>Computer Vision Pipeline:</b> CNN/ViT-based models extract visual signals from property images (condition, brightness, space perception, finishing quality).</li>
</ul>

<p>
A fusion layer combines outputs from these pipelines into a unified valuation score. Explainability mechanisms (feature importance, SHAP values) ensure interpretability.
</p>



<h4>🔸 Regional Trend Forecasting Agent (DSO3)</h4>

<p>
This module models price evolution at multiple geographic levels (city, district, neighborhood). It integrates:
</p>

<ul>
<li>Historical transaction data</li>
<li>Macroeconomic indicators</li>
<li>Demand-supply signals</li>
<li>Urban expansion patterns</li>
</ul>

<p>
Advanced time-series and machine learning models generate forward-looking indicators such as:
</p>
<ul>
<li>Price growth projections</li>
<li>Volatility indexes</li>
<li>Emerging zone detection</li>
<li>Risk exposure scores</li>
</ul>

<p>
This agent supports strategic investment and planning decisions.
</p>



<h4>🔸 Conversational AI Module (DSO4)</h4>

<p>
This module connects structured data and analytical outputs to a large language model (LLM) via Retrieval-Augmented Generation (RAG). It enables users to:
</p>

<ul>
<li>Ask natural language queries</li>
<li>Request property comparisons</li>
<li>Generate analytical summaries</li>
<li>Receive investment recommendations</li>
</ul>

<p>
The conversational agent does not generate generic responses; instead, it grounds its answers in structured market data retrieved dynamically from the database.
</p>



<h4>🔸 Spatial Reconstruction & Visualization Engine (DSO5)</h4>

<p>
This module enhances spatial interpretability by generating 2D and 3D representations of properties. It combines:
</p>

<ul>
<li>Layout inference models</li>
<li>Vision-based structural feature extraction</li>
<li>Geometric consistency validation</li>
</ul>

<p>
The goal is not architectural precision but spatial coherence, allowing users to better understand volume, layout distribution, and structural organization.
</p>



<h3 style="color:#5c8399;"> 🔹 3. Integration & Data Flow</h3>

<p>
Data flows from the Data Layer into a centralized structured repository. The Intelligence Layer consumes this repository through API-based or pipeline-based access. Each analytical module produces enriched outputs (valuation scores, forecasts, explanations, 3D models) that are reintegrated into the system and exposed through user interfaces or dashboards.
</p>

<p>
The architecture supports:
</p>

<ul>
<li>Scalability (new agents can be added)</li>
<li>Modularity (independent model updates)</li>
<li>Continuous retraining</li>
<li>Future integration with external APIs</li>
</ul>



<h3 style="color:#5c8399;"> 🔹 Architectural Philosophy</h3>

<p>
The system follows an agent-based intelligence paradigm where autonomous components specialize in specific analytical functions while remaining orchestrated within a unified ecosystem. This design ensures robustness, extensibility, and long-term adaptability to evolving market conditions.
</p>

<p>
Rather than functioning as a simple listing aggregator, the architecture operates as a full-stack real estate intelligence platform capable of transforming raw digital signals into structured, predictive, and explainable market knowledge.
</p>

<h2 style="color: #53629E;">
    🏠 Data Standardization 
</h2>

<p style="font-size:15px;">
Data standardization is the process of transforming raw, heterogeneous datasets collected from multiple platforms into a single, schema-consistent, analysis-ready master dataset. Each real estate platform in this pipeline (Mubawab, Tayara, BigDatis, and Tunisie Annonce) uses different column names, different languages (French vs. English), different categorical label conventions, and different levels of field granularity. The standardization phase resolves all of these inconsistencies through a systematic sequence of operations: loading, shape verification, column removal, feature engineering, renaming, schema alignment, and final merge. The following cells walk through each transformation step in detail.
</p>

<h2 style="color: #87BAC3;">📥 Loading Mubawab Listings</h2>

<h3>🎯 Purpose</h3>
<p>
This first step focuses on loading all raw CSV files collected from the <strong>Mubawab</strong> platform into a single unified DataFrame. Since the scraping process was split into multiple runs — each producing its own CSV file — we need to aggregate them into one cohesive dataset before any processing can begin.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
The <code>glob.glob()</code> function scans the <code>Data/raw_listings/mubaweb/</code> directory and returns the paths of all <code>.csv</code> files stored there. Each file corresponds to a scraping batch. These files are then read individually using <code>pd.read_csv()</code> inside a generator expression and concatenated using <code>pd.concat()</code> with <code>ignore_index=True</code>, which resets the row index to avoid conflicts between files. The result is a single flat DataFrame called <code>mubaweb_df</code> containing all Mubawab listings discovered across all scraping runs. The <code>.head()</code> call at the end gives us a quick preview of the first rows to visually verify the data was loaded correctly.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Keeping data in separate files is a common practice when scraping in batches, but it creates fragmentation. Merging all files at this early stage ensures a single consistent working object throughout the pipeline. Failing to reset the index could cause duplicate index issues during joins or filtering later in the workflow.
</p>

````python
import pandas as pd
import glob

# Get all CSV file paths
mubaweb_files = glob.glob("Data/raw_listings/mubaweb/*.csv")

# Read and combine them into one DataFrame
mubaweb_df = pd.concat(
    (pd.read_csv(file) for file in mubaweb_files),
    ignore_index=True
)

# Check the result
mubaweb_df.head()
````

</p>

<h3>🔍 Technical Breakdown</h3>
<p>
<code>mubaweb_df.shape</code> returns a tuple <code>(rows, columns)</code>. The <code>print()</code> statement displays both dimensions with a descriptive label. This is a lightweight sanity check that costs no computation but saves significant debugging time.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Knowing the shape before any transformations establishes a reference point. If columns are unexpectedly missing or row counts seem off after subsequent transformations, we can trace changes back to this baseline. It also gives us a first quantitative understanding of the dataset volume before deeper exploration.
</p>

````python
print("number of lines " ,mubaweb_df.shape)
````

<h2 style="color: #87BAC3;">🏷️ Assigning Transaction Type to Mubawab</h2>

<h3>🎯 Purpose</h3>
<p>
Mubawab's raw data does not include a standardized <code>transaction_type</code> column. Since all listings scraped from Mubawab correspond exclusively to <strong>sale transactions</strong> (as opposed to rentals), we can safely add a constant column marking every row as <em>"Sale"</em>. This is a critical step for consistency when merging with other platforms later.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
A new column <code>transaction_type</code> is assigned a constant string value <code>"Sale"</code> for every row. The <code>.value_counts()</code> call then verifies that the assignment was applied uniformly across the dataset. This simple validation confirms that no row was accidentally left without a transaction type.

<h2 style="color: #87BAC3;">🚿 Adding Derived Feature Flags for Mubawab</h2>

<h3>🎯 Purpose</h3>
<p>
This step enriches the Mubawab dataset with two additional boolean feature flags that will be needed for schema alignment with other platforms. The first flag, <code>has_bathrooms</code>, is derived from existing data. The second, <code>has_heating</code>, is set to a default value since the information was not available in the Mubawab scrape.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
<code>mubaweb_df["has_bathrooms"]</code> is created by checking whether the <code>bathrooms</code> column contains a non-null value using <code>.notna()</code>. This produces a boolean column — <code>True</code> if the listing specifies a bathroom count, <code>False</code> otherwise. In parallel, <code>has_heating</code> is set to <code>False</code> across the board since heating information was not captured during the Mubawab scraping phase. These flags follow the naming convention used across all datasets to ensure compatibility during the merge step.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Derived binary flags are a standard feature engineering practice in real estate modeling. They allow machine learning models to treat the <em>presence or absence</em> of an amenity as an independent signal, even when the underlying numeric attribute is missing. Setting <code>has_heating</code> to <code>False</code> rather than <code>None</code> prevents null propagation issues during model training and schema alignment.
</p>

````python
mubaweb_df["has_bathrooms"] = mubaweb_df["bathrooms"].notna()
mubaweb_df["has_heating"] = False
````

<h3>⚠️ Why This Matters</h3>

In data preparation workflows, it is easy to lose track of the schema after multiple transformation steps. <h3>🔍 Technical Breakdown</h3>

This is a non-destructive operation on the source files since the transformations apply only to the in-memory object.

<h3>⚠️ Why This Matters</h3>
<p>
Retaining columns that cannot be aligned across sources introduces schema mismatches during concatenation. More importantly, incomplete or inconsistent fields like <code>price_type</code> can mislead models during training if left in the dataset without proper handling. <p>
The <code>district</code> column is removed from the Mubawab dataset at this stage. While district-level geographic precision is desirable, this field was not available or consistently populated across all platforms in the pipeline, making it unsuitable for inclusion in the unified schema at this stage.

Geospatial granularity is valuable in real estate analysis, but only when it is reliable and available across datasets. <h3>🔍 Technical Breakdown</h3>

<h3>⚠️ Why This Matters</h3>

<h2 style="color: #87BAC3;">🔤 Renaming Columns to English Standard for Mubawab</h2>

<h3>🎯 Purpose</h3>
<p>
This step translates all French-language amenity column names from the Mubawab scrape into their English equivalents using a standardized naming convention. This is a critical harmonization step that ensures that all four platform datasets share identical column names, enabling a clean final merge.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
<code>pd.DataFrame.rename(columns={...}, inplace=True)</code> applies a mapping dictionary that translates each source column name to its target English name. For example, <code>has_ascenceur</code> (French for elevator) becomes <code>has_elevator</code>, and <code>has_piscine</code> (French for pool) becomes <code>has_pool</code>. Note that both <code>has_ascenceur</code> and <code>has_ascenseur</code> (two common spelling variants in the scraped data) are mapped to the same target <code>has_elevator</code>, handling a real-world spelling inconsistency discovered during data exploration.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Schema harmonization across multi-source datasets is one of the most critical and error-prone steps in data integration. A single column name mismatch between two DataFrames will cause misaligned columns during concatenation, producing unexpected <code>NaN</code> values instead of properly merged data. By standardizing to English at the source level, the final merge becomes a straightforward union rather than a complex alignment exercise.
</p>

<h2 style="color: #87BAC3;">📥 Loading Tayara Listings</h2>

It is the analytical equivalent of a schema diff tool used in database migration workflows.

````python
# Get sets of columns
tayara_cols = set(tayara_df.columns)
mubaweb_cols = set(mubaweb_df.columns)

# All unique columns
all_cols = sorted(tayara_cols | mubaweb_cols)

# Prepare aligned DataFrame
comparison_list = []
for col in all_cols:
    tayara_val = col if col in tayara_cols else ""
    mubaweb_val = col if col in mubaweb_cols else ""
    comparison_list.append({"tayara_df": tayara_val, "mubaweb_df": mubaweb_val})

comparison_df = pd.DataFrame(comparison_list)
comparison_df
````

<h2 style="color: #87BAC3;">🧩 Injecting Missing Columns into Tayara from Mubawab Schema</h2>

<h3>🎯 Purpose</h3>
<p>
Based on the schema comparison performed in the previous step, this cell automatically adds all columns present in the Mubawab DataFrame but absent from the Tayara DataFrame, filling them with appropriate default values based on their data type. This is the final alignment step before the two datasets can be concatenated without schema conflicts.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
The code computes the set difference between Mubawab columns and Tayara columns, then iterates over each missing column. For each missing column, it checks the data type in the Mubawab DataFrame:
</p>
<ul>
<li><strong>Boolean columns</strong> are filled with <code>False</code> — assuming the absence of an amenity when no data is available is safer than assuming its presence.</li>
<li><strong>Numeric columns</strong> are filled with <code>np.nan</code> — a standard missing value representation for numerical features in Pandas and scikit-learn.</li>
<li><strong>All other columns</strong> (strings, objects, categories) are filled with <code>None</code> — the Python null equivalent, which Pandas stores as <code>NaN</code> for object dtypes.</li>
</ul>
<p>
The list of added columns is printed at the end for auditability.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Type-aware default value assignment is a sophisticated and important detail. Using <code>False</code> for boolean amenity flags reflects the real-world assumption that Tayara listings without an explicit elevator flag do not have an elevator — rather than leaving this as an unknown. Using <code>np.nan</code> for numerics preserves the distinction between "zero" and "unknown", which is critical for imputation strategies and model interpretability. This step ensures the concatenation operation produces a clean, type-consistent merged DataFrame rather than one polluted with mistyped defaults.
</p>

````python
import numpy as np

# Columns missing in tayara
missing_cols = set(mubaweb_df.columns) - set(tayara_df.columns)

for col in missing_cols:
    dtype = mubaweb_df[col].dtype

    # Boolean → set False
    if dtype == bool:
        tayara_df[col] = False

    # Numeric → use np.nan
    elif np.issubdtype(dtype, np.number):
        tayara_df[col] = np.nan

    # Everything else (object, string, category, etc.)
    else:
        tayara_df[col] = None

print("Added columns:", missing_cols)
````

<h2 style="color: #87BAC3;">✅ Re-verifying Schema Alignment After Missing Column Injection</h2>

<h3>🎯 Purpose</h3>
<p>
This step repeats the column comparison matrix from Step 32, now reflecting the state of both DataFrames after the missing column injection. The goal is to confirm that all schema gaps have been closed and that both DataFrames now share identical column sets — a prerequisite for a clean merge.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
The same comparison logic from Step 32 is applied again. After a successful missing column injection, the expected result is a comparison matrix where every row shows both <code>tayara_df</code> and <code>mubaweb_df</code> populated — meaning no column is exclusive to either DataFrame. Any remaining empty cell in the output would indicate a gap that was missed and needs to be addressed before concatenation.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Schema verification after alignment is not optional — it is the only reliable way to confirm that the injection step performed correctly. Attempting to concatenate two DataFrames with mismatched schemas will not raise an error in Pandas; instead, it silently produces a result with misaligned columns and unexpected null columns. This silent failure mode makes pre-merge schema verification absolutely critical in multi-source data pipelines.
</p>

````python
# Get sets of columns
tayara_cols = set(tayara_df.columns)
mubaweb_cols = set(mubaweb_df.columns)

# All unique columns
all_cols = sorted(tayara_cols | mubaweb_cols)

# Prepare aligned DataFrame
comparison_list = []
for col in all_cols:
    tayara_val = col if col in tayara_cols else ""
    mubaweb_val = col if col in mubaweb_cols else ""
    comparison_list.append({"tayara_df": tayara_val, "mubaweb_df": mubaweb_val})

comparison_df = pd.DataFrame(comparison_list)
comparison_df
````

<h2 style="color: #87BAC3;">🏗️ Loading, Standardizing, and Aligning BigDatis Listings</h2>

<h3>🎯 Purpose</h3>
<p>
This comprehensive cell handles the complete standardization workflow for the <strong>BigDatis</strong> platform dataset in a single block. Following the same principles applied to Mubawab and Tayara, it loads the raw files, removes irrelevant columns, renames fields to the shared schema, adds missing columns, normalizes categorical values, and validates the result — all within one organized, commented pipeline.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
The cell is structured into seven sequential stages:
</p>
<ul>
<li><strong>Stage 1 — Load:</strong> All BigDatis CSV files are discovered and concatenated using the same <code>glob</code> + <code>pd.concat</code> pattern. The shape and head are printed for immediate validation.</li>
<li><strong>Stage 2 — Drop:</strong> Platform-specific columns not aligned with the shared schema (<code>price_type</code>, <code>district</code>, <code>language</code>) are removed, following the same rationale as with Mubawab.</li>
<li><strong>Stage 3 — Rename:</strong> BigDatis uses French field names similar to Mubawab but with some variations (e.g., <code>has_climatisseur</code> instead of <code>has_climatisation</code>, <code>has_meubles</code> instead of <code>has_meuble</code>). All are mapped to the shared English schema.</li>
<li><strong>Stage 4 — Add Missing Columns:</strong> Fields present in the shared schema but absent from BigDatis are added with appropriate defaults: <code>country</code> = "Tunisia", <code>total_rooms</code> = None, <code>last_updated</code> = None, and <code>has_standing</code> = False.</li>
<li><strong>Stage 5 — Normalize transaction_type:</strong> BigDatis uses French labels ("Vente", "Location") instead of the shared English labels ("Sale", "Rent"). A mapping dictionary is applied via <code>.map(lambda x: transaction_map.get(x, x))</code>, which translates known values while safely passing through any unknown values unchanged.</li>
<li><strong>Stage 6 — Verify Shape and Columns:</strong> Shape and column list are printed to confirm the result.</li>
<li><strong>Stage 7 — Deduplication Check:</strong> Before the merge, all three DataFrames are checked for duplicate column names — a common issue when scraping introduces repeated header rows. Duplicate columns are removed using <code>.loc[:, ~df.columns.duplicated()]</code>.</li>
</ul>

<h3>⚠️ Why This Matters</h3>
<p>
The transaction type normalization step is particularly important. Mixing French and English labels in the same column after the merge would break any analysis or model that relies on <code>transaction_type</code> as a categorical filter. Normalizing at the source level — before the merge — ensures categorical consistency throughout the pipeline. The duplicate column detection step addresses a subtle but dangerous issue: if a CSV file contains duplicate column headers (which can happen due to malformed scraping output), Pandas will silently keep both, leading to unpredictable behavior during merges and aggregations.
</p>

````python
# ─────────────────────────────────────────────
# 1. Load all bigdatis CSV files
# ─────────────────────────────────────────────
bigdatis_files = glob.glob("DATA/raw_listings/big_datis/*.csv")

bigdatis_df = pd.concat(
    (pd.read_csv(file) for file in bigdatis_files),
    ignore_index=True
)

print("Shape after loading:", bigdatis_df.shape)
print(bigdatis_df.head())

# ─────────────────────────────────────────────
# 2. Drop columns not needed for the merge
# ─────────────────────────────────────────────
bigdatis_df = bigdatis_df.drop(columns=["price_type"])
bigdatis_df = bigdatis_df.drop(columns=["district"])
bigdatis_df = bigdatis_df.drop(columns=["language"])

# ─────────────────────────────────────────────
# 3. Rename columns to match the shared schema
# ─────────────────────────────────────────────
bigdatis_df.rename(columns={
    "id"                  : "listing_id",
    "link"                : "listing_url",
    "surface"             : "surface_m2",
    "number_of_rooms"     : "rooms",
    "number_of_salle_de_bain": "bathrooms",
    "total_floor"         : "total_floors",
    "image_url"           : "image_urls",
    "scraped_at"          : "scrape_timestamp",
    "has_ascenseur"       : "has_elevator",
    "has_cave"            : "has_basement",
    "has_chauffage"       : "has_heating",
    "has_climatisseur"    : "has_air_conditioning",
    "has_jardin"          : "has_garden",
    "has_meubles"         : "has_furniture",
    "has_piscine"         : "has_pool",
    "has_terrasse"        : "has_terrace",
    "has_vue_sur_mer"     : "has_sea_view",
}, inplace=True)

# ─────────────────────────────────────────────
# 4. Add missing columns (present in mubawab/tayara but not in bigdatis)
# ─────────────────────────────────────────────
bigdatis_df["country"]      = "Tunisia"
bigdatis_df["total_rooms"]  = None
bigdatis_df["last_updated"] = None
bigdatis_df["has_standing"] = False

# floor already exists in bigdatis, keep it as-is

# ─────────────────────────────────────────────
# 5. Normalize transaction_type to match the shared schema
#    (bigdatis uses "Vente" / "Location", others use "Sale" / "Rent")
# ─────────────────────────────────────────────
transaction_map = {
    "Vente"    : "Sale",
    "Location" : "Rent",
}
bigdatis_df["transaction_type"] = bigdatis_df["transaction_type"].map(
    lambda x: transaction_map.get(x, x)
)

# ─────────────────────────────────────────────
# 6. Verify final columns
# ─────────────────────────────────────────────
print("\nShape after treatment:", bigdatis_df.shape)
print("\nColumns:", bigdatis_df.columns.tolist())

# ─────────────────────────────────────────────
# 7. (Optional) Compare columns across the three datasets
#    Uncomment once mubaweb_df and tayara_df are also loaded
# ─────────────────────────────────────────────


# ─────────────────────────────────────────────
# Check for duplicates in each df
print("mubaweb duplicates:", mubaweb_df.columns[mubaweb_df.columns.duplicated()].tolist())
print("tayara duplicates: ", tayara_df.columns[tayara_df.columns.duplicated()].tolist())
print("bigdatis duplicates:", bigdatis_df.columns[bigdatis_df.columns.duplicated()].tolist())

# Remove duplicate columns (keeps the first occurrence)
mubaweb_df  = mubaweb_df.loc[:, ~mubaweb_df.columns.duplicated()]
tayara_df   = tayara_df.loc[:, ~tayara_df.columns.duplicated()]
bigdatis_df = bigdatis_df.loc[:, ~bigdatis_df.columns.duplicated()]
````

<h2 style="color: #87BAC3;">🗺️ Loading, Standardizing, and Aligning Tunisie Annonce Listings</h2>

<h3>🎯 Purpose</h3>
<p>
This comprehensive cell applies the full standardization workflow to the <strong>Tunisie Annonce</strong> platform dataset. It introduces one unique challenge not present in the other datasets: the geographic information for this platform is stored in a single compound <code>region</code> string (e.g., "Localisation Tunisie>Tunis>La Marsa>El Aouina") that must be parsed and split into the shared <code>governorate</code> and <code>city</code> fields before the merge.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
The standardization is organized into seven stages:
</p>
<ul>
<li><strong>Stage 1 — Load:</strong> All Tunisie Annonce CSV files are loaded and concatenated. Shape and head are verified immediately.</li>
<li><strong>Stage 2 — Drop:</strong> <code>price_type</code> and <code>language</code> are removed. Note that <code>district</code> is not explicitly dropped here since the geographic parsing step (Stage 3) replaces the entire <code>region</code> field.</li>
<li><strong>Stage 3 — Parse Region:</strong> A custom <code>parse_region()</code> function is defined and applied using <code>.apply()</code>. This function splits the compound region string on the "&gt;" delimiter and extracts the governorate (index 1) and city (index 2) as separate fields. The index 0 ("Localisation Tunisie") and index 3 (district) are discarded. Null-safe handling is included via <code>pd.isna()</code> to prevent errors on missing values.</li>
<li><strong>Stage 4 — Rename:</strong> Platform-specific field names are mapped to the shared English schema, following the same renaming conventions applied to BigDatis.</li>
<li><strong>Stage 5 — Add Missing Columns:</strong> <code>country</code>, <code>total_rooms</code>, <code>last_updated</code>, <code>scrape_timestamp</code>, and <code>has_standing</code> are added with appropriate defaults.</li>
<li><strong>Stage 6 — Normalize transaction_type:</strong> French labels ("Vente", "Location") are mapped to English equivalents ("Sale", "Rent") using the same translation dictionary as BigDatis.</li>
<li><strong>Stage 7 — Verify:</strong> Shape, column list, and duplicate column check are printed.</li>
</ul>

<h3>⚠️ Why This Matters</h3>
<p>
The region parsing logic is the most technically sophisticated standardization step across all four platforms. It demonstrates an important principle: when geographic information is stored in compound strings rather than structured fields, a custom parsing function must be carefully designed to handle edge cases — including missing values, variable-length strings, and unexpected formats. The null-safe implementation using <code>pd.isna()</code> ensures that listings without a region string do not crash the pipeline. The discarding of the district-level component (index 3) is consistent with the decision made during Mubawab and BigDatis standardization to exclude district-level geography from the shared schema, maintaining geographic field consistency across all four datasets.
</p>

````python
import pandas as pd
import glob

# ─────────────────────────────────────────────
# 1. Load the tunisie_annonce CSV file(s)
#    (adjust the path/glob to match your folder structure)
# ─────────────────────────────────────────────
tunisie_files = glob.glob("DATA/raw_listings/tunisieannonce/*.csv")

tunisie_df = pd.concat(
    (pd.read_csv(file) for file in tunisie_files),
    ignore_index=True
)

print("Shape after loading:", tunisie_df.shape)
print(tunisie_df.head())

# ─────────────────────────────────────────────
# 2. Drop columns not needed for the merge
# ─────────────────────────────────────────────
tunisie_df = tunisie_df.drop(columns=["price_type"])
tunisie_df = tunisie_df.drop(columns=["language"])

# ─────────────────────────────────────────────
# 3. Parse 'region' into 'governorate' and 'city'
#    Raw format: "Localisation Tunisie>Tunis>La Marsa>El Aouina"
#    We take index [1] as governorate and [2] as city
# ─────────────────────────────────────────────
def parse_region(region_str):
    if pd.isna(region_str):
        return pd.Series({"governorate": None, "city": None})
    parts = region_str.split(">")
    # parts[0] = "Localisation Tunisie", parts[1] = governorate, parts[2] = city, parts[3] = district (ignored)
    governorate = parts[1].strip() if len(parts) > 1 else None
    city        = parts[2].strip() if len(parts) > 2 else None
    return pd.Series({"governorate": governorate, "city": city})

tunisie_df[["governorate", "city"]] = tunisie_df["region"].apply(parse_region)
tunisie_df = tunisie_df.drop(columns=["region"])

# ─────────────────────────────────────────────
# 4. Rename columns to match the shared schema
# ─────────────────────────────────────────────
tunisie_df.rename(columns={
    "id"                     : "listing_id",
    "link"                   : "listing_url",
    "surface"                : "surface_m2",
    "number_of_rooms"        : "rooms",
    "number_of_salle_de_bain": "bathrooms",
    "total_floor"            : "total_floors",
    "image_url"              : "image_urls",
    "has_ascenseur"          : "has_elevator",
    "has_cave"               : "has_basement",
    "has_chauffage"          : "has_heating",
    "has_climatisseur"       : "has_air_conditioning",
    "has_jardin"             : "has_garden",
    "has_meubles"            : "has_furniture",
    "has_piscine"            : "has_pool",
    "has_terrasse"           : "has_terrace",
    "has_vue_sur_mer"        : "has_sea_view",
}, inplace=True)

# ─────────────────────────────────────────────
# 5. Add missing columns (present in other dfs but not here)
# ─────────────────────────────────────────────
tunisie_df["country"]          = "Tunisia"
tunisie_df["total_rooms"]      = None
tunisie_df["last_updated"]     = None
tunisie_df["scrape_timestamp"] = None
tunisie_df["has_standing"]     = False

# ─────────────────────────────────────────────
# 6. Normalize transaction_type
#    tunisie_annonce uses "Vente" / "Location" → "Sale" / "Rent"
# ─────────────────────────────────────────────
transaction_map = {
    "Vente"    : "Sale",
    "Location" : "Rent",
}
tunisie_df["transaction_type"] = tunisie_df["transaction_type"].map(
    lambda x: transaction_map.get(x, x)
)

# ─────────────────────────────────────────────
# 7. Verify
# ─────────────────────────────────────────────
print("\nShape after treatment:", tunisie_df.shape)
print("\nColumns:", tunisie_df.columns.tolist())
print("\nDuplicate columns:", tunisie_df.columns[tunisie_df.columns.duplicated()].tolist())
````

<h2 style="color: #87BAC3;">🔗 Merging All Four Datasets into a Unified DataFrame</h2>

<h3>🎯 Purpose</h3>
<p>
This is the culminating step of the data standardization phase: all four standardized DataFrames — Mubawab, Tayara, BigDatis, and Tunisie Annonce — are concatenated into a single unified <code>merged_df</code>. This merged dataset becomes the master analytical dataset that will power all subsequent modeling, forecasting, and intelligence modules in the pipeline.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
Before the merge, the Tunisie Annonce DataFrame is deduplicated for column names using <code>.loc[:, ~tunisie_df.columns.duplicated()]</code>, consistent with the deduplication applied to the other three DataFrames in Step 35. The merge itself is performed using <code>pd.concat([mubaweb_df, tayara_df, bigdatis_df, tunisie_df], ignore_index=True)</code>. This operation stacks all four DataFrames vertically — appending rows from each source beneath the previous — while aligning on column names. Because all four DataFrames now share the same schema after the standardization steps, every column in the result contains data from all four platforms (or null values where a platform didn't capture that field). The final shape is printed for immediate size validation.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
The quality of this merge is entirely dependent on the rigor of all preceding standardization steps. A successful merge with no unexpected null columns confirms that the schema alignment was performed correctly across all four platforms. Any schema mismatches — column name differences, type inconsistencies, or missing columns — would silently manifest as unexpected null columns in the merged result. The shape validation at the end provides the first opportunity to catch such issues: the total row count should equal the sum of rows from all four individual DataFrames, and the column count should equal the size of the shared schema.
</p>

````python
# ─────────────────────────────────────────────
# 8. Merge all four datasets
#    Uncomment once all four dataframes are ready
# ─────────────────────────────────────────────
# # Remove duplicate columns from each df before merging
tunisie_df  = tunisie_df.loc[:, ~tunisie_df.columns.duplicated()]

merged_df = pd.concat([mubaweb_df, tayara_df, bigdatis_df, tunisie_df], ignore_index=True)
print("\nMerged shape:", merged_df.shape)
````

<h2 style="color: #87BAC3;">💾 Assigning IDs and Saving the Merged Dataset</h2>

<h3>🎯 Purpose</h3>
<p>
This final step assigns a unique sequential integer identifier to every row in the merged DataFrame and saves the result as a CSV file. The saved file becomes the primary input for all subsequent analytical phases of the LensEstate pipeline, including exploratory data analysis, feature engineering, and machine learning model training.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
<code>merged_df.insert(0, "id", range(1, len(merged_df) + 1))</code> inserts a new column <code>id</code> at position 0 (the first column) with integer values from 1 to the total number of rows. Using <code>range(1, N+1)</code> produces 1-indexed IDs, which are generally more intuitive for human reading and database compatibility than 0-indexed identifiers. The DataFrame is then exported to CSV using <code>.to_csv()</code> with <code>index=False</code> to exclude the Pandas default integer index from the file, since the explicitly created <code>id</code> column already serves this purpose. The save path is printed as confirmation that the operation completed successfully.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Persisting the merged dataset to disk is a critical checkpoint in the pipeline. It means that all the standardization work performed across Steps 1–37 is preserved and does not need to be re-executed for every subsequent analysis. The unique <code>id</code> column is essential for data lineage tracking — it allows any processed or predicted result to be traced back to the original raw listing without ambiguity. Without a stable unique identifier, it becomes impossible to perform reliable joins, deduplications, or post-analysis result linking against the merged dataset.
</p>

````python
merged_df.insert(0, "id", range(1, len(merged_df) + 1))
merged_df.to_csv("Data/raw_listings/merged_listings.csv", index=False)
print("Saved to Data/raw_listings/merged_listings.csv")
````

<h2 style="color: #87BAC3;">📥 Loading Houni Blog Articles</h2>

<h3>🎯 Purpose</h3>
<p>
This cell loads the blog articles scraped from <strong>Houni.tn</strong>, a Tunisian real estate information platform that publishes neighborhood guides, mortgage comparison articles, legal advice, and lifestyle-oriented property content. Houni's editorial tone is more consumer-facing and educational, targeting first-time buyers and renters rather than professional investors.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
The Houni CSV was scraped with a character encoding inconsistency: the file was saved in <strong>Latin-1</strong> (ISO-8859-1) rather than UTF-8, causing French accented characters (é, è, ê, à, ç, etc.) to appear as garbled sequences (e.g., <code>crÃ©dit</code> instead of <code>crédit</code>) when read naively. To fix this, the file is first loaded with <code>encoding="latin-1"</code> to prevent parsing errors, and then each string value is re-encoded to Latin-1 bytes and decoded as UTF-8 — the standard mojibake correction technique. The <code>fix_mojibake()</code> helper function handles this safely, with a try/except block that leaves values unchanged if the correction fails (e.g., for already-clean strings or irreparable sequences).
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Encoding corruption is one of the most common and damaging issues in web-scraped text data. If left uncorrected, garbled characters would propagate into every downstream NLP operation — breaking tokenizers, polluting vocabulary dictionaries, and producing nonsensical embeddings. Fixing the encoding at the loading stage is the only correct remedy: it ensures that the Houni articles enter the pipeline as clean, human-readable French text, fully compatible with French-language NLP models and tokenizers.
</p>

````python
# Load Houni blog articles — requires encoding fix (Latin-1 mojibake)
def fix_mojibake(text):
    """Fix text incorrectly read as Latin-1 that was originally UTF-8."""
    if not isinstance(text, str):
        return text
    try:
        return text.encode('latin-1').decode('utf-8')
    except (UnicodeDecodeError, UnicodeEncodeError):
        return text  # Return as-is if correction fails

houni_blog_df = pd.read_csv("Data/articles/houni/houni_blog_articles.csv", encoding="latin-1")
houni_blog_df = houni_blog_df.apply(lambda col: col.map(fix_mojibake))

print("Shape:", houni_blog_df.shape)
print("Columns:", houni_blog_df.columns.tolist())
houni_blog_df.head()
````

<h2 style="color: #87BAC3;">📊 Schema Comparison Across All Three Article Sources</h2>

<h3>🎯 Purpose</h3>
<p>
Before merging, this cell performs a final schema verification by comparing the column lists of all three article DataFrames side by side. This confirms that the standardization steps applied to each source have produced identical schemas, ensuring a clean and conflict-free concatenation.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
The column sets of all three DataFrames are compared using a union-based alignment table — the same technique used during the listing dataset standardization. For each column in the union of all names, the table shows whether it is present in each source. After the selection and labeling steps applied above, all three DataFrames should show identical columns: <code>title</code>, <code>content</code>, and <code>source</code> — resulting in a perfectly aligned comparison table with no empty cells.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
Schema verification before merge is a non-negotiable checkpoint. Even with three simple three-column DataFrames, a subtle naming discrepancy (e.g., <code>Content</code> vs <code>content</code>) would silently produce a six-column merged DataFrame with half the values as nulls. This check costs nothing and prevents a class of silent bugs that are difficult to detect once data has been combined.
</p>

````python
# Compare schemas across all three article sources
all_cols = sorted(set(mubawab_blog_df.columns) | set(bnb_blog_df.columns) | set(houni_blog_df.columns))

comparison = []
for col in all_cols:
    comparison.append({
        "mubawab_blog_df": col if col in mubawab_blog_df.columns else "",
        "bnb_blog_df":     col if col in bnb_blog_df.columns else "",
        "houni_blog_df":   col if col in houni_blog_df.columns else "",
    })

pd.DataFrame(comparison)
````

<h2 style="color: #87BAC3;">🔗 Merging All Three Article Datasets</h2>

<h3>🎯 Purpose</h3>
<p>
This is the culminating step of the article standardization phase. All three standardized blog DataFrames — from Mubawab, BnB, and Houni — are concatenated into a single unified <code>merged_articles_df</code>. The result is a clean, source-labeled, text-ready corpus covering the full editorial landscape of the Tunisian real estate market.
</p>

<h3>🔍 Technical Breakdown</h3>
<p>
<code>pd.concat([mubawab_blog_df, bnb_blog_df, houni_blog_df], ignore_index=True)</code> stacks all three DataFrames vertically, resetting the index. Because all three share the same three-column schema (<code>title</code>, <code>content</code>, <code>source</code>), the concatenation produces a perfectly aligned result with no null columns. The final shape and a distribution of articles per source (via <code>value_counts()</code>) are printed to confirm the merge was successful and to give a clear picture of the relative contribution of each platform to the corpus.
</p>

<h3>⚠️ Why This Matters</h3>
<p>
The <code>source</code> column distribution printed at the end of this cell is particularly informative: it shows at a glance whether the corpus is dominated by a single platform (which could bias topic models or retrieval systems toward that source's editorial perspective). <h2 style="color: #87BAC3;">🕳️ Missing Values Analysis</h2>

<h3>🎯 What This Cell Does</h3>
<p>This cell quantifies and visualizes the extent of missing data across all columns — one of the most important diagnostic steps in any real estate EDA. The bar chart shows the percentage of missing values per column, and the heatmap reveals the structural pattern of missingness across a 500-row sample.</p>

<h3>📊 Results & Observations</h3>
<p>The results reveal a severe missingness problem in several critical columns. <code>last_updated</code> and <code>total_floors</code> are missing in <strong>100%</strong> of records, making them entirely unusable without a dedicated re-scraping effort. <code>floor</code> is also nearly fully missing at <strong>100%</strong>. <code>total_rooms</code> is missing in <strong>91.3%</strong> of rows. Core modeling features are also significantly impacted: <code>rooms</code> is missing in <strong>75%</strong> of listings, <code>bathrooms</code> in <strong>67%</strong>, <code>agency_name</code> in <strong>64.4%</strong>, and <code>bedrooms</code> in <strong>63%</strong>. Even the most fundamental fields like <code>property_type</code> (<strong>41%</strong> missing) and <code>governorate</code> (<strong>39.2%</strong> missing) show significant gaps.</p>
<p>More encouragingly, the most analytically critical field — <code>price</code> — is only missing in <strong>5.5%</strong> of listings, and <code>description</code> is nearly complete at just <strong>0.1%</strong> missing.</p>

<h3>💡 Key Takeaways</h3>
<p>The missingness pattern is not random — it is strongly structured by platform. Columns like <code>floor</code>, <code>total_floors</code>, and <code>total_rooms</code> were simply not captured by most platforms' scrapers. This means imputation strategies must be platform-aware and feature-specific. Columns with over 90% missing (<code>last_updated</code>, <code>total_floors</code>, <code>floor</code>) should be dropped from the modeling pipeline. The <code>rooms</code> and <code>bathrooms</code> fields, despite significant missingness, carry enough signal to be retained with careful imputation.</p>

````python
# Missing values analysis
missing = df.isnull().mean().sort_values(ascending=False)
missing_df = missing[missing > 0].reset_index()
missing_df.columns = ["column", "missing_pct"]
missing_df["missing_pct_label"] = (missing_df["missing_pct"] * 100).round(1).astype(str) + "%"

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# Bar chart
colors = [PALETTE[0] if v < 0.5 else PALETTE[4] for v in missing_df["missing_pct"]]
axes[0].barh(missing_df["column"], missing_df["missing_pct"] * 100, color=colors)
axes[0].axvline(50, color="red", linestyle="--", linewidth=1, label="50% threshold")
axes[0].set_xlabel("Missing (%)")
axes[0].set_title("Missing Values per Column")
axes[0].legend()
for i, (v, lbl) in enumerate(zip(missing_df["missing_pct"], missing_df["missing_pct_label"])):
    axes[0].text(v * 100 + 0.5, i, lbl, va="center", fontsize=8)

# Heatmap sample (500 rows)
sample = df[missing_df["column"].tolist()].sample(min(500, len(df)), random_state=42).isnull()
sns.heatmap(sample.T, cmap=["#d9e6ec", "#426d84"], cbar=False,
            yticklabels=True, xticklabels=False, ax=axes[1])
axes[1].set_title("Missingness Pattern (500-row sample)")
axes[1].set_xlabel("Listings (sample)")

plt.tight_layout()
plt.savefig("Data/eda_missing_values.png", bbox_inches="tight")
plt.show()
print(missing_df.to_string(index=False))
````

Prices were pre-filtered to strictly positive, finite values to avoid artifacts from zero-priced or corrupted entries.</p>

<h3>📊 Results & Observations</h3>
<p>After cleaning, <strong>54,368</strong> valid prices were retained, with 1,865 zero/negative/non-numeric entries removed. A critical issue is revealed: the rent price series returns <strong>0 valid records</strong>, meaning all rental listings in the dataset have their <code>transaction_type</code> labeled in a non-standard way (<code>À Louer</code> rather than <code>Rent</code>) — confirming the normalization gap identified in the previous cell. Sale price analysis shows a median of <strong>370,000 TND</strong> and a mean of approximately <strong>283 million TND</strong>, reflecting the extreme positive skew caused by luxury property outliers. The maximum recorded price is an extraordinary <strong>6.3 trillion TND</strong> — an obvious data entry error. The 25th percentile is 214,000 TND and the 75th percentile is 630,000 TND, placing the interquartile range of the bulk of the market between roughly 214K and 630K TND.</p>

<h3>💡 Key Takeaways</h3>
<p>The price column requires two immediate cleaning actions: (1) normalizing transaction type labels so that "À Louer" maps to "Rent", enabling separate rental price analysis; (2) applying a robust outlier removal strategy for price, as values like 6.3 trillion TND are clearly data entry errors that will destabilize any regression model. A log transformation of price is strongly recommended before model training, given the heavy right skew of the sale price distribution.</p>

````python
# Price distribution analysis
# ── Clean: keep only strictly positive, finite prices ──────────────────────
price_raw = pd.to_numeric(df["price"], errors="coerce")
price = price_raw[price_raw > 0].dropna()

price_sale = pd.to_numeric(
    df[df["transaction_type"] == "Sale"]["price"], errors="coerce"
)
price_sale = price_sale[price_sale > 0].dropna()

price_rent = pd.to_numeric(
    df[df["transaction_type"] == "Rent"]["price"], errors="coerce"
)
price_rent = price_rent[price_rent > 0].dropna()

print(f"Valid prices (> 0): {len(price):,}  |  "
      f"Sale: {len(price_sale):,}  |  Rent: {len(price_rent):,}")
print(f"Removed {len(price_raw.dropna()) - len(price):,} zero/negative/non-numeric prices\n")

# Cap outliers at 99th percentile for visualization
p99_sale = price_sale.quantile(0.99)
p99_rent = price_rent.quantile(0.99)
p99_all  = price.quantile(0.99)

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Overall price histogram (log scale) — safe because price > 0
axes[0, 0].hist(np.log1p(price[price <= p99_all]), bins=60,
                color=PALETTE[0], edgecolor="white")
axes[0, 0].set_title("Price Distribution (log scale, capped at 99th pct)")
axes[0, 0].set_xlabel("log(price + 1)")
axes[0, 0].set_ylabel("Count")

# Sale prices (capped)
axes[0, 1].hist(price_sale[price_sale <= p99_sale], bins=60,
                color=PALETTE[0], edgecolor="white")
axes[0, 1].set_title(f"Sale Price Distribution (\u2264 99th pct: {p99_sale:,.0f} TND)")
axes[0, 1].set_xlabel("Price (TND)")
axes[0, 1].xaxis.set_major_formatter(
    mtick.FuncFormatter(lambda x, _: f"{x/1e6:.1f}M" if x >= 1e6 else f"{x/1e3:.0f}K"))

# Rent prices (capped)
axes[1, 0].hist(price_rent[price_rent <= p99_rent], bins=60,
                color=PALETTE[1], edgecolor="white")
axes[1, 0].set_title(f"Rent Price Distribution (\u2264 99th pct: {p99_rent:,.0f} TND)")
axes[1, 0].set_xlabel("Price (TND)")

# Box plot by transaction type (capped)
df_box = df.copy()
df_box["price"] = pd.to_numeric(df_box["price"], errors="coerce")
df_box = df_box[(df_box["price"] > 0) & (df_box["price"] <= p99_all)]
df_box.boxplot(column="price", by="transaction_type", ax=axes[1, 1],
               boxprops=dict(color=PALETTE[0]),
               medianprops=dict(color="red", linewidth=2))
axes[1, 1].set_title("Price by Transaction Type")
axes[1, 1].set_xlabel("Transaction Type")
axes[1, 1].set_ylabel("Price (TND)")
plt.suptitle("")

plt.tight_layout()
plt.savefig("Data/eda_price_distribution.png", bbox_inches="tight")
plt.show()

print("Price summary (all valid):")
print(price.describe().apply(lambda x: f"{x:,.0f}"))
print("\nSale price summary:")
print(price_sale.describe().apply(lambda x: f"{x:,.0f}"))
print("\nRent price summary:")
print(price_rent.describe().apply(lambda x: f"{x:,.0f}"))
````

<h2 style="color: #87BAC3;">⚠️ Outlier Detection — IQR Method on Price</h2>

<h3>🎯 What This Cell Does</h3>
<p>This cell applies the standard Interquartile Range (IQR) method to formally identify price outliers — computing the lower and upper fences and flagging listings that fall outside them. The top 10 highest-priced listings are also printed for manual inspection.</p>

<h3>📊 Results & Observations</h3>
<p>The IQR method defines valid price bounds as <strong>[−785,000 TND, 1,335,000 TND]</strong>, flagging <strong>3,582 listings (6.0%)</strong> as outliers. Manual inspection of the top 10 highest prices reveals clearly corrupt entries: a Sousse immeuble listed at <strong>6.3 trillion TND</strong>, La Marsa villas at <strong>1.4 trillion</strong> and <strong>1.28 trillion TND</strong>, and a Nabeul terrain at <strong>722 billion TND</strong>. These are almost certainly data entry errors where prices were entered in millimes instead of dinars, or digits were duplicated during scraping. Notably, all extreme outliers originate from the <strong>Mubawab</strong> source — suggesting either a scraping artifact or a listing management issue on that platform.</p>

<h3>💡 Key Takeaways</h3>
<p>The 6% outlier rate by IQR is manageable but consequential. For modeling purposes, a price ceiling of approximately 1.5–2 million TND should be applied as a pre-processing rule, or a log transformation combined with a percentile cap (e.g., 99th percentile capping at the preprocessing stage) should be enforced. The concentration of extreme outliers in Mubawab data points to a platform-specific data quality issue that should be flagged and monitored in the data pipeline.</p>

````python
# Outlier detection — IQR method on price
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["price"] < lower) | (df["price"] > upper)]
print(f"IQR bounds: [{lower:,.0f}, {upper:,.0f}]")
print(f"Outliers detected: {len(outliers):,} ({len(outliers)/len(df)*100:.1f}%)")
print(f"\nTop 10 highest prices:")
print(df.nlargest(10, "price")[["source", "transaction_type", "property_type",
                                  "governorate", "city", "price", "surface_m2"]].to_string(index=False))
````

<h2 style="color: #87BAC3;">🔁 Duplicate Detection</h2>

<h3>🎯 What This Cell Does</h3>
<p>This cell performs a two-level duplicate analysis: exact row-level duplicates and near-duplicates defined as listings sharing the same title, price, surface, and governorate — a proxy for reposted or cross-platform duplicate listings.</p>

<h3>📊 Results & Observations</h3>
<p>There are <strong>zero exact row-level duplicates</strong> — confirming that the <code>ignore_index=True</code> concatenation and the id assignment during standardization were correctly applied. However, the near-duplicate analysis reveals a significant problem: <strong>12,990 listings</strong> share the same combination of title + price + surface + governorate, representing approximately <strong>21.8%</strong> of the total dataset. Additionally, the dataset contains only <strong>48,648 unique <code>listing_id</code> values</strong> out of 59,493 total rows — a gap of <strong>10,845 rows</strong> — indicating that the same listing was captured from multiple platforms or across multiple scraping batches. This is consistent with the known behavior of Tayara users who repost listings regularly.</p>

<h3>💡 Key Takeaways</h3>
<p>Near-duplicate deduplication is a mandatory preprocessing step before any statistical analysis or model training. Including 12,990 near-duplicate listings would artificially inflate the frequency of popular listing types, bias price distribution estimates, and degrade model generalization. A deduplication strategy based on a combination of listing URL, title similarity, price, and geography is recommended — with the platform-source label used as a tiebreaker to decide which copy to retain.</p>

````python
# Duplicate detection
exact_dupes = df.duplicated().sum()
title_dupes = df.duplicated(subset=["title", "price", "surface_m2", "governorate"]).sum() if "title" in df.columns else "N/A"

print(f"Exact duplicate rows: {exact_dupes:,}")
print(f"Near-duplicates (title+price+surface+gov): {title_dupes}")
print(f"\nTotal rows: {len(df):,}")
print(f"Unique listing_id values: {df['listing_id'].nunique():,}" if "listing_id" in df.columns else "listing_id not present")

# Per-source duplicate rate
for src in df["source"].unique():
    sub = df[df["source"] == src]
    d = sub.duplicated().sum()
    print(f"  {src}: {d:,} exact dupes ({d/len(sub)*100:.1f}%)")
````

<h2 style="color: #87BAC3;">📁 Image Inventory — Coverage & Volume</h2>

<h3>🎯 Purpose</h3>
<p>This cell scans the entire image directory tree to build a comprehensive inventory of all available property images. It maps each listing folder to its image count, computes overall coverage relative to the merged listings dataset, and flags listings with no images at all.</p>

<h3>🔍 Technical Breakdown</h3>
<p><code>os.walk()</code> traverses the folder hierarchy recursively. For each subfolder named after a listing ID, all files with image extensions (<code>.jpg</code>, <code>.jpeg</code>, <code>.png</code>, <code>.webp</code>) are counted. The result is joined against the merged listings DataFrame to compute image coverage per listing and identify zero-image listings. Summary statistics are printed for immediate situational awareness.</p>

<h3>⚠️ Why This Matters</h3>
<p>Image coverage is a critical quality dimension for multi-modal valuation models. Listings without images are much less informative — both for computer vision features and for user trust signals. Understanding coverage gaps early allows the modeling pipeline to design appropriate fallback strategies for image-absent records.</p>

````python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
from pathlib import Path
from collections import Counter
import warnings
warnings.filterwarnings("ignore")

PALETTE = ["#426d84", "#5c8399", "#7b98a6", "#87BAC3", "#d9e6ec", "#2c3e50"]
plt.rcParams.update({"figure.dpi": 130, "axes.titlesize": 13, "axes.labelsize": 11})

IMAGE_ROOT = Path("Data/raw_listings/Img/images")
IMG_EXTS   = {".jpg", ".jpeg", ".png", ".webp", ".JPG", ".JPEG", ".PNG"}

# ── Build inventory ────────────────────────────────────────────────────────
inventory = {}
for folder in IMAGE_ROOT.iterdir():
    if folder.is_dir():
        imgs = [f for f in folder.iterdir() if f.suffix in IMG_EXTS]
        try:
            listing_id = int(folder.name)
            inventory[listing_id] = imgs
        except ValueError:
            pass

img_counts = {lid: len(imgs) for lid, imgs in inventory.items()}
img_series = pd.Series(img_counts, name="image_count")

total_folders  = len(inventory)
total_images   = sum(img_counts.values())
zero_img       = sum(1 for v in img_counts.values() if v == 0)
has_img        = total_folders - zero_img
avg_per_listing = total_images / has_img if has_img > 0 else 0

print(f"Listing folders found : {total_folders:,}")
print(f"Total images          : {total_images:,}")
print(f"Listings with images  : {has_img:,}")
print(f"Listings with 0 images: {zero_img:,}")
print(f"Avg images per listing: {avg_per_listing:.2f}")
print(f"\nImage count distribution:")
print(img_series.describe())
````

<h2 style="color: #87BAC3;">📊 Image Coverage vs Merged Listings Dataset</h2>

<h3>🎯 Purpose</h3>
<p>This cell joins the image inventory against the full merged listings DataFrame to compute the overall image coverage rate — i.e., what percentage of the 59,493 merged listings have at least one image in the folder. It also breaks coverage down by data source to identify which platforms contributed images.</p>

<h3>🔍 Technical Breakdown</h3>
<p>The merged dataset is loaded and its <code>id</code> column is cross-referenced against the image inventory dictionary. A new binary column <code>has_images</code> is derived, and coverage statistics are computed globally and grouped by <code>source</code>. This reveals whether image availability is evenly distributed or concentrated in specific platforms.</p>

<h3>⚠️ Why This Matters</h3>
<p>Multi-modal models that fuse image features with structured attributes will only work for listings with images. Knowing coverage by source allows the modeling team to set realistic expectations about which platform's listings can benefit from visual feature extraction, and to design separate model branches for image-present vs. image-absent listings.</p>

````python
# Load merged listings to compute coverage
df = pd.read_csv("Data/raw_listings/merged_listings.csv", low_memory=False)

df["has_images"]   = df["id"].isin(img_counts.keys())
df["image_count"]  = df["id"].map(img_counts).fillna(0).astype(int)

# Overall coverage
total_listings   = len(df)
covered          = df["has_images"].sum()
coverage_pct     = covered / total_listings * 100

print(f"Total listings in merged dataset : {total_listings:,}")
print(f"Listings with image folder       : {covered:,} ({coverage_pct:.1f}%)")
print(f"Listings without image folder    : {total_listings - covered:,} ({100-coverage_pct:.1f}%)")

# Coverage by source
cov_by_source = df.groupby("source").agg(
    total=("id", "count"),
    with_images=("has_images", "sum"),
    avg_images=("image_count", "mean")
).reset_index()
cov_by_source["coverage_pct"] = (cov_by_source["with_images"] / cov_by_source["total"] * 100).round(1)
cov_by_source = cov_by_source.sort_values("with_images", ascending=False)
print("\nCoverage by source:")
print(cov_by_source.to_string(index=False))
````

<h2 style="color: #87BAC3;">📈 Images Per Listing — Distribution Analysis</h2>

<h3>🎯 Purpose</h3>
<p>This cell analyzes how many images are available per listing — a key quality signal. A listing with 10 high-resolution images provides far more visual information than one with a single low-quality photo. The distribution reveals whether multi-image listings are the norm or the exception.</p>

<h3>🔍 Technical Breakdown</h3>
<p>The image count per listing is plotted as a histogram (capped at the 99th percentile to avoid distortion from rare extreme values) and as a cumulative distribution function. A bar chart breaking down image counts by source reveals platform-specific image upload behaviors. Summary statistics are printed for precise characterization of the distribution.</p>

<h3>⚠️ Why This Matters</h3>
<p>For computer vision pipelines, the number of available images per listing determines what strategies are feasible: single-image models for low-coverage listings, multi-image aggregation for well-photographed properties, or image selection heuristics (e.g., always use the first image as the primary). This analysis directly informs those architectural decisions.</p>

````python
# Images per listing distribution
img_count_series = df[df["image_count"] > 0]["image_count"]
p99 = img_count_series.quantile(0.99)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Histogram capped at 99th pct
axes[0].hist(img_count_series[img_count_series <= p99], bins=40,
             color=PALETTE[0], edgecolor="white")
axes[0].set_title(f"Images per Listing (capped at p99={p99:.0f})")
axes[0].set_xlabel("Number of Images")
axes[0].set_ylabel("Number of Listings")

# CDF
sorted_counts = np.sort(img_count_series.values)
cdf = np.arange(1, len(sorted_counts) + 1) / len(sorted_counts)
axes[1].plot(sorted_counts, cdf, color=PALETTE[0], linewidth=2)
axes[1].axvline(img_count_series.median(), color="red", linestyle="--",
                label=f"Median = {img_count_series.median():.0f}")
axes[1].axvline(img_count_series.mean(), color=PALETTE[2], linestyle="--",
                label=f"Mean = {img_count_series.mean():.1f}")
axes[1].set_title("CDF — Images per Listing")
axes[1].set_xlabel("Image Count")
axes[1].set_ylabel("Cumulative Fraction")
axes[1].legend()

# Average images by source
avg_by_src = df[df["image_count"] > 0].groupby("source")["image_count"].mean().sort_values(ascending=False).head(12)
axes[2].barh(avg_by_src.index[::-1], avg_by_src.values[::-1], color=PALETTE[1])
axes[2].set_title("Avg Images per Listing by Source")
axes[2].set_xlabel("Average Image Count")

plt.tight_layout()
plt.savefig("Data/eda_images_per_listing.png", bbox_inches="tight")
plt.show()

print("Images per listing summary (listings with ≥1 image):")
print(img_count_series.describe().apply(lambda x: f"{x:,.2f}"))
````

<h2 style="color: #87BAC3;">💾 File Format & Size Analysis</h2>

<h3>🎯 Purpose</h3>
<p>This cell samples the image file inventory to analyze the distribution of file formats (JPEG, PNG, WebP) and file sizes in kilobytes. File size is a proxy for image quality — very small files often indicate low-resolution thumbnails rather than usable property photographs.</p>

<h3>🔍 Technical Breakdown</h3>
<p>A random sample of up to 5,000 image files is drawn from the full inventory. For each sampled file, the extension (format) and size in KB are recorded. Format counts are visualized as a pie chart, and file sizes are plotted as a histogram with percentile markers. Files below a minimum viable size threshold (e.g., 10 KB) are flagged as likely thumbnails or corrupt files.</p>

<h3>⚠️ Why This Matters</h3>
<p>Computer vision models require images above a minimum resolution and file size to extract meaningful features. Thumbnails below ~10–15 KB are typically too small for CNN or ViT-based feature extraction. Identifying the proportion of such files early informs the image preprocessing and filtering strategy that must precede any model training.</p>

````python
import random

# Sample up to 5000 image files for format/size analysis
all_image_files = []
for lid, imgs in inventory.items():
    all_image_files.extend(imgs)

sample_files = random.sample(all_image_files, min(5000, len(all_image_files)))

formats    = []
file_sizes = []

for fpath in sample_files:
    formats.append(fpath.suffix.lower())
    try:
        file_sizes.append(fpath.stat().st_size / 1024)  # KB
    except:
        file_sizes.append(0)

fmt_counts = Counter(formats)
file_sizes = np.array(file_sizes)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Format pie
labels = [f"{k} ({v})" for k, v in fmt_counts.most_common()]
axes[0].pie([v for _, v in fmt_counts.most_common()],
            labels=labels, colors=PALETTE[:len(fmt_counts)],
            autopct="%1.1f%%", startangle=140,
            wedgeprops={"edgecolor": "white", "linewidth": 1.5})
axes[0].set_title("Image Format Distribution")

# File size histogram
p99_size = np.percentile(file_sizes[file_sizes > 0], 99)
axes[1].hist(file_sizes[(file_sizes > 0) & (file_sizes <= p99_size)],
             bins=50, color=PALETTE[0], edgecolor="white")
axes[1].axvline(10, color="red", linestyle="--", label="10 KB threshold")
axes[1].axvline(np.median(file_sizes[file_sizes > 0]), color=PALETTE[2],
                linestyle="--", label=f"Median = {np.median(file_sizes[file_sizes>0]):.0f} KB")
axes[1].set_title("File Size Distribution (KB, capped at p99)")
axes[1].set_xlabel("File Size (KB)")
axes[1].set_ylabel("Count")
axes[1].legend()

# Cumulative size — how many images are above threshold
thresholds = np.arange(0, 200, 5)
pct_above  = [np.mean(file_sizes >= t) * 100 for t in thresholds]
axes[2].plot(thresholds, pct_above, color=PALETTE[0], linewidth=2)
axes[2].axvline(10, color="red", linestyle="--", alpha=0.7, label="10 KB")
axes[2].axvline(50, color=PALETTE[2], linestyle="--", alpha=0.7, label="50 KB")
axes[2].set_title("% Images Above File Size Threshold")
axes[2].set_xlabel("Min File Size (KB)")
axes[2].set_ylabel("% of Images")
axes[2].legend()

plt.tight_layout()
plt.savefig("Data/eda_image_formats_sizes.png", bbox_inches="tight")
plt.show()

tiny = np.sum(file_sizes < 10)
print(f"Format breakdown: {dict(fmt_counts.most_common())}")
print(f"\nFile size summary (KB):")
print(f"  Min    : {file_sizes[file_sizes>0].min():.1f}")
print(f"  Median : {np.median(file_sizes[file_sizes>0]):.1f}")
print(f"  Mean   : {np.mean(file_sizes[file_sizes>0]):.1f}")
print(f"  p95    : {np.percentile(file_sizes[file_sizes>0], 95):.1f}")
print(f"  Max    : {file_sizes.max():.1f}")
print(f"\nImages < 10 KB (likely thumbnails): {tiny:,} ({tiny/len(sample_files)*100:.1f}%)")
````

<h2 style="color: #87BAC3;">🔍 Image Resolution Analysis</h2>

<h3>🎯 Purpose</h3>
<p>This cell reads a random sample of images to extract their actual pixel dimensions (width × height), analyzes the resolution distribution, computes aspect ratio statistics, and identifies images below the minimum resolution threshold for reliable computer vision feature extraction.</p>

<h3>🔍 Technical Breakdown</h3>
<p>Pillow (<code>PIL.Image</code>) is used to read image metadata without loading the full pixel data (using <code>Image.open()</code> which is lazy). Width, height, total pixel count, and aspect ratio are extracted for each sampled image. Results are visualized as scatter plots (width vs. height), resolution histograms, and aspect ratio distributions. Images below 224×224 pixels — the standard minimum for most CNN architectures — are flagged.</p>

<h3>⚠️ Why This Matters</h3>
<p>Resolution is the most direct quality gating criterion for computer vision. Images below 224×224 cannot be fed into standard ResNet, EfficientNet, or ViT architectures without aggressive upscaling that degrades feature quality. Knowing the resolution distribution determines whether images need resizing, padding, or filtering before model inference, and informs the choice of input resolution for training.</p>

````python
from PIL import Image as PILImage

# Sample up to 2000 images for resolution analysis (reading metadata is slow)
sample_res = random.sample(all_image_files, min(2000, len(all_image_files)))

widths, heights, aspects = [], [], []
failed = 0

for fpath in sample_res:
    try:
        with PILImage.open(fpath) as im:
            w, h = im.size
            widths.append(w)
            heights.append(h)
            aspects.append(w / h if h > 0 else 0)
    except Exception:
        failed += 1

widths  = np.array(widths)
heights = np.array(heights)
aspects = np.array(aspects)
pixels  = widths * heights

print(f"Successfully read: {len(widths):,} images | Failed: {failed}")

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Width vs Height scatter
axes[0, 0].scatter(widths, heights, alpha=0.15, s=8, color=PALETTE[0])
axes[0, 0].axhline(224, color="red", linestyle="--", linewidth=1, label="224px")
axes[0, 0].axvline(224, color="red", linestyle="--", linewidth=1)
axes[0, 0].set_title("Width vs Height (pixels)")
axes[0, 0].set_xlabel("Width (px)")
axes[0, 0].set_ylabel("Height (px)")
axes[0, 0].legend()

# Width distribution
axes[0, 1].hist(widths[widths <= np.percentile(widths, 99)], bins=50,
                color=PALETTE[0], edgecolor="white")
axes[0, 1].axvline(224, color="red", linestyle="--", label="224px min")
axes[0, 1].axvline(np.median(widths), color=PALETTE[2], linestyle="--",
                   label=f"Median={np.median(widths):.0f}px")
axes[0, 1].set_title("Width Distribution")
axes[0, 1].set_xlabel("Width (px)")
axes[0, 1].legend()

# Height distribution
axes[1, 0].hist(heights[heights <= np.percentile(heights, 99)], bins=50,
                color=PALETTE[1], edgecolor="white")
axes[1, 0].axvline(224, color="red", linestyle="--", label="224px min")
axes[1, 0].axvline(np.median(heights), color=PALETTE[2], linestyle="--",
                   label=f"Median={np.median(heights):.0f}px")
axes[1, 0].set_title("Height Distribution")
axes[1, 0].set_xlabel("Height (px)")
axes[1, 0].legend()

# Aspect ratio distribution
axes[1, 1].hist(aspects[(aspects > 0) & (aspects <= np.percentile(aspects, 99))],
                bins=50, color=PALETTE[2], edgecolor="white")
axes[1, 1].axvline(1.0, color="red", linestyle="--", label="1:1 (square)")
axes[1, 1].axvline(4/3, color=PALETTE[0], linestyle="--", label="4:3")
axes[1, 1].axvline(16/9, color=PALETTE[4], linestyle="--", label="16:9")
axes[1, 1].set_title("Aspect Ratio Distribution")
axes[1, 1].set_xlabel("Width / Height")
axes[1, 1].legend()

plt.tight_layout()
plt.savefig("Data/eda_image_resolution.png", bbox_inches="tight")
plt.show()

below_224 = np.sum((widths < 224) | (heights < 224))
print(f"\nResolution summary:")
print(f"  Median width  : {np.median(widths):.0f} px")
print(f"  Median height : {np.median(heights):.0f} px")
print(f"  Median aspect : {np.median(aspects):.3f}")
print(f"  Images < 224px on either dimension: {below_224:,} ({below_224/len(widths)*100:.1f}%)")
print(f"  Mean megapixels: {np.mean(pixels)/1e6:.2f} MP")
````

<h2 style="color: #87BAC3;">🎨 Color Statistics & Brightness Analysis</h2>

<h3>🎯 Purpose</h3>
<p>This cell analyzes the color and brightness characteristics of the image corpus — computing per-image mean brightness, saturation, and channel statistics. These metrics serve as proxies for image quality and listing professionalism: well-lit, properly exposed images tend to correlate with higher-quality listings and better user engagement.</p>

<h3>🔍 Technical Breakdown</h3>
<p>Each sampled image is resized to a small thumbnail (64×64) for fast processing, then converted to HSV (Hue-Saturation-Value) color space. The Value channel gives brightness, and the Saturation channel measures color intensity. Mean and standard deviation of brightness are computed per image. Very dark images (mean brightness < 50/255) and very bright/overexposed images (mean brightness > 220/255) are flagged as potentially low quality. Channel-level RGB statistics are also computed to detect grayscale or near-grayscale images.</p>

<h3>⚠️ Why This Matters</h3>
<p>Brightness and color diversity are informative quality signals for real estate imagery. Extremely dark or washed-out images are unlikely to provide useful visual features for property condition assessment or luxury detection. A computer vision preprocessing pipeline should filter or flag such images before training to avoid introducing noise into learned visual representations.</p>

````python
import cv2

# Sample up to 1500 images for color/brightness analysis
sample_color = random.sample(all_image_files, min(1500, len(all_image_files)))

brightness_vals = []
saturation_vals = []
r_means, g_means, b_means = [], [], []
grayscale_count = 0
dark_count      = 0
bright_count    = 0
failed_c        = 0

for fpath in sample_color:
    try:
        img = cv2.imread(str(fpath))
        if img is None:
            failed_c += 1
            continue
        # Resize to 64x64 for speed
        thumb = cv2.resize(img, (64, 64))

        # BGR → RGB
        rgb = cv2.cvtColor(thumb, cv2.COLOR_BGR2RGB)
        r_means.append(rgb[:, :, 0].mean())
        g_means.append(rgb[:, :, 1].mean())
        b_means.append(rgb[:, :, 2].mean())

        # BGR → HSV for brightness & saturation
        hsv = cv2.cvtColor(thumb, cv2.COLOR_BGR2HSV)
        v   = hsv[:, :, 2].mean()   # brightness 0–255
        s   = hsv[:, :, 1].mean()   # saturation 0–255
        brightness_vals.append(v)
        saturation_vals.append(s)

        if v < 50:   dark_count   += 1
        if v > 220:  bright_count += 1
        if s < 20:   grayscale_count += 1

    except Exception:
        failed_c += 1

brightness_vals = np.array(brightness_vals)
saturation_vals = np.array(saturation_vals)
r_means = np.array(r_means)
g_means = np.array(g_means)
b_means = np.array(b_means)
n = len(brightness_vals)

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Brightness histogram
axes[0, 0].hist(brightness_vals, bins=50, color=PALETTE[0], edgecolor="white")
axes[0, 0].axvline(50,  color="navy",  linestyle="--", label="Dark threshold (50)")
axes[0, 0].axvline(220, color="orange", linestyle="--", label="Overexposed (220)")
axes[0, 0].set_title("Image Brightness Distribution (HSV Value channel)")
axes[0, 0].set_xlabel("Mean Brightness (0–255)")
axes[0, 0].set_ylabel("Count")
axes[0, 0].legend()

# Saturation histogram
axes[0, 1].hist(saturation_vals, bins=50, color=PALETTE[1], edgecolor="white")
axes[0, 1].axvline(20, color="red", linestyle="--", label="Near-grayscale (<20)")
axes[0, 1].set_title("Image Saturation Distribution")
axes[0, 1].set_xlabel("Mean Saturation (0–255)")
axes[0, 1].set_ylabel("Count")
axes[0, 1].legend()

# Brightness vs Saturation scatter
axes[1, 0].scatter(brightness_vals, saturation_vals, alpha=0.2, s=6, color=PALETTE[0])
axes[1, 0].axvline(50,  color="navy",   linestyle="--", alpha=0.6, label="Dark")
axes[1, 0].axvline(220, color="orange", linestyle="--", alpha=0.6, label="Overexposed")
axes[1, 0].axhline(20,  color="red",    linestyle="--", alpha=0.6, label="Grayscale")
axes[1, 0].set_title("Brightness vs Saturation")
axes[1, 0].set_xlabel("Brightness")
axes[1, 0].set_ylabel("Saturation")
axes[1, 0].legend(fontsize=8)

# Mean RGB channel values
channel_means = [r_means.mean(), g_means.mean(), b_means.mean()]
channel_stds  = [r_means.std(),  g_means.std(),  b_means.std()]
axes[1, 1].bar(["Red", "Green", "Blue"], channel_means,
               yerr=channel_stds, color=["#c0392b", "#27ae60", "#2980b9"],
               capsize=6, edgecolor="white")
axes[1, 1].set_title("Mean RGB Channel Values (± std)")
axes[1, 1].set_ylabel("Mean Pixel Value (0–255)")

plt.tight_layout()
plt.savefig("Data/eda_image_color.png", bbox_inches="tight")
plt.show()

print(f"Analyzed: {n:,} images | Failed: {failed_c}")
print(f"\nBrightness stats:")
print(f"  Mean   : {brightness_vals.mean():.1f}")
print(f"  Median : {np.median(brightness_vals):.1f}")
print(f"  Std    : {brightness_vals.std():.1f}")
print(f"\nQuality flags:")
print(f"  Dark images (V < 50)        : {dark_count:,}  ({dark_count/n*100:.1f}%)")
print(f"  Overexposed images (V > 220) : {bright_count:,}  ({bright_count/n*100:.1f}%)")
print(f"  Near-grayscale (S < 20)      : {grayscale_count:,}  ({grayscale_count/n*100:.1f}%)")
````

<h2 style="color: #87BAC3;">🖼️ Sample Image Grid — Visual Quality Inspection</h2>

<h3>🎯 Purpose</h3>
<p>This cell displays a 5×6 grid of randomly sampled listing images to provide a qualitative visual impression of the image corpus. Numerical statistics alone cannot capture whether images show interiors, exteriors, floor plans, or irrelevant content — a direct visual scan is irreplaceable for this assessment.</p>

<h3>🔍 Technical Breakdown</h3>
<p>30 images are randomly sampled across different listings (one per listing to maximize diversity) and rendered in a grid using Matplotlib. Each image is displayed without axis labels, with the listing ID shown as the subplot title. The sample is drawn from listings across all available sources to avoid source-specific biases in the visual impression.</p>

<h3>⚠️ Why This Matters</h3>
<p>A qualitative image scan is essential for detecting common real-world data quality issues that numerical metrics miss: images showing watermarks or platform logos, floor plan diagrams mixed with photographs, exterior shots vs. interior shots imbalance, images of documents (e.g., land titles), or images that are completely irrelevant to the property (e.g., seller profile photos). These insights feed directly into image preprocessing and filtering rules.</p>

<h2 style="color: #87BAC3;">📚 Vocabulary Richness by Source</h2>

<h3>🎯 What This Cell Does</h3>
<p>This cell computes three vocabulary richness metrics for each source: total word count, unique word count, and the Type-Token Ratio (TTR) — a normalized measure of lexical diversity calculated as the ratio of unique words to total words. A higher TTR indicates more diverse vocabulary relative to corpus size.</p>

<h3>📊 Results & Observations</h3>
<p><strong>Mubawab Blog</strong> has the largest raw corpus: <strong>164,044 total words</strong> and <strong>14,490 unique words</strong>, but a relatively low TTR of <strong>0.088</strong> — meaning only 8.8% of all Mubawab words are unique. This low ratio is expected for a large corpus (larger corpora always have lower TTRs) but also reflects Mubawab's repetitive editorial style centered on market price reporting with recurring terminology. <strong>BnB Blog</strong> shows a much higher TTR of <strong>0.129</strong> (12.9%) across 48,013 total words — indicating more lexically diverse, long-form content covering a broader range of real estate topics. <strong>Houni Blog</strong> has the highest TTR at <strong>0.182</strong> (18.2%) despite having the smallest corpus (8,854 words), consistent with its consumer-oriented content covering diverse lifestyle and practical topics.</p>

<h3>💡 Key Takeaways</h3>
<p>Vocabulary richness analysis confirms that BnB and Houni articles, despite their smaller volume, contribute disproportionately diverse vocabulary to the corpus — which is valuable for training a broad-coverage domain embedding model. Mubawab's lower TTR is not a quality problem but a reflection of its focused, data-centric editorial style. For building a real estate vocabulary or fine-tuning a domain language model, a sampling strategy that over-represents BnB and Houni content relative to their raw frequency would produce a more lexically balanced training corpus.</p>

````python
# Vocabulary richness by source
def vocab_stats(text_series):
    all_words = []
    for t in text_series.fillna(""):
        all_words.extend(re.findall(r'\b[a-zéèêëàâùûüôîïç]{3,}\b', t.lower()))
    total = len(all_words)
    unique = len(set(all_words))
    return pd.Series({"total_words": total, "unique_words": unique,
                      "ttr": round(unique / total, 4) if total > 0 else 0})

vocab_df = art.groupby("source")["content"].apply(vocab_stats).reset_index()
vocab_df.columns = ["source", "metric", "value"]
vocab_pivot = vocab_df.pivot(index="source", columns="metric", values="value")
print("Vocabulary richness by source:")
print(vocab_pivot.to_string())

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
metrics = ["total_words", "unique_words", "ttr"]
titles  = ["Total Words", "Unique Words", "Type-Token Ratio (TTR)"]
for ax, metric, title in zip(axes, metrics, titles):
    vals = vocab_pivot[metric]
    ax.bar(vals.index, vals.values, color=PALETTE[:len(vals)])
    ax.set_title(title)
    ax.set_ylabel(metric)
    for i, v in enumerate(vals.values):
        ax.text(i, v * 1.01, f"{v:,.3f}" if metric == "ttr" else f"{v:,.0f}",
                ha="center", fontsize=9)

plt.tight_layout()
plt.savefig("Data/eda_vocab_richness.png", bbox_inches="tight")
plt.show()
````

These are the same platform and must be unified to a single canonical lowercase label so that any groupby, filter, or model that uses source as a feature treats them as one category.</p>

<h3>🔍 Technical Breakdown</h3>
<p>All source values are lowercased using <code>.str.lower().str.strip()</code>. This collapses all case variants in a single operation without needing to enumerate each platform explicitly — making it robust to any other mixed-case variants that may exist in smaller sources.</p>

````python
# Normalize source labels to lowercase
df["source"] = df["source"].str.lower().str.strip()

print("source values after normalization:")
print(df["source"].value_counts().to_string())
print(f"\nUnique sources: {df['source'].nunique()}")
````

<h2 style="color: #87BAC3;">🏷️ Fix 2 — Standardize Transaction Type</h2>

<h3>🎯 Purpose</h3>
<p>The <code>transaction_type</code> column contains the same concepts under multiple labels: <code>"Sale"</code>, <code>"À Vendre"</code>, and <code>"sale"</code> all mean a sale transaction, while <code>"À Louer"</code> and <code>"rent"</code> both mean a rental. These must be unified to two canonical English values — <code>"Sale"</code> and <code>"Rent"</code> — so that downstream models and analyses see a clean binary categorical variable.</p>

<h3>🔍 Technical Breakdown</h3>
<p>A mapping dictionary translates every known variant to its canonical form. <code>.map()</code> applies the mapping, and <code>.fillna(df["transaction_type"])</code> preserves any value not explicitly listed in the map (preventing accidental nullification of edge cases). ````python
# Standardize transaction_type to canonical English values
tx_map = {
    "Sale"      : "Sale",
    "sale"      : "Sale",
    "À Vendre"  : "Sale",
    "à vendre"  : "Sale",
    "Vente"     : "Sale",
    "vente"     : "Sale",
    "rent"      : "Rent",
    "Rent"      : "Rent",
    "À Louer"   : "Rent",
    "à louer"   : "Rent",
    "Location"  : "Rent",
    "location"  : "Rent",
}

df["transaction_type"] = (
    df["transaction_type"]
    .map(tx_map)
    .fillna(df["transaction_type"])
)

print("transaction_type values after standardization:")
print(df["transaction_type"].value_counts(dropna=False).to_string())
````

<h2 style="color: #87BAC3;">🔁 Duplicate Verification Based on Description</h2>

<h3>🎯 Purpose</h3>
<p>Description text is the richest deduplication signal in the dataset — two listings sharing an identical description are almost certainly the same property reposted across platforms or across time. This cell identifies exact description duplicates, analyzes their distribution by source, and inspects sample pairs to decide whether to drop them.</p>

<h3>🔍 Technical Breakdown</h3>
<p>Exact description duplicates are detected with <code>duplicated(subset=["description"])</code>. The analysis reports total duplicate count, which sources contribute them most, and how many unique descriptions are shared by 2, 3, or more listings. A sample of duplicated description groups is displayed for manual inspection. The index is kept intact at this stage — deduplication is applied in the next cell after reviewing the results.</p>

````python
# ── Exact description duplicates ──────────────────────────────────────────
desc_dupes = df[df["description"].notna()].duplicated(subset=["description"], keep=False)
dupes_df   = df[df["description"].notna()][desc_dupes]

total_dupes    = df.duplicated(subset=["description"], keep=False).sum()
unique_duped   = df[df.duplicated(subset=["description"], keep=False)]["description"].nunique()
rows_to_remove = df.duplicated(subset=["description"], keep="first").sum()

print(f"Total rows with a duplicated description : {total_dupes:,}")
print(f"Unique descriptions that appear >1 time  : {unique_duped:,}")
print(f"Rows that would be dropped (keep first)  : {rows_to_remove:,}")
print(f"Rows that would remain                   : {len(df) - rows_to_remove:,}")

# How many listings share the same description
freq = dupes_df["description"].value_counts()
print(f"\nDuplicate group size distribution:")
print(freq.value_counts().sort_index().rename_axis("copies").rename("groups").to_string())

# By source
print(f"\nDuplicate rows by source:")
print(dupes_df["source"].value_counts().to_string())
````

````python
# Drop duplicates based on description — keep first occurrence
before = len(df)

df = df.drop_duplicates(subset=["description"], keep="first").reset_index(drop=True)

after = len(df)
print(f"Rows BEFORE : {before:,}")
print(f"Rows AFTER  : {after:,}")
print(f"Dropped     : {before - after:,} duplicate rows")
````

<h2 style="color: #87BAC3;">🗑️ Step 1 — Drop Useless Columns</h2>

<h3>🎯 Purpose</h3>

<h2 style="color: #87BAC3;">📝 Step 2 — Drop the Single Null Description Row</h2>

<h2 style="color: #87BAC3;">📊 Step 3 — Add Description Word Count Feature</h2>

<h3>🎯 Purpose</h3>
<p>Description length is a meaningful structured signal for the valuation model: longer, more detailed descriptions correlate with more effort from the seller and often with higher-value properties. Adding <code>description_word_count</code> now as a derived structured feature makes this signal available to the structured branch independently of the NLP branch.</p>

````python
# Add description word count as a structured feature
df["description_word_count"] = df["description"].str.split().str.len()

print("description_word_count summary:")
print(df["description_word_count"].describe().apply(lambda x: f"{x:,.1f}"))
````

<h2 style="color: #87BAC3;">🖼️ Step 4 — Image Availability Flag & Fill image_urls</h2>

<p>Since images form a dedicated model branch, the pipeline needs to know which listings have visual data available. A binary <code>has_images</code> flag is added as a structured feature. Null <code>image_urls</code> values are filled with an empty string so downstream image loading code does not need to handle NaN separately.</p>

````python
# Add has_images flag and fill null image_urls
df["has_images"] = df["image_count"] > 0
df["image_urls"] = df["image_urls"].fillna("")

print(f"Listings with images    : {df['has_images'].sum():,} ({df['has_images'].mean()*100:.1f}%)")
print(f"Listings without images : {(~df['has_images']).sum():,} ({(~df['has_images']).mean()*100:.1f}%)")
print(f"Null image_urls remaining: {df['image_urls'].isna().sum()}")
````

<h2 style="color: #87BAC3;">🔒 Step 5 — Fill Amenity Flag NaNs → False</h2>

<h3>🎯 Purpose</h3>
<p>The five amenity columns stored as <code>object</code> type (<code>has_elevator</code>, <code>has_heating</code>, <code>has_air_conditioning</code>, <code>has_parking</code>, <code>has_security</code>) have ~11.4% missing values because certain scraping sources simply did not capture these fields. The standard real estate imputation rule applies: <strong>if an amenity is not mentioned, it is assumed absent</strong> — sellers consistently highlight desirable features when present. NaNs are filled with <code>False</code> and all five columns are coerced to boolean type.</p>

````python
# Fill amenity flag NaNs with False and coerce to bool
amenity_obj_cols = [
    "has_elevator", "has_heating", "has_air_conditioning",
    "has_parking", "has_security"
]

for col in amenity_obj_cols:
    df[col] = df[col].fillna(False)
    df[col] = df[col].map(
        lambda x: False if str(x).strip().lower() in ("false", "0", "nan", "") else bool(x)
    )

print("Amenity flags — null counts after fill:")
for col in amenity_obj_cols:
    print(f"  {col:<25} nulls: {df[col].isna().sum()}  dtype: {df[col].dtype}")
````

<h2 style="color: #87BAC3;">👤 Step 6 — Encode seller_type as is_agency</h2>

<h3>🎯 Purpose</h3>
<p>Rather than imputing the raw <code>seller_type</code> string, it is more useful for the model to encode it as a binary feature: <code>is_agency = 1</code> if the listing was posted by an agency, <code>0</code> if by a private individual, and <code>-1</code> if unknown. This retains the missingness signal while providing a clean numeric input to the structured branch.</p>

````python
# Encode seller_type as binary is_agency feature
def encode_seller(val):
    if isinstance(val, str):
        v = val.strip().lower()
        if any(k in v for k in ["agence", "agency", "professionnel", "pro", "agent"]):
            return 1
        elif any(k in v for k in ["particulier", "individual", "private", "privé"]):
            return 0
    return -1  # unknown

df["is_agency"] = df["seller_type"].apply(encode_seller)

print("is_agency distribution:")
print(df["is_agency"].value_counts().rename({1: "Agency", 0: "Individual", -1: "Unknown"}).to_string())
````

<h2 style="color: #87BAC3;">🏠 Step 7 — Recover property_type from Title & Description</h2>

<h3>🎯 Purpose</h3>
<p>With 21.1% of <code>property_type</code> values missing, and given that property type is both a critical structured feature AND a conditioning signal for the image branch (the visual appearance of an apartment differs fundamentally from a farm or a villa), recovery is a priority. A keyword-matching approach scans the listing title and description for property-type indicator terms in French and English, and assigns the matched type. Rows where no keyword is found are labeled <code>"Unknown"</code>.</p>

````python
# Keyword-based property_type recovery from title + description
import re

TYPE_KEYWORDS = {
    "Apartment" : [
        "appartement", "appart", "apartment", "flat", "s\+1", "s\+2", "s\+3",
        "s\+4", "s\+5", "studio", "f2", "f3", "f4"
    ],
    "Villa"     : ["villa", "propriété", "propriete", "chalet"],
    "House"     : ["maison", "house", "dar", "riad"],
    "Land"      : ["terrain", "land", "lot", "parcelle", "foncier"],
    "Commercial": ["local commercial", "commerce", "fonds de commerce",
                   "boutique", "magasin", "local", "fond de commerce"],
    "Office"    : ["bureau", "office", "bureaux"],
    "Building"  : ["immeuble", "building", "résidence"],
    "Farm"      : ["ferme", "farm", "exploitation"],
    "Duplex"    : ["duplex"],
}

def infer_property_type(row):
    if pd.notna(row["property_type"]):
        return row["property_type"]
    text = " ".join([
        str(row.get("title", "") or ""),
        str(row.get("description", "") or "")
    ]).lower()
    for ptype, keywords in TYPE_KEYWORDS.items():
        for kw in keywords:
            if re.search(r'\b' + kw + r'\b', text):
                return ptype
    return "Unknown"

before_null = df["property_type"].isna().sum()
df["property_type"] = df.apply(infer_property_type, axis=1)
after_null  = df["property_type"].isna().sum()

print(f"property_type nulls BEFORE : {before_null:,}")
print(f"property_type nulls AFTER  : {after_null:,}")
print(f"Recovered                  : {before_null - after_null:,}")
print(f"\nproperty_type distribution:")
print(df["property_type"].value_counts().to_string())
````

<h2 style="color: #87BAC3;">🗺️ Step 8 — Recover City & Governorate from Title & Description</h2>

<h3>🎯 Purpose</h3>
<p>Location is the single strongest price driver in real estate valuation. For listings where both <code>city</code> and <code>governorate</code> are null, we scan the listing title and description for known Tunisian city and governorate names. Matches are used to fill the missing geographic fields. This regex-based approach is fast, deterministic, and recovers the majority of cases where the seller mentions a location in their listing text. Rows where no geographic signal is found are dropped — they would introduce significant noise into a location-aware valuation model.</p>

````python
# Regex-based city + governorate recovery from title + description
import re, pandas as pd

# Full city → governorate lookup (reuse the same map from Fix 4)
CITY_GOV = {
    "tunis": "Tunis", "la marsa": "Tunis", "carthage": "Tunis",
    "sidi bou said": "Tunis", "le bardo": "Tunis", "la goulette": "Tunis",
    "el menzah": "Tunis", "gammarth": "Tunis", "ain zaghouan": "Tunis",
    "la soukra": "Tunis", "chotrana": "Tunis", "el ghazela": "Tunis",
    "l'aouina": "Tunis", "el aouina": "Tunis", "jardins de carthage": "Tunis",
    "riadh andalous": "Tunis", "centre urbain nord": "Tunis",
    "cité ennasr": "Tunis", "ennasr": "Tunis", "el kram": "Tunis",
    "sidi hassine": "Tunis", "el ouardia": "Tunis", "bab souika": "Tunis",
    "montplaisir": "Tunis", "le lac": "Tunis", "berges du lac": "Tunis",
    "ariana": "Ariana", "raoued": "Ariana", "mnihla": "Ariana",
    "ettadhamen": "Ariana", "borj louzir": "Ariana",
    "ben arous": "Ben Arous", "rades": "Ben Arous", "radès": "Ben Arous",
    "hammam lif": "Ben Arous", "hammam chott": "Ben Arous",
    "boumhel": "Ben Arous", "el mourouj": "Ben Arous",
    "mégrine": "Ben Arous", "fouchana": "Ben Arous", "mornag": "Ben Arous",
    "ezzahra": "Ben Arous", "borj cedria": "Ben Arous",
    "manouba": "Manouba", "oued ellil": "Manouba", "douar hicher": "Manouba",
    "tebourba": "Manouba", "jedaida": "Manouba", "mornaguia": "Manouba",
    "nabeul": "Nabeul", "hammamet": "Nabeul", "kelibia": "Nabeul",
    "kélibia": "Nabeul", "grombalia": "Nabeul", "soliman": "Nabeul",
    "korba": "Nabeul", "dar chaabane": "Nabeul", "maamoura": "Nabeul",
    "tazarka": "Nabeul", "beni khiar": "Nabeul", "el haouaria": "Nabeul",
    "zaghouan": "Zaghouan", "el fahs": "Zaghouan",
    "bizerte": "Bizerte", "menzel bourguiba": "Bizerte", "raf raf": "Bizerte",
    "mateur": "Bizerte", "ras jebel": "Bizerte",
    "beja": "Béja", "béja": "Béja", "testour": "Béja", "medjez el bab": "Béja",
    "jendouba": "Jendouba", "tabarka": "Jendouba", "ain draham": "Jendouba",
    "le kef": "Le Kef", "kef": "Le Kef",
    "siliana": "Siliana",
    "sousse": "Sousse", "hammam sousse": "Sousse", "akouda": "Sousse",
    "sahloul": "Sousse", "el kantaoui": "Sousse", "hergla": "Sousse",
    "msaken": "Sousse", "kalaa kebira": "Sousse", "enfidha": "Sousse",
    "monastir": "Monastir", "moknine": "Monastir", "ksar hellal": "Monastir",
    "jemmal": "Monastir", "skanes": "Monastir",
    "mahdia": "Mahdia", "el jem": "Mahdia", "ksour essaf": "Mahdia",
    "salakta": "Mahdia", "rejiche": "Mahdia",
    "kairouan": "Kairouan",
    "kasserine": "Kasserine", "sbeitla": "Kasserine",
    "sidi bouzid": "Sidi Bouzid",
    "sfax": "Sfax", "sakiet ezzit": "Sfax", "gremda": "Sfax",
    "kerkennah": "Sfax", "mahrès": "Sfax", "tantana": "Sfax",
    "gafsa": "Gafsa", "metlaoui": "Gafsa", "redeyef": "Gafsa",
    "tozeur": "Tozeur", "nefta": "Tozeur",
    "kebili": "Kébili", "douz": "Kébili",
    "gabes": "Gabès", "gabès": "Gabès", "el hamma": "Gabès",
    "medenine": "Médenine", "médenine": "Médenine",
    "djerba": "Médenine", "zarzis": "Médenine", "houmt souk": "Médenine",
    "midoun": "Médenine", "aghir": "Médenine",
    "tataouine": "Tataouine",
}

# Sort by length descending so longer phrases match before substrings
SORTED_CITIES = sorted(CITY_GOV.keys(), key=len, reverse=True)

def extract_location(text):
    if not isinstance(text, str):
        return None, None
    t = text.lower()
    for city in SORTED_CITIES:
        if re.search(r'\b' + re.escape(city) + r'\b', t):
            return city.title(), CITY_GOV[city]
    return None, None

both_null_before = df[df["governorate"].isna() & df["city"].isna()].shape[0]
recovered_city, recovered_gov = 0, 0

for idx, row in df[df["governorate"].isna() & df["city"].isna()].iterrows():
    text = " ".join([str(row.get("title","") or ""), str(row.get("description","") or "")])
    city_found, gov_found = extract_location(text)
    if gov_found:
        df.at[idx, "governorate"] = gov_found
        recovered_gov += 1
    if city_found:
        df.at[idx, "city"] = city_found
        recovered_city += 1

both_null_after = df[df["governorate"].isna() & df["city"].isna()].shape[0]
print(f"Rows with both null BEFORE  : {both_null_before:,}")
print(f"Recovered governorate       : {recovered_gov:,}")
print(f"Recovered city              : {recovered_city:,}")
print(f"Rows with both null AFTER   : {both_null_after:,}")
````

````python
# Drop rows where both city and governorate remain null after all recovery attempts
before = len(df)
df = df[~(df["governorate"].isna() & df["city"].isna())].reset_index(drop=True)
print(f"Rows BEFORE : {before:,}")
print(f"Rows AFTER  : {len(df):,}")
print(f"Dropped     : {before - len(df):,} rows with no recoverable location")
````

<h2 style="color: #87BAC3;">💰 Step 9 — Price Imputation (3-Tier Median Strategy)</h2>

<h3>🎯 Purpose</h3>
<p>Price is the model target — null prices cannot be used for training. The 3-tier imputation strategy fills each missing price with the most granular available median: first by <strong>property_type + governorate</strong> (most precise), then by <strong>property_type alone</strong> if the combination has insufficient data, and finally by the <strong>global dataset median</strong> as a last resort. An <code>is_price_imputed</code> flag is added so the model can distinguish real from imputed targets during training.</p>

````python
# 3-tier price imputation
import numpy as np

df["price"] = pd.to_numeric(df["price"], errors="coerce")

# Pre-compute medians
median_by_type_gov  = df.groupby(["property_type", "governorate"])["price"].median()
median_by_type      = df.groupby("property_type")["price"].median()
global_median       = df["price"].median()

df["is_price_imputed"] = df["price"].isna()

def impute_price(row):
    if pd.notna(row["price"]):
        return row["price"]
    # Tier 1 — property_type + governorate
    key1 = (row["property_type"], row["governorate"])
    if key1 in median_by_type_gov.index and pd.notna(median_by_type_gov[key1]):
        return median_by_type_gov[key1]
    # Tier 2 — property_type only
    key2 = row["property_type"]
    if key2 in median_by_type.index and pd.notna(median_by_type[key2]):
        return median_by_type[key2]
    # Tier 3 — global median
    return global_median

df["price"] = df.apply(impute_price, axis=1)

tier1 = df[df["is_price_imputed"]].apply(
    lambda r: (r["property_type"], r["governorate"]) in median_by_type_gov.index, axis=1).sum()
print(f"Price nulls imputed total  : {df['is_price_imputed'].sum():,}")
print(f"  → Tier 1 (type+gov)      : {tier1:,}")
print(f"  → Tier 2+3 (type/global) : {df['is_price_imputed'].sum() - tier1:,}")
print(f"Price nulls remaining      : {df['price'].isna().sum()}")
print(f"\nGlobal median used         : {global_median:,.0f} TND")
````

<h2 style="color: #87BAC3;">📐 Step 10 — Surface Area Soft Imputation + Indicator Flag</h2>

<h3>🎯 Purpose</h3>
<p>Surface area is one of the top three price predictors in the structured branch. Rather than leaving 17.9% of values as NaN — which would prevent surface-based features like price-per-m² from being computed — a soft imputation fills missing values with the median surface by <strong>property_type + governorate</strong>, falling back to property_type median. A binary <code>has_surface</code> flag preserves the information about which values are real vs imputed, allowing the model to down-weight imputed surfaces during training.</p>

````python
# Surface area soft imputation + has_surface indicator
df["surface_m2"] = pd.to_numeric(df["surface_m2"], errors="coerce")
df["surface_m2"] = df["surface_m2"].where(df["surface_m2"] > 0)  # zero → NaN

df["has_surface"] = df["surface_m2"].notna()

median_surf_type_gov = df.groupby(["property_type", "governorate"])["surface_m2"].median()
median_surf_type     = df.groupby("property_type")["surface_m2"].median()
global_surf_median   = df["surface_m2"].median()

def impute_surface(row):
    if pd.notna(row["surface_m2"]):
        return row["surface_m2"]
    key1 = (row["property_type"], row["governorate"])
    if key1 in median_surf_type_gov.index and pd.notna(median_surf_type_gov[key1]):
        return median_surf_type_gov[key1]
    key2 = row["property_type"]
    if key2 in median_surf_type.index and pd.notna(median_surf_type[key2]):
        return median_surf_type[key2]
    return global_surf_median

before_null = df["surface_m2"].isna().sum()
df["surface_m2"] = df.apply(impute_surface, axis=1)

print(f"surface_m2 nulls BEFORE : {before_null:,}")
print(f"surface_m2 nulls AFTER  : {df['surface_m2'].isna().sum():,}")
print(f"has_surface=True        : {df['has_surface'].sum():,} ({df['has_surface'].mean()*100:.1f}%)")
print(f"has_surface=False       : {(~df['has_surface']).sum():,} (soft imputed)")
````

<h2 style="color: #87BAC3;">🛏️ Step 11 — Room Feature Indicator Flags</h2>

<h3>🎯 Purpose</h3>
<p>Rooms, bathrooms, and bedrooms are 71–85% missing — too sparse for reliable imputation. Rather than imputing potentially misleading values, binary indicator flags <code>has_rooms_info</code>, <code>has_bathrooms_info</code>, and <code>has_bedrooms_info</code> are added to signal availability to the model. The raw values are kept as-is with NaN — tree-based models (XGBoost, LightGBM) handle NaN natively and will extract signal where the data is available. <code>total_rooms</code> is dropped as it is 85.4% missing and redundant with <code>rooms</code>.</p>

````python
# Add indicator flags for sparse room features and drop total_rooms
df["rooms"]    = pd.to_numeric(df["rooms"],    errors="coerce")
df["bathrooms"]= pd.to_numeric(df["bathrooms"],errors="coerce")
df["bedrooms"] = pd.to_numeric(df["bedrooms"], errors="coerce")

df["has_rooms_info"]    = df["rooms"].notna()
df["has_bathrooms_info"]= df["bathrooms"].notna()
df["has_bedrooms_info"] = df["bedrooms"].notna()

df = df.drop(columns=["total_rooms"])

print("Indicator flags:")
for col in ["has_rooms_info", "has_bathrooms_info", "has_bedrooms_info"]:
    n = df[col].sum()
    print(f"  {col:<25} True: {n:,} ({n/len(df)*100:.1f}%)")
print(f"\ntotal_rooms dropped.")
print(f"Columns remaining: {df.shape[1]}")
````

<h2 style="color: #87BAC3;">💾 Step 12 — Final Missing Values Snapshot & Save</h2>

<h3>🎯 Purpose</h3>
<p>A final missing values audit confirms the effect of every treatment step applied in this section. The cleaned, model-ready dataset is saved as <strong>merged_listings_model_ready.csv</strong> — the canonical input for all feature engineering, training, and evaluation phases that follow.</p>

````python
# Final missing values snapshot
missing_final = df.isnull().mean().sort_values(ascending=False)
mf = missing_final.reset_index()
mf.columns = ["column", "missing_pct"]
mf["missing_count"] = (mf["missing_pct"] * len(df)).astype(int)
mf["pct_label"] = (mf["missing_pct"] * 100).round(1).astype(str) + "%"

print(f"Final dataset shape: {df.shape}")
print(f"\n{'Column':<28} {'Missing':>10} {'%':>8}")
print("-" * 48)
for _, row in mf.iterrows():
    marker = " ⚠️" if row["missing_pct"] > 0.5 else ""
    print(f"{row['column']:<28} {row['missing_count']:>10,} {row['pct_label']:>8}{marker}")

# Save
df.to_csv("Data/raw_listings/merged_listings_model_ready.csv", index=False)
print(f"\n✅ Saved → Data/raw_listings/merged_listings_model_ready.csv")
print(f"   Rows: {len(df):,}  |  Columns: {df.shape[1]}")
````

<h1 style="color: #2c3e50; font-size: 34px; font-weight: bold;">
    ⚙️ Feature Engineering
</h1>
<p style="font-size:15px;">
Feature engineering transforms the cleaned dataset into a set of structured numerical and categorical inputs optimised for the property valuation model. Every feature created here is derived from domain knowledge of the Tunisian real estate market and from the patterns surfaced during EDA. The engineering is organised in five groups: <strong>price & surface derived features</strong>, <strong>geographic encoding</strong>, <strong>property characteristics encoding</strong>, <strong>amenity composite scores</strong>, and <strong>text-derived signals</strong> from the listing description.
</p>

<h2 style="color: #87BAC3;">🔧 Setup — Load Model-Ready Dataset</h2>

<h3>🎯 Purpose</h3>
<p>Reload the cleaned model-ready CSV as the starting point for feature engineering. A copy is worked on throughout this section; the original file is not overwritten until the final save step.</p>

````python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("Data/raw_listings/merged_listings_model_ready.csv", low_memory=False)

# Drop the 248 remaining null governorate rows before feature engineering
before = len(df)
df = df[df["governorate"].notna()].reset_index(drop=True)
print(f"Rows after dropping null governorate: {len(df):,} (dropped {before - len(df)})")
print(f"Shape: {df.shape}")
df.head(3)
````

<h2 style="color: #87BAC3;">💰 Group 1 — Price & Surface Derived Features</h2>

<h3>🎯 Purpose</h3>
<p>Raw price and surface values are useful but several derived features extract additional signal that linear and tree-based models can exploit directly:</p>
<ul>
  <li><code>log_price</code> — log-transformed price, the standard target for real estate regression models (reduces skew, makes residuals more normal)</li>
  <li><code>price_per_m2</code> — price divided by surface area, only for listings with real (non-imputed) surface data</li>
  <li><code>log_price_per_m2</code> — log of price per m², the primary market density metric</li>
  <li><code>log_surface</code> — log-transformed surface area</li>
  <li><code>surface_bin</code> — discretised surface into 6 market-relevant size buckets</li>
</ul>

````python
# ── Log price (model target) ───────────────────────────────────────────────
df["log_price"] = np.log1p(df["price"])

# ── Log surface ────────────────────────────────────────────────────────────
df["log_surface"] = np.log1p(df["surface_m2"])

# ── Price per m² — only where surface is real (not imputed) ───────────────
df["price_per_m2"] = np.where(
    df["has_surface"],
    df["price"] / df["surface_m2"].replace(0, np.nan),
    np.nan
)
df["log_price_per_m2"] = np.log1p(df["price_per_m2"])

# ── Surface bins (m²) ──────────────────────────────────────────────────────
bins   = [0, 60, 100, 150, 250, 500, np.inf]
labels = ["micro", "small", "medium", "large", "xlarge", "estate"]
df["surface_bin"] = pd.cut(
    df["surface_m2"], bins=bins, labels=labels, right=False
).astype(str)

print("New features: log_price, log_surface, price_per_m2, log_price_per_m2, surface_bin")
print(f"\nlog_price  — mean: {df['log_price'].mean():.2f}, std: {df['log_price'].std():.2f}")
print(f"price_per_m2 — non-null: {df['price_per_m2'].notna().sum():,} ({df['price_per_m2'].notna().mean()*100:.1f}%)")
print(f"\nsurface_bin distribution:")
print(df["surface_bin"].value_counts().sort_index())
````

<h2 style="color: #87BAC3;">🗺️ Group 2 — Geographic Features</h2>

<h3>🎯 Purpose</h3>
<p>Location is the strongest price predictor in real estate. Three complementary geographic encodings are created:</p>
<ul>
  <li><code>gov_median_price</code> — median sale price per governorate (target encoding), capturing the overall price level of each region</li>
  <li><code>gov_median_price_per_m2</code> — median price/m² per governorate, a normalised density metric</li>
  <li><code>gov_listing_count</code> — number of listings per governorate, a market activity proxy</li>
  <li><code>is_greater_tunis</code> — binary flag for the Greater Tunis area (Tunis, Ariana, Ben Arous, Manouba), the premium market cluster</li>
  <li><code>is_coastal</code> — binary flag for coastal governorates with tourism demand (Nabeul, Sousse, Monastir, Mahdia, Sfax, Médenine)</li>
</ul>

<h3>⚠️ Important</h3>
<p>Target encoding (<code>gov_median_price</code>) uses the full training set here. In the final model pipeline this must be computed only on the training fold to prevent data leakage during cross-validation.</p>

````python
# ── Governorate-level target encoding ─────────────────────────────────────
gov_stats = df.groupby("governorate").agg(
    gov_median_price       = ("price",        "median"),
    gov_median_price_per_m2= ("price_per_m2", "median"),
    gov_listing_count      = ("id",           "count"),
).reset_index()

df = df.merge(gov_stats, on="governorate", how="left")

# ── Geographic cluster flags ───────────────────────────────────────────────
GREATER_TUNIS = {"Tunis", "Ariana", "Ben Arous", "Manouba"}
COASTAL       = {"Nabeul", "Sousse", "Monastir", "Mahdia", "Sfax",
                 "Médenine", "Gabès", "Bizerte"}

df["is_greater_tunis"] = df["governorate"].isin(GREATER_TUNIS).astype(int)
df["is_coastal"]       = df["governorate"].isin(COASTAL).astype(int)

# ── Log-transform the target-encoded price ────────────────────────────────
df["log_gov_median_price"] = np.log1p(df["gov_median_price"])

print("Geographic features added:")
print(f"  gov_median_price        — range: {df['gov_median_price'].min():,.0f} – {df['gov_median_price'].max():,.0f} TND")
print(f"  gov_listing_count       — range: {df['gov_listing_count'].min()} – {df['gov_listing_count'].max():,}")
print(f"  is_greater_tunis = 1    — {df['is_greater_tunis'].sum():,} listings ({df['is_greater_tunis'].mean()*100:.1f}%)")
print(f"  is_coastal = 1          — {df['is_coastal'].sum():,} listings ({df['is_coastal'].mean()*100:.1f}%)")
print(f"\nTop 5 governorates by median price:")
print(gov_stats.nlargest(5, "gov_median_price")[["governorate","gov_median_price","gov_listing_count"]].to_string(index=False))
````

<h2 style="color: #87BAC3;">🏠 Group 3 — Property Type & Categorical Encoding</h2>

<h3>🎯 Purpose</h3>
<p>Categorical columns (<code>property_type</code>, <code>source</code>) are encoded for model consumption:</p>
<ul>
  <li><code>property_type_encoded</code> — ordinal target encoding of property type by median price, capturing the implicit price hierarchy (Building > Villa > Duplex > Apartment > Land...)</li>
  <li>One-hot encoding for <code>property_type</code> — preserves full categorical signal for tree models</li>
  <li><code>source_encoded</code> — label encoding of the data source, a proxy for listing quality and completeness</li>
  <li><code>type_median_price</code> — median price per property type, direct target encoding</li>
</ul>

````python
# ── Property type target encoding ─────────────────────────────────────────
type_stats = df.groupby("property_type").agg(
    type_median_price = ("price", "median"),
    type_count        = ("id",    "count"),
).reset_index()

df = df.merge(type_stats, on="property_type", how="left")
df["log_type_median_price"] = np.log1p(df["type_median_price"])

# ── One-hot encode property_type ───────────────────────────────────────────
pt_dummies = pd.get_dummies(df["property_type"], prefix="type", drop_first=False)
df = pd.concat([df, pt_dummies], axis=1)

# ── Source label encoding ──────────────────────────────────────────────────
source_order = df["source"].value_counts().index.tolist()
source_map   = {s: i for i, s in enumerate(source_order)}
df["source_encoded"] = df["source"].map(source_map).fillna(-1).astype(int)

print("Property type features added:")
print(type_stats.sort_values("type_median_price", ascending=False).to_string(index=False))
print(f"\nOne-hot columns added: {[c for c in pt_dummies.columns]}")
print(f"source_encoded range: {df['source_encoded'].min()} – {df['source_encoded'].max()}")
````

<h2 style="color: #87BAC3;">🔑 Group 4 — Amenity Composite Score</h2>

<h3>🎯 Purpose</h3>
<p>Individual binary amenity flags are meaningful but a composite score aggregates them into a single <em>listing quality</em> signal. Two composite features are created:</p>
<ul>
  <li><code>amenity_score</code> — raw count of amenities present (0–12)</li>
  <li><code>luxury_score</code> — weighted score emphasising premium amenities (pool, sea view, standing, elevator) that carry a strong price premium in the Tunisian market</li>
  <li><code>has_any_amenity</code> — binary flag: does this listing mention at least one amenity?</li>
</ul>

````python
# ── Amenity columns ───────────────────────────────────────────────────────
amenity_cols = [
    "has_elevator", "has_basement", "has_heating", "has_air_conditioning",
    "has_garden", "has_furniture", "has_parking", "has_pool",
    "has_security", "has_standing", "has_terrace", "has_sea_view",
]

# Coerce all to numeric (some may still be bool)
for col in amenity_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col].map({True:1, False:0, 1:1, 0:0}).fillna(0))

# ── Raw amenity count ──────────────────────────────────────────────────────
df["amenity_score"] = df[[c for c in amenity_cols if c in df.columns]].sum(axis=1)

# ── Weighted luxury score ──────────────────────────────────────────────────
luxury_weights = {
    "has_pool"           : 3,
    "has_sea_view"       : 3,
    "has_standing"       : 2,
    "has_elevator"       : 2,
    "has_security"       : 2,
    "has_garden"         : 1,
    "has_parking"        : 1,
    "has_terrace"        : 1,
    "has_air_conditioning": 1,
    "has_heating"        : 1,
    "has_furniture"      : 1,
    "has_basement"       : 1,
}
df["luxury_score"] = sum(
    df[col] * weight
    for col, weight in luxury_weights.items()
    if col in df.columns
)
df["has_any_amenity"] = (df["amenity_score"] > 0).astype(int)

print("Amenity features added:")
print(f"  amenity_score  — mean: {df['amenity_score'].mean():.2f}, max: {df['amenity_score'].max()}")
print(f"  luxury_score   — mean: {df['luxury_score'].mean():.2f}, max: {df['luxury_score'].max()}")
print(f"  has_any_amenity=1 : {df['has_any_amenity'].sum():,} ({df['has_any_amenity'].mean()*100:.1f}%)")
print(f"\nAmenity score distribution:")
print(df["amenity_score"].value_counts().sort_index().head(10))
````

<h2 style="color: #87BAC3;">📝 Group 5 — Text-Derived Structured Signals</h2>

<h3>🎯 Purpose</h3>
<p>The raw description text feeds the NLP branch of the valuation model. Here, lightweight regex-based signal extraction produces <strong>structured features</strong> from text that the structured branch of the model can use directly — without needing NLP inference at prediction time:</p>
<ul>
  <li><code>desc_mentions_price</code> — description explicitly states a price (may indicate negotiability)</li>
  <li><code>desc_mentions_new</code> — keywords indicating new construction (neuf, nouveau, nouvelle) — strong price premium signal</li>
  <li><code>desc_mentions_renovated</code> — renovation or refurbishment keywords</li>
  <li><code>desc_mentions_view</code> — mentions of sea/panoramic views (beyond the structured flag)</li>
  <li><code>desc_mentions_investment</code> — investment or rental yield language</li>
  <li><code>title_word_count</code> — length of the listing title (longer titles tend to be more professional)</li>
  <li><code>description_density</code> — words per m² of described surface (a proxy for listing effort relative to property size)</li>
</ul>

````python
import re

def flag_keywords(series, patterns):
    combined = '|'.join(patterns)
    return series.str.lower().str.contains(combined, na=False, regex=True).astype(int)

# ── Description keyword flags ──────────────────────────────────────────────
desc = df["description"].fillna("") + " " + df["title"].fillna("")

df["desc_mentions_new"]        = flag_keywords(desc, [r"\bneuf\b", r"nouvelle construction",
                                                        r"\bnew\b", r"nouvelle résidence"])
df["desc_mentions_renovated"]  = flag_keywords(desc, [r"rénov", r"renov", r"refait",
                                                        r"remis à neuf", r"retapé"])
df["desc_mentions_view"]       = flag_keywords(desc, [r"vue mer", r"vue sur mer",
                                                        r"vue panoramique", r"sea view"])
df["desc_mentions_investment"] = flag_keywords(desc, [r"investissement", r"rentabilité",
                                                        r"rendement", r"investment"])
df["desc_mentions_price"]      = flag_keywords(desc, [r"\d+\s*(dt|tnd|dinar)",
                                                        r"prix négociable", r"à débattre"])
df["desc_mentions_furnished"]  = flag_keywords(desc, [r"\bmeublé\b", r"\bfurnished\b",
                                                        r"\bmeublée\b"])

# ── Title word count ───────────────────────────────────────────────────────
df["title_word_count"] = df["title"].fillna("").str.split().str.len()

# ── Description density (words per m²) ────────────────────────────────────
df["description_density"] = np.where(
    df["has_surface"] & (df["surface_m2"] > 0),
    df["description_word_count"] / df["surface_m2"],
    np.nan
)

text_flags = ["desc_mentions_new", "desc_mentions_renovated", "desc_mentions_view",
              "desc_mentions_investment", "desc_mentions_price", "desc_mentions_furnished"]

print("Text-derived features:")
for col in text_flags:
    n = df[col].sum()
    print(f"  {col:<32} {n:>6,} ({n/len(df)*100:.1f}%)")
print(f"  title_word_count — mean: {df['title_word_count'].mean():.1f}")
print(f"  description_density — non-null: {df['description_density'].notna().sum():,}")
````

<h2 style="color: #87BAC3;">🧠 LLM Setup — Qwen2.5-0.5B-Instruct</h2>
<h3>🎯 Why Qwen2.5-0.5B-Instruct?</h3>
<ul>
  <li><strong>500M parameters</strong> — fits in ~1 GB RAM, runs on CPU in 3–5 seconds per call</li>
  <li><strong>Instruction-tuned</strong> — reliably outputs structured JSON</li>
  <li><strong>Multilingual</strong> — handles French/Arabic property text natively</li>
  <li><strong>Apache 2.0</strong> — fully open source, no API key, works offline</li>
</ul>

````python
# -----------------------------------------------------------------------------
# CELL 111 -- Imports, Logging, and LLM Setup (OpenRouter + Token Factory)
# Replace your existing Cell 111 entirely with this cell.
#
# Primary  : OpenRouter  -> google/gemma-4-31b-it:free
# Fallback : Token Factory (ESPRIT) -> hosted_vllm/Llama-3.1-70B-Instruct
# -----------------------------------------------------------------------------

import json, logging, os, re, subprocess, sys, time, threading, requests
from abc import ABC, abstractmethod
from datetime import datetime
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# -- Logging ------------------------------------------------------------------
Path("logs").mkdir(exist_ok=True)

def _build_logger(name: str) -> logging.Logger:
    logger = logging.getLogger(name)
    if logger.handlers:
        return logger
    logger.setLevel(logging.DEBUG)
    fmt = logging.Formatter(
        "%(asctime)s | %(name)-26s | %(levelname)-8s | %(message)s",
        datefmt="%H:%M:%S"
    )
    ch = logging.StreamHandler(sys.stdout)
    ch.setLevel(logging.INFO)
    ch.setFormatter(fmt)
    logger.addHandler(ch)
    fh = logging.FileHandler("logs/pipeline.log", encoding="utf-8")
    fh.setLevel(logging.DEBUG)
    fh.setFormatter(fmt)
    logger.addHandler(fh)
    return logger

_root = logging.getLogger()
_root.setLevel(logging.DEBUG)
print("Logging configured -- INFO to console, DEBUG to logs/pipeline.log")

# -- OpenRouter config (PRIMARY) ----------------------------------------------
# Paste your key here, or set the environment variable OPENROUTER_API_KEY
# before launching Jupyter.  The env-var approach is safer (never commits to git).
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY", "YOUR_OPENROUTER_KEY_HERE")

# Primary model -- Gemma 4 31B: free-tier on OpenRouter, strong reasoning
LLM_PRIMARY = "google/gemma-4-31b-it:free"

OPENROUTER_URL     = "https://openrouter.ai/api/v1/chat/completions"
OPENROUTER_HEADERS = {
    "Authorization": f"Bearer {OPENROUTER_API_KEY}",
    "Content-Type":  "application/json",
    "HTTP-Referer":  "https://github.com/lenstate-pipeline",
    "X-Title":       "LensEstate Pipeline",
}

# -- Token Factory config (FALLBACK -- ESPRIT university platform) -------------
# Set TOKEN_FACTORY_API_KEY as an environment variable, or paste it below.
# Endpoint: https://tokenfactory.esprit.tn/api  (OpenAI-compatible)
# Available models: hosted_vllm/Llama-3.1-70B-Instruct, hosted_vllm/llava-1.5-7b-hf
TOKEN_FACTORY_API_KEY  = os.getenv("TOKEN_FACTORY_API_KEY", "YOUR_TOKEN_FACTORY_KEY_HERE")
TOKEN_FACTORY_URL      = "https://tokenfactory.esprit.tn/api/chat/completions"
TOKEN_FACTORY_HEADERS  = {
    "Authorization": f"Bearer {TOKEN_FACTORY_API_KEY}",
    "Content-Type":  "application/json",
}
# Fallback models served by Token Factory
LLM_FALLBACK       = "hosted_vllm/Llama-3.1-70B-Instruct"
LLM_FALLBACK_SMALL = "hosted_vllm/llava-1.5-7b-hf"   # last-resort tiny fallback

_LLM_SYSTEM_PROMPT = (
    "You are a data-pipeline reasoning agent for LensEstate, "
    "a Tunisian real-estate platform. "
    "Always reply with a single valid JSON object and nothing else. "
    "No markdown, no prose outside the JSON. "
    "Include a 'thought' field where you reason step-by-step "
    "before committing to an action."
)

# -- LLM helper posts ---------------------------------------------------------
_llm_log = _build_logger("LLM")

def _post_openrouter(model: str, prompt: str, max_tokens: int, timeout: int) -> requests.Response:
    """POST to OpenRouter."""
    return requests.post(
        OPENROUTER_URL,
        headers=OPENROUTER_HEADERS,
        json={
            "model":       model,
            "max_tokens":  max_tokens,
            "temperature": 0.15,
            "messages": [
                {"role": "system", "content": _LLM_SYSTEM_PROMPT},
                {"role": "user",   "content": prompt},
            ],
        },
        timeout=timeout,
    )


def _post_token_factory(model: str, prompt: str, max_tokens: int, timeout: int) -> requests.Response:
    """POST to Token Factory (ESPRIT) -- OpenAI-compatible endpoint."""
    return requests.post(
        TOKEN_FACTORY_URL,
        headers=TOKEN_FACTORY_HEADERS,
        json={
            "model":             model,
            "max_tokens":        max_tokens,
            "temperature":       0.15,
            "top_p":             0.9,
            "frequency_penalty": 0.0,
            "presence_penalty":  0.0,
            "messages": [
                {"role": "system", "content": _LLM_SYSTEM_PROMPT},
                {"role": "user",   "content": prompt},
            ],
        },
        timeout=timeout,
        verify=False,   # Token Factory uses self-signed cert (-k equivalent)
    )


def call_llm(
    prompt:     str,
    model:      str  = LLM_PRIMARY,
    max_tokens: int  = 512,
    retries:    int  = 3,
    timeout:    int  = 60,
) -> str:
    """
    Send a prompt and return the raw text response.

    Routing logic:
      1. Try OpenRouter with the requested model (default: LLM_PRIMARY).
      2. On 429 (rate-limit) or repeated 5xx -> switch to Token Factory
         with LLM_FALLBACK (Llama-3.1-70B-Instruct).
      3. If Token Factory also fails -> try LLM_FALLBACK_SMALL as last resort.
      4. All paths exhausted -> return a safe deterministic JSON string.
    """
    primary_key_ok  = OPENROUTER_API_KEY  != "YOUR_OPENROUTER_KEY_HERE"
    fallback_key_ok = TOKEN_FACTORY_API_KEY != "YOUR_TOKEN_FACTORY_KEY_HERE"

    if not primary_key_ok:
        _llm_log.warning("OpenRouter key not set -- will attempt Token Factory directly")

    use_token_factory = not primary_key_ok or model.startswith("hosted_vllm/")
    current_model     = LLM_FALLBACK if use_token_factory else model
    backend           = "token_factory" if use_token_factory else "openrouter"
    backoff           = 5

    for attempt in range(1, retries + 2):   # +1 extra slot for fallback switch
        _llm_log.debug(f"LLM attempt {attempt} | backend={backend} | model={current_model}")
        try:
            if backend == "openrouter":
                resp = _post_openrouter(current_model, prompt, max_tokens, timeout)
            else:
                if not fallback_key_ok:
                    _llm_log.error("Token Factory key not set -- cannot call fallback")
                    break
                resp = _post_token_factory(current_model, prompt, max_tokens, timeout)

            # Rate-limit: escalate to Token Factory
            if resp.status_code == 429:
                if backend == "openrouter":
                    _llm_log.warning(
                        f"OpenRouter rate-limited on {current_model} "
                        f"-- switching to Token Factory ({LLM_FALLBACK})"
                    )
                    backend       = "token_factory"
                    current_model = LLM_FALLBACK
                else:
                    _llm_log.warning(f"Token Factory rate-limited -- waiting {backoff}s")
                    time.sleep(backoff)
                    backoff *= 2
                continue

            # Server errors: wait and retry, escalate after 2nd failure
            if resp.status_code >= 500:
                _llm_log.warning(f"Server error {resp.status_code} ({backend}) -- waiting {backoff}s")
                time.sleep(backoff)
                backoff *= 2
                if backend == "openrouter" and attempt >= 2:
                    _llm_log.warning("Escalating to Token Factory after repeated 5xx")
                    backend       = "token_factory"
                    current_model = LLM_FALLBACK
                continue

            resp.raise_for_status()
            raw = resp.json()["choices"][0]["message"]["content"].strip()
            _llm_log.debug(f"LLM raw response ({len(raw)} chars): {raw[:300]}")
            return raw

        except requests.exceptions.SSLError:
            _llm_log.warning("SSL error on Token Factory (verify=False set -- check connection)")
        except requests.exceptions.Timeout:
            _llm_log.warning(f"LLM timeout after {timeout}s (attempt {attempt})")
        except requests.exceptions.ConnectionError as e:
            _llm_log.warning(f"LLM connection error (attempt {attempt}): {e}")
        except Exception as e:
            _llm_log.error(f"Unexpected LLM error (attempt {attempt}): {e}")

        if attempt < retries + 1:
            time.sleep(backoff)
            backoff *= 2

    # All retries exhausted -- try tiny fallback one last time
    if fallback_key_ok and current_model != LLM_FALLBACK_SMALL:
        _llm_log.warning(f"Trying last-resort model {LLM_FALLBACK_SMALL}")
        try:
            resp = _post_token_factory(LLM_FALLBACK_SMALL, prompt, min(max_tokens, 256), timeout)
            resp.raise_for_status()
            raw = resp.json()["choices"][0]["message"]["content"].strip()
            _llm_log.debug(f"Last-resort response: {raw[:200]}")
            return raw
        except Exception as e:
            _llm_log.error(f"Last-resort model also failed: {e}")

    _llm_log.error("All LLM paths exhausted -- returning safe deterministic fallback")
    return json.dumps({
        "thought":    "All LLM paths exhausted. Defaulting to safe continue.",
        "action":     "continue",
        "reason":     "llm_unavailable",
        "confidence": 0.3,
    })


# -- JSON parser --------------------------------------------------------------
def parse_llm_json(raw: str) -> Dict:
    """
    Robustly extract a JSON object from an LLM response.
    Handles: plain JSON, ```json fences, leading/trailing prose.
    """
    try:
        clean = raw.strip()
        if "```" in clean:
            parts = clean.split("```")
            for part in parts:
                if part.startswith("json"):
                    part = part[4:]
                part = part.strip()
                if part.startswith("{"):
                    clean = part
                    break
        m = re.search(r'\{.*\}', clean, re.DOTALL)
        if m:
            return json.loads(m.group())
        return json.loads(clean)
    except Exception:
        _llm_log.warning(f"JSON parse failed on: {raw[:200]}")
        return {
            "thought":    "Could not parse LLM response as JSON.",
            "action":     "continue",
            "reason":     "parse_error",
            "confidence": 0.3,
        }


# -- Misc helpers -------------------------------------------------------------
def _now() -> str:
    return datetime.utcnow().isoformat() + "Z"


# -- Smoke test ---------------------------------------------------------------
def test_llm_connection() -> bool:
    """Quick sanity check -- call this once after setting your API keys."""
    print("Testing LLM connection...")
    print(f"  Primary  (OpenRouter)    : {LLM_PRIMARY}")
    print(f"  Fallback (Token Factory) : {LLM_FALLBACK}")
    print(f"  OpenRouter key   : {'yes' if OPENROUTER_API_KEY  != 'YOUR_OPENROUTER_KEY_HERE'  else 'NOT SET'}")
    print(f"  Token Factory key: {'yes' if TOKEN_FACTORY_API_KEY != 'YOUR_TOKEN_FACTORY_KEY_HERE' else 'NOT SET'}")
    raw    = call_llm('Reply with exactly: {"status": "ok", "thought": "connection works"}')
    result = parse_llm_json(raw)
    if result.get("status") == "ok":
        print("  LLM connection OK")
        print(f"  thought: {result.get('thought')}")
        return True
    else:
        print(f"  Unexpected response: {result}")
        return False

print("\nAll utilities ready")
print(f"   Primary  (OpenRouter)    : {LLM_PRIMARY}")
print(f"   Fallback (Token Factory) : {LLM_FALLBACK}")
print(f"   OpenRouter key  : {'set' if OPENROUTER_API_KEY  != 'YOUR_OPENROUTER_KEY_HERE'  else 'NOT SET -- set OPENROUTER_API_KEY'}")
print(f"   Token Factory key: {'set' if TOKEN_FACTORY_API_KEY != 'YOUR_TOKEN_FACTORY_KEY_HERE' else 'NOT SET -- set TOKEN_FACTORY_API_KEY'}")
print("\nRun test_llm_connection() after setting your API keys to verify everything works.")
````

<h2 style="color: #87BAC3;">💾 Agent Memory System</h2>
<p>Each agent has <strong>short-term</strong> (runtime dict, cleared each run) and <strong>long-term</strong> (persisted JSON in <code>memory/state/</code>) memory. Long-term memory allows agents to learn across runs — e.g. skipping scrapers that consistently fail.</p>

````python
class AgentMemory:
    """
    Two-tier memory system for each agent.

    Tier 1 — Episodic (short-term, per-run scratchpad):
        self.short  — cleared at the start of every run() call.
        Stores the current run's action, result, errors, warnings.

    Tier 2 — Semantic (long-term, persisted to disk as JSON):
        self.long   — survives across runs. Contains:
            successes       list  — every successful scrape/operation
            failures        list  — every failure with reason + timestamp
            decisions       list  — every LLM routing decision with thought
            quality         list  — quality metrics per run
            schema_repairs  list  — LLM-inferred column mappings (StandardizationAgent)
            suspicious_flags list — LLM anomaly detection results (CleaningAgent)
            knowledge        dict — free-form cross-run facts agents can read/write
                                    e.g. {"mubawab": {"typical_rows": 1200,
                                                       "known_issues": ["slow sundays"]}}

    The `knowledge` sub-dict is the semantic memory store. Agents read it
    at the start of their ReAct loop and write to it when they discover
    something durable (a typical row range, a recurring failure pattern, etc.).
    """

    def __init__(self, name: str, state_dir: str = "memory/state/"):
        self.name   = name
        self._dir   = Path(state_dir)
        self._dir.mkdir(parents=True, exist_ok=True)
        self._lock  = threading.Lock()
        self._path  = self._dir / f"{name}_memory.json"
        self.short  : Dict[str, Any] = {
            "action": None, "result": None, "errors": [], "warnings": []
        }
        self.long   : Dict[str, Any] = self._load()
        self._log   = _build_logger(f"Memory.{name}")

    # ── Episodic (short-term) ─────────────────────────────────────────────────
    def set(self, k: str, v: Any):
        self.short[k] = v

    def get(self, k: str, d=None):
        return self.short.get(k, d)

    def add_error(self, e: str):
        self.short["errors"].append({"ts": _now(), "msg": e})
        self._log.debug(f"[ERROR] {e}")

    def add_warning(self, w: str):
        self.short["warnings"].append({"ts": _now(), "msg": w})
        self._log.debug(f"[WARNING] {w}")

    def clear(self):
        self.short = {"action": None, "result": None, "errors": [], "warnings": []}

    # ── Semantic (long-term event log) ────────────────────────────────────────
    def remember(self, cat: str, entry: Dict):
        """Append an event to a named category list (capped at 200 entries)."""
        entry["ts"] = _now()
        with self._lock:
            self.long.setdefault(cat, []).append(entry)
            if len(self.long[cat]) > 200:
                self.long[cat] = self.long[cat][-200:]
            self._save()
        self._log.debug(f"[MEMORY:{cat}] {entry}")

    def recall(self, cat: str, n: int = 10) -> List[Dict]:
        """Return the last n entries from a category."""
        return self.long.get(cat, [])[-n:]

    # ── Semantic knowledge store (free-form facts) ────────────────────────────
    def know(self, key: str, value: Any):
        """
        Write a durable fact to the knowledge store.
        key   — dot-separated path, e.g. "mubawab.typical_rows"
        value — any JSON-serialisable value

        Example:
            agent.memory.know("mubawab.typical_rows", 1250)
            agent.memory.know("mubawab.known_issues", ["slow on sundays"])
        """
        parts = key.split(".")
        with self._lock:
            node = self.long.setdefault("knowledge", {})
            for part in parts[:-1]:
                node = node.setdefault(part, {})
            node[parts[-1]] = value
            self.long["knowledge"]["_last_updated"] = _now()
            self._save()
        self._log.debug(f"[KNOWLEDGE] {key} = {value}")

    def recall_knowledge(self, key: str = None) -> Any:
        """
        Read from the knowledge store.
        key=None  → return the entire knowledge dict
        key="mubawab.typical_rows" → return that specific value, or None
        """
        store = self.long.get("knowledge", {})
        if key is None:
            return store
        parts = key.split(".")
        node  = store
        for part in parts:
            if not isinstance(node, dict):
                return None
            node = node.get(part)
        return node

    # ── Long-term scalar setters (backward compat) ─────────────────────────────
    def set_lt(self, k: str, v: Any):
        with self._lock:
            self.long[k] = v
            self._save()

    def get_lt(self, k: str, d=None):
        return self.long.get(k, d)

    # ── Persistence ───────────────────────────────────────────────────────────
    def _load(self) -> Dict:
        if self._path.exists():
            try:
                with open(self._path, encoding="utf-8") as f:
                    return json.load(f)
            except Exception:
                pass
        return {
            "successes":        [],
            "failures":         [],
            "decisions":        [],
            "quality":          [],
            "schema_repairs":   [],
            "suspicious_flags": [],
            "knowledge":        {},
        }

    def _save(self):
        tmp = self._path.with_suffix(".tmp")
        with open(tmp, "w", encoding="utf-8") as f:
            json.dump(self.long, f, indent=2, ensure_ascii=False)
        os.replace(tmp, self._path)

    def summary(self) -> Dict:
        """Return a compact summary for logging/debugging."""
        return {
            "name":       self.name,
            "successes":  len(self.long.get("successes", [])),
            "failures":   len(self.long.get("failures", [])),
            "decisions":  len(self.long.get("decisions", [])),
            "knowledge":  list(self.long.get("knowledge", {}).keys()),
        }

print("✅ AgentMemory defined (episodic + semantic knowledge store)")
````

<h2 style="color: #87BAC3;">🧩 Base Agent</h2>
<p>Every agent inherits from <code>BaseAgent</code> which provides: shared logger, memory instance, LLM <code>decide()</code> interface with graceful fallback, <code>log_decision()</code>, and the <code>run(state)</code> abstract contract.</p>

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 113 — BaseAgent  (Phase 4 update: heuristics injection in decide())
# ─────────────────────────────────────────────────────────────────────────────

class BaseAgent(ABC):
    def __init__(self, name: str):
        self.name   = name
        self.memory = AgentMemory(name)
        self.log    = _build_logger(name)

    @abstractmethod
    def run(self, state: Dict[str, Any]) -> Dict[str, Any]: ...

    # ── LLM reasoning ─────────────────────────────────────────────────────────
    def decide(
        self,
        prompt:       str,
        context:      Optional[Dict] = None,
        model:        str            = LLM_PRIMARY,
        scope:        Optional[str]  = None,
    ) -> Dict:
        """
        Ask the LLM to reason about a situation and return a structured decision.

        Phase 4: automatically prepends any relevant heuristics from
        HeuristicsStore so the LLM benefits from past-run experience.

        Args:
            prompt   — the question / task for the LLM
            context  — structured data serialised as JSON and prepended
            model    — which OpenRouter model to use
            scope    — optional scraper name to filter heuristics (e.g. 'mubawab')

        Returns dict always containing: thought, action, reason, confidence
        """
        # ── Heuristics block ──────────────────────────────────────────────
        heuristics_block = ""
        try:
            # HEURISTICS is the global singleton defined in Cell 119b
            h_text = HEURISTICS.format_for_prompt(self.name, scope=scope)
            if h_text:
                heuristics_block = h_text + "\n\n"
        except NameError:
            pass   # HeuristicsStore not yet defined (early cells running alone)

        # ── Context block ─────────────────────────────────────────────────
        ctx_block = ""
        if context:
            ctx_block = (
                "PIPELINE CONTEXT (current run data):\n"
                + json.dumps(context, indent=2, default=str)
                + "\n\n"
            )

        full_prompt = heuristics_block + ctx_block + prompt
        self.log.debug(f"decide() prompt ({len(full_prompt)} chars):\n{full_prompt[:600]}")

        raw    = call_llm(full_prompt, model=model)
        result = parse_llm_json(raw)

        result.setdefault("thought",    "(no thought provided)")
        result.setdefault("action",     "continue")
        result.setdefault("reason",     "no reason provided")
        result.setdefault("confidence", 0.5)

        self.log.debug(f"decide() result: {result}")
        return result

    def log_decision(self, d: Dict, ctx: str = ""):
        """Persist decision to memory and print a structured log block."""
        self.memory.remember("decisions", {"decision": d, "context": ctx})
        self.log.info(f"┌─ DECISION [{ctx}]")
        thought = d.get("thought", "")
        if thought and thought != "(no thought provided)":
            for i, line in enumerate(thought.split(". ")):
                prefix = "│  thought   : " if i == 0 else "│             "
                self.log.info(f"{prefix}{line.strip()}.")
        self.log.info(f"│  action     : {d.get('action', '?')}")
        self.log.info(f"│  reason     : {d.get('reason', '?')}")
        self.log.info(f"└─ confidence : {d.get('confidence', '?')}")

    # ── State helpers ──────────────────────────────────────────────────────────
    def update_state(self, state: Dict, k: str, v: Any) -> Dict:
        state[k] = v
        self.log.debug(f"State updated — {k}: {str(v)[:120]}")
        return state

    # ── Subprocess helpers (unchanged) ─────────────────────────────────────────
    @staticmethod
    def _resolve_python(cwd: str) -> str:
        from pathlib import Path as _P
        cwd_path   = _P(cwd)
        candidates = [
            cwd_path / "venv"  / "Scripts" / "python.exe",
            cwd_path / "venv"  / "bin"     / "python",
            cwd_path / ".venv" / "Scripts" / "python.exe",
            cwd_path / ".venv" / "bin"     / "python",
        ]
        for c in candidates:
            if c.exists():
                return str(c)
        return sys.executable

    def _run_subprocess(
        self, cmd: List[str], cwd: str, timeout: int = 3600,
    ) -> Tuple[bool, str, str]:
        self.log.info(f"  $ {' '.join(cmd)}")
        self.log.info(f"  cwd: {cwd}")
        try:
            r = subprocess.run(
                cmd, cwd=cwd, capture_output=True, text=True, timeout=timeout
            )
            if r.returncode == 0:
                self.log.info("  ✅ exit 0")
                if r.stdout:
                    self.log.debug(f"  STDOUT: {r.stdout[-500:]}")
            else:
                self.log.error(f"  ❌ exit {r.returncode}")
                self.log.error(f"  STDERR: {r.stderr[-800:]}")
            return r.returncode == 0, r.stdout, r.stderr
        except subprocess.TimeoutExpired:
            self.log.error(f"  ❌ TIMEOUT after {timeout}s")
            return False, "", "TimeoutExpired"
        except FileNotFoundError as e:
            self.log.error(f"  ❌ Script not found: {e}")
            return False, "", str(e)
        except Exception as e:
            self.log.error(f"  ❌ Unexpected: {e}")
            return False, "", str(e)

    def _count_rows(self, path: str) -> int:
        try:
            with open(path, errors="ignore") as f:
                n = sum(1 for _ in f) - 1
            self.log.debug(f"Row count [{path}]: {n:,}")
            return max(n, 0)
        except Exception:
            return 0

print("✅ BaseAgent defined (Phase 4: heuristics injected into every decide() call)")
````

<h2 style="color: #87BAC3;">📰 Agent 1 — Articles Collection Agent <span style="font-size:13px; color:#87BAC3;">(runs first)</span></h2>

<h3>🎯 What it does</h3>
<p>Runs <strong>BnB blog scraper</strong> (<code>C:\bnbblog\bnb_tunisie_scraper.py</code>) and <strong>Houni scraper</strong> (<code>C:\houni.scraper\houni_scraper.py</code>) in sequence. Both scrapers overwrite their output on every run — no checkpoint system. The agent checks if existing output is less than 7 days old before deciding to re-run.</p>

<h3>📁 Scraper details</h3>
<ul>
  <li><strong>BnB</strong>: <code>cd C:\bnbblog && python bnb_tunisie_scraper.py</code> → outputs <code>bnb_blog_articles.csv</code> + <code>.json</code> (~78 articles, ~3–5 min)</li>
  <li><strong>Houni</strong>: <code>cd C:\houni.scraper && python houni_scraper.py</code> → outputs <code>houni_blog_articles.csv</code> (~51 articles, fast)</li>
</ul>

<h3>🧠 Decision logic (deterministic)</h3>
<ul>
  <li>Check if output CSV exists AND is < 7 days old → skip re-scraping, use existing</li>
  <li>Check long-term memory for recent failures → skip if failed 3+ times recently</li>
  <li>Run with retry (max 2 attempts, 15s backoff)</li>
  <li>After scraping: compute quality score (empty %, duplicate title %) → store in memory</li>
</ul>

````python
class ArticlesAgent(BaseAgent, FallbackMixin):
    """
    Runs BnB and Houni blog scrapers.
    Runs FIRST in the pipeline — fast (~5-10 min total), non-blocking.

    Phase 2: ReAct loop — LLM decides run/skip per scraper with reasoning.
    Phase 3: FallbackMixin — degradation ladder when a scraper fails.
             Semantic memory — reads/writes known article counts per source.
    """

    SCRAPERS = {
        "bnb": {
            "cwd"          : r"C:\bnbblog",
            "script"       : "bnb_tunisie_scraper.py",
            "output_csv"   : r"C:\bnbblog\bnb_blog_articles.csv",
            "output_json"  : r"C:\bnbblog\bnb_blog_articles.json",
            "min_rows"     : 10,
            "max_age_hours": 6,
            "timeout"      : 600,
        },
        "houni": {
            "cwd"          : r"C:\houni.scraper",
            "script"       : "houni_scraper.py",
            "args"         : [],
            "output_csv"   : r"C:\houni.scraper\houni_blog_articles.csv",
            "min_rows"     : 10,
            "max_age_hours": 6,
            "timeout"      : 600,
        },
    }
    MAX_RECENT_FAILURES = 3
    MAX_RETRIES         = 2
    BACKOFF_SECONDS     = 15
    MAX_REACT_STEPS     = 10

    def __init__(self):
        super().__init__("ArticlesAgent")

    # ── Public entry point ────────────────────────────────────────────────────
    def run(self, state: Dict[str, Any]) -> Dict[str, Any]:
        self.log.info("=" * 60)
        self.log.info("ArticlesAgent — starting (ReAct + Fallback)")
        self.log.info("=" * 60)
        self.memory.clear()
        self.memory.set("action", "collect_articles")

        # Log semantic knowledge about these scrapers if we have any
        knowledge = self.memory.recall_knowledge()
        if knowledge:
            self.log.info(f"  Semantic knowledge: {list(knowledge.keys())}")

        collected, results = self._react_loop()
        merged_path = self._merge_articles(collected)

        state = self.update_state(state, "articles_raw_paths",   collected)
        state = self.update_state(state, "articles_results",     results)
        state = self.update_state(state, "merged_articles_path", merged_path or "")

        self.log.info(f"\nArticlesAgent done — {len(collected)} sources collected")
        if merged_path:
            self.log.info(f"  merged → {merged_path}")
        return state

    # ── ReAct loop ────────────────────────────────────────────────────────────
    def _react_loop(self) -> Tuple[List[str], Dict]:
        """
        Tools:
          inspect_scraper(name)        — state snapshot
          run_scraper(name)            — execute subprocess
          skip_scraper(name, reason)   — skip, use cache if available
          activate_fallback(name)      — trigger degradation ladder explicitly
          finish()                     — exit loop
        """
        collected: List[str] = []
        results:   Dict      = {}
        handled:   set       = set()

        scraper_summary = {
            name: self._inspect_scraper(name, cfg)
            for name, cfg in self.SCRAPERS.items()
        }

        # Include semantic knowledge in context
        knowledge = self.memory.recall_knowledge()

        system_context = (
            "You are ArticlesAgent for LensEstate, a Tunisian real estate platform.\n"
            "Manage two article blog scrapers: bnb, houni.\n"
            "Articles are non-blocking — pipeline continues even if both fail.\n\n"
            "Available tools (call ONE per step):\n"
            "  inspect_scraper(name)         — output age, rows, failure history\n"
            "  run_scraper(name)             — execute subprocess\n"
            "  skip_scraper(name, reason)    — use cache or skip\n"
            "  activate_fallback(name)       — trigger degradation ladder\n"
            "  finish()                      — all scrapers handled\n\n"
            "Rules:\n"
            "  SKIP  if output fresh (< max_age_hours) AND rows >= min_rows\n"
            "  SKIP  if recent_failures >= 3 AND no valid output\n"
            "  RUN   if output stale, missing, or below min_rows\n"
            "  FALLBACK if scraper failed and cache exists\n\n"
            f"Scraper states:\n{json.dumps(scraper_summary, indent=2)}\n\n"
            f"Semantic knowledge from past runs:\n{json.dumps(knowledge, indent=2)}\n\n"
            'Reply ONLY JSON: {"thought": "reasoning", "tool": "name", "args": {}}'
        )

        messages = [{"role": "user", "content": system_context}]

        for step in range(self.MAX_REACT_STEPS):
            self.log.info(f"\n  [ReAct step {step+1}/{self.MAX_REACT_STEPS}]")

            recent = messages[-3:] if len(messages) > 3 else messages
            prompt = "\n---\n".join(m["content"] for m in recent)
            raw    = call_llm(prompt, max_tokens=512)
            parsed = parse_llm_json(raw)

            thought = parsed.get("thought", "")
            tool    = parsed.get("tool", "finish")
            args    = parsed.get("args", {})

            self.log.info(f"  thought : {thought[:200]}")
            self.log.info(f"  tool    : {tool}  args={args}")

            # ── Tool dispatch ─────────────────────────────────────────────
            if tool == "inspect_scraper":
                name        = args.get("name", "")
                result_data = self._inspect_scraper(name, self.SCRAPERS.get(name, {}))
                tool_result = json.dumps(result_data)
                self.log.info(f"  → {tool_result[:250]}")

            elif tool == "run_scraper":
                name = args.get("name", "")
                cfg  = self.SCRAPERS.get(name)
                if not cfg:
                    tool_result = f"Unknown scraper: {name}"
                else:
                    success, path = self._run_with_retry(name, cfg)
                    if success and path:
                        quality = self._assess_quality(path)
                        collected.append(path)
                        results[name] = {"status": "success", "path": path, **quality}
                        self.memory.remember("successes", {"scraper": name, **quality})
                        # Update semantic knowledge with typical article count
                        self.memory.know(f"{name}.typical_rows", quality["total"])
                        tool_result = (
                            f"Success — {quality['total']} articles, "
                            f"empty={quality['empty_pct']*100:.1f}%, "
                            f"dupes={quality['dupe_pct']*100:.1f}%"
                        )
                        self.log.info(f"  ✅ {name} → {tool_result}")
                    else:
                        # Scraper failed — try fallback ladder automatically
                        self.log.warning(f"  {name} failed — activating fallback")
                        rung, fpath, explanation = self.resolve_fallback(
                            name, cfg,
                            health_status="unknown",
                            find_output_fn=lambda c: c.get("output_csv") if Path(c.get("output_csv","")).exists() else None,
                            count_rows_fn=self._count_rows,
                        )
                        if fpath:
                            collected.append(fpath)
                            results[name] = {
                                "status": "fallback", "fallback_rung": rung,
                                "path": fpath, "explanation": explanation,
                            }
                            tool_result = f"Failed → fallback {rung}: {explanation}"
                            self.log.warning(f"  ⚠️  {name} → {tool_result}")
                        else:
                            results[name] = {"status": "failed"}
                            self.memory.remember("failures", {
                                "scraper": name, "reason": "retries_exhausted_no_fallback"
                            })
                            tool_result = "Failed — no fallback available (non-blocking)"
                            self.log.warning(f"  ⚠️  {name} → {tool_result}")
                    handled.add(name)

            elif tool == "activate_fallback":
                name = args.get("name", "")
                cfg  = self.SCRAPERS.get(name, {})
                rung, fpath, explanation = self.resolve_fallback(
                    name, cfg,
                    health_status="unknown",
                    find_output_fn=lambda c: c.get("output_csv") if Path(c.get("output_csv","")).exists() else None,
                    count_rows_fn=self._count_rows,
                )
                if fpath:
                    collected.append(fpath)
                    results[name] = {
                        "status": "fallback", "fallback_rung": rung,
                        "path": fpath, "explanation": explanation,
                    }
                    tool_result = f"Fallback {rung}: {explanation}"
                else:
                    results[name] = {"status": "skipped", "reason": "no_fallback"}
                    tool_result = "No fallback available — skipping"
                handled.add(name)
                self.log.info(f"  ⚠️  {name} → {tool_result}")

            elif tool == "skip_scraper":
                name   = args.get("name", "")
                reason = args.get("reason", "")
                cfg    = self.SCRAPERS.get(name, {})
                existing = cfg.get("output_csv", "")
                if existing and Path(existing).exists():
                    collected.append(existing)
                    results[name] = {"status": "skipped", "reason": reason, "path": existing}
                    tool_result = f"Skipped — using cache: {self._count_rows(existing)} rows"
                else:
                    results[name] = {"status": "skipped", "reason": reason}
                    tool_result = f"Skipped — no cache"
                handled.add(name)
                self.log.info(f"  ⏭  {name}: {tool_result}")

            elif tool == "finish":
                self.log.info("  LLM called finish() — exiting loop")
                break

            else:
                tool_result = f"Unknown tool '{tool}'"
                self.log.warning(f"  ⚠  {tool_result}")

            messages.append({"role": "assistant", "content": raw})
            messages.append({"role": "user",      "content": f"Tool result: {tool_result}"})

            if handled >= set(self.SCRAPERS.keys()):
                self.log.info("  All scrapers handled — exiting loop")
                break

        # Safety net for any scraper the LLM missed
        for name, cfg in self.SCRAPERS.items():
            if name not in handled and name not in results:
                self.log.warning(f"  Safety net: {name}")
                skip, reason = self._should_skip_fallback(name, cfg)
                if skip:
                    existing = cfg.get("output_csv", "")
                    if existing and Path(existing).exists():
                        collected.append(existing)
                    results[name] = {"status": "skipped", "reason": reason}
                else:
                    success, path = self._run_with_retry(name, cfg)
                    if success and path:
                        quality = self._assess_quality(path)
                        collected.append(path)
                        results[name] = {"status": "success", "path": path, **quality}
                        self.memory.remember("successes", {"scraper": name, **quality})
                    else:
                        results[name] = {"status": "failed"}
                        self.memory.remember("failures", {"scraper": name, "reason": "safety_net_failed"})

        return collected, results

    # ── Tool implementations ───────────────────────────────────────────────────
    def _inspect_scraper(self, name: str, cfg: Dict) -> Dict:
        csv_path      = cfg.get("output_csv", "")
        max_age_hours = cfg.get("max_age_hours", 6)
        min_rows      = cfg.get("min_rows", 10)

        output_info = {"exists": False, "rows": 0, "age_hours": None, "fresh": False}
        if csv_path and Path(csv_path).exists():
            age_h = (time.time() - os.path.getmtime(csv_path)) / 3600
            rows  = self._count_rows(csv_path)
            output_info = {
                "exists":    True,
                "rows":      rows,
                "age_hours": round(age_h, 1),
                "fresh":     age_h < max_age_hours and rows >= min_rows,
            }

        recent_failures = [
            f for f in self.memory.recall("failures", 20)
            if f.get("scraper") == name
        ]
        typical_rows = self.memory.recall_knowledge(f"{name}.typical_rows")

        return {
            "scraper":         name,
            "min_rows":        min_rows,
            "max_age_hours":   max_age_hours,
            "output":          output_info,
            "recent_failures": len(recent_failures),
            "last_failure":    recent_failures[-1] if recent_failures else None,
            "typical_rows":    typical_rows,
            "manual_override": self._find_manual(name) is not None,
        }

    def _should_skip_fallback(self, name: str, cfg: Dict) -> Tuple[bool, str]:
        csv_path      = cfg.get("output_csv", "")
        max_age_hours = cfg.get("max_age_hours", 6)
        if Path(csv_path).exists():
            age_h = (time.time() - os.path.getmtime(csv_path)) / 3600
            rows  = self._count_rows(csv_path)
            if age_h < max_age_hours and rows >= cfg["min_rows"]:
                return True, f"Fresh ({age_h:.1f}h, {rows} rows)"
        recent_fail = [f for f in self.memory.recall("failures", 20) if f.get("scraper") == name]
        if len(recent_fail) >= self.MAX_RECENT_FAILURES:
            return True, f"Failed {len(recent_fail)}x recently"
        return False, "Output missing or stale"

    def _run_with_retry(self, name: str, cfg: Dict) -> Tuple[bool, Optional[str]]:
        backoff = self.BACKOFF_SECONDS
        python  = self._resolve_python(cfg["cwd"])
        self.log.info(f"  python : {python}")
        cmd = [python, cfg["script"]] + cfg.get("args", [])
        for attempt in range(1, self.MAX_RETRIES + 1):
            self.log.info(f"  Attempt {attempt}/{self.MAX_RETRIES}")
            success, _, _ = self._run_subprocess(cmd, cfg["cwd"], cfg.get("timeout", 600))
            if success:
                path = cfg["output_csv"]
                if Path(path).exists():
                    rows = self._count_rows(path)
                    if rows >= cfg["min_rows"]:
                        return True, path
                    self.log.warning(f"  Too few rows: {rows} < {cfg['min_rows']}")
                else:
                    self.log.warning(f"  Output not found: {path}")
            if attempt < self.MAX_RETRIES:
                self.log.info(f"  Retrying in {backoff}s...")
                time.sleep(backoff)
                backoff *= 2
        return False, None

    def _assess_quality(self, path: str) -> Dict:
        try:
            df    = pd.read_csv(path, encoding="utf-8")
            total = len(df)
            empty = df["content"].isna().mean() if "content" in df.columns else 0.0
            dupes = df.duplicated(subset=["title"]).mean() if "title" in df.columns else 0.0
            short = (df["content"].fillna("").str.split().str.len() < 50).mean() \
                    if "content" in df.columns else 0.0
            return {
                "total":     total,
                "empty_pct": round(empty, 3),
                "dupe_pct":  round(dupes, 3),
                "short_pct": round(short, 3),
            }
        except Exception as e:
            self.log.error(f"Quality assessment error: {e}")
            return {"total": 0, "empty_pct": 1.0, "dupe_pct": 0.0, "short_pct": 1.0}

    def _merge_articles(self, paths: List[str]) -> Optional[str]:
        dfs = []
        for path in paths:
            try:
                source = "bnb_blog" if "bnb" in path.lower() else "houni_blog"
                enc    = "latin-1" if "houni" in path.lower() else "utf-8"
                df     = pd.read_csv(path, encoding=enc)
                if "houni" in path.lower():
                    def fix(t):
                        try:    return t.encode("latin-1").decode("utf-8")
                        except: return t
                    df = df.apply(lambda c: c.map(fix) if c.dtype == object else c)
                if "title" in df.columns and "content" in df.columns:
                    df = df[["title", "content"]].assign(source=source)
                    dfs.append(df)
                    self.log.info(f"  Loaded {len(df)} articles from {source}")
            except Exception as e:
                self.log.error(f"Failed to load {path}: {e}")
        if not dfs:
            self.log.warning("No article DataFrames to merge")
            return None
        merged = pd.concat(dfs, ignore_index=True)
        merged.insert(0, "article_id", range(1, len(merged) + 1))
        out = "Data/raw_articles/merged_articles.csv"
        Path(out).parent.mkdir(parents=True, exist_ok=True)
        merged.to_csv(out, index=False)
        self.log.info(f"  Articles merged: {len(merged)} total → {out}")
        return out

print("✅ ArticlesAgent defined (Phase 2 ReAct + Phase 3 Fallback + Semantic Memory)")
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 114b — Phase 3: ScraperHealthAgent + FallbackMixin
# INSERT this cell between Cell 114 (ArticlesAgent) and Cell 115 (ListingsAgent)
# ─────────────────────────────────────────────────────────────────────────────

import glob as _glob
import shutil
import requests as _requests

# ── Scraper health check ──────────────────────────────────────────────────────

class ScraperHealthAgent(BaseAgent):
    """
    Pre-flight agent — runs before ListingsAgent.
    Sends a lightweight HEAD request to each target website and classifies
    the failure mode so ListingsAgent can reason about it:

      healthy      — site responds 200/30x
      blocked      — 403/429 (IP block / rate limit — wait or rotate)
      down         — 5xx or connection error (site issue — retry later)
      unknown      — any other status

    Result stored in state["scraper_health"] for the LLM to read.
    """

    TARGETS = {
        "mubawab"        : "https://www.mubawab.tn",
        "tayara"         : "https://www.tayara.tn",
        "tunisie_annonce": "https://www.tunisie-annonce.com",
    }
    TIMEOUT = 8   # seconds per request

    def __init__(self):
        super().__init__("ScraperHealthAgent")

    def run(self, state: Dict[str, Any]) -> Dict[str, Any]:
        self.log.info("=" * 60)
        self.log.info("ScraperHealthAgent — pre-flight checks")
        self.log.info("=" * 60)

        health: Dict[str, Dict] = {}

        for name, url in self.TARGETS.items():
            self.log.info(f"  Checking {name} ({url}) ...")
            status, detail = self._check(url)
            health[name] = {"status": status, "detail": detail, "url": url}

            icon = {"healthy": "✅", "blocked": "🚫", "down": "❌", "unknown": "⚠️"}.get(status, "?")
            self.log.info(f"  {icon}  {name:<20} → {status}  ({detail})")

        # LLM summary — one sentence of context for downstream agents
        summary_prompt = (
            f"Scraper pre-flight results: {json.dumps(health)}.\n"
            f"Summarize in one sentence what this means for running the scrapers.\n"
            'Return ONLY JSON: {"thought": "...", "summary": "one sentence"}'
        )
        dec     = self.decide(summary_prompt)
        summary = dec.get("summary", "Health check complete.")
        self.log.info(f"\n  LLM summary: {summary}")

        state = self.update_state(state, "scraper_health",         health)
        state = self.update_state(state, "scraper_health_summary", summary)
        return state

    def _check(self, url: str) -> Tuple[str, str]:
        try:
            r = _requests.head(url, timeout=self.TIMEOUT,
                               headers={"User-Agent": "Mozilla/5.0"}, allow_redirects=True)
            code = r.status_code
            if code in (200, 301, 302, 303, 307, 308):
                return "healthy", f"HTTP {code}"
            if code in (403, 429):
                return "blocked", f"HTTP {code} — IP block or rate limit"
            if code >= 500:
                return "down", f"HTTP {code} — server error"
            return "unknown", f"HTTP {code}"
        except _requests.exceptions.ConnectionError:
            return "down", "connection refused"
        except _requests.exceptions.Timeout:
            return "down", f"timeout after {self.TIMEOUT}s"
        except Exception as e:
            return "unknown", str(e)[:80]


print("✅ ScraperHealthAgent defined")


# ── Fallback mixin ────────────────────────────────────────────────────────────

class FallbackMixin:
    """
    Mixin that adds the 4-rung degradation ladder to any agent that runs scrapers.

    Rung 1 — fresh scrape         (normal path)
    Rung 2 — cached output < 48h  (stale but usable)
    Rung 3 — stale output any age (last resort real data)
    Rung 4 — synthetic rows       (unblock pipeline only; stripped before model CSV)

    Usage:
        class ListingsAgent(BaseAgent, FallbackMixin): ...
        # then call self.resolve_fallback(name, cfg, health_status) inside ReAct loop
    """

    MANUAL_OVERRIDE_DIR  = "Data/manual_override"
    SYNTHETIC_SOURCE_TAG = "synthetic_fallback"
    CACHE_SOFT_LIMIT_H   = 48    # rung 2 threshold
    MIN_SYNTHETIC_ROWS   = 5     # rows to generate on rung 4

    def resolve_fallback(
        self,
        name:          str,
        cfg:           Dict,
        health_status: str = "unknown",
        find_output_fn = None,
        count_rows_fn  = None,
    ) -> Tuple[str, Optional[str], str]:
        """
        Walk the degradation ladder and return:
            (rung_label, path_or_None, explanation)

        rung_label one of: "manual", "cache_soft", "cache_stale", "synthetic", "none"
        """
        find   = find_output_fn  or (lambda c: None)
        nrows  = count_rows_fn   or (lambda p: 0)
        log    = getattr(self, "log", _build_logger("FallbackMixin"))

        # ── Manual override (highest priority) ────────────────────────────
        manual_path = self._find_manual(name)
        if manual_path:
            rows = nrows(manual_path)
            log.info(f"  [Fallback] RUNG-MANUAL — {manual_path} ({rows:,} rows)")
            return "manual", manual_path, f"Manual override file found: {rows:,} rows"

        existing = find(cfg)

        # ── Rung 2: cache < 48h ───────────────────────────────────────────
        if existing and Path(existing).exists():
            age_h = (time.time() - os.path.getmtime(existing)) / 3600
            rows  = nrows(existing)
            if age_h < self.CACHE_SOFT_LIMIT_H and rows >= cfg.get("min_rows", 50):
                log.info(f"  [Fallback] RUNG-2 — cache {age_h:.1f}h old, {rows:,} rows")
                return "cache_soft", existing, f"Cached output {age_h:.1f}h old ({rows:,} rows)"

            # ── Rung 3: stale cache any age ───────────────────────────────
            if rows > 0:
                log.warning(f"  [Fallback] RUNG-3 — stale cache {age_h:.1f}h old, {rows:,} rows")
                return "cache_stale", existing, f"Stale cache {age_h:.1f}h old ({rows:,} rows) — quality degraded"

        # ── Rung 4: synthetic rows ────────────────────────────────────────
        # Only generate if site is blocked/down — not if we've never scraped it
        if health_status in ("blocked", "down"):
            synth_path = self._generate_synthetic(name)
            if synth_path:
                log.warning(f"  [Fallback] RUNG-4 — synthetic placeholder at {synth_path}")
                return "synthetic", synth_path, f"Synthetic placeholder rows (site {health_status})"

        log.warning(f"  [Fallback] RUNG-NONE — no fallback available for {name}")
        return "none", None, "No fallback available"

    # ── Helpers ───────────────────────────────────────────────────────────────
    def _find_manual(self, name: str) -> Optional[str]:
        """Check Data/manual_override/{name}_manual.csv (< 7 days old)."""
        path = Path(self.MANUAL_OVERRIDE_DIR) / f"{name}_manual.csv"
        if path.exists():
            age_days = (time.time() - os.path.getmtime(str(path))) / 86400
            if age_days < 7:
                return str(path)
        return None

    def _generate_synthetic(self, name: str) -> Optional[str]:
        """
        Generate a minimal set of clearly-tagged synthetic rows.
        These are placeholder rows only — stripped before the final model CSV.
        Every field is either a realistic Tunisian value or NaN.
        """
        import numpy as _np

        out_dir = Path("Data/raw_listings/synthetic")
        out_dir.mkdir(parents=True, exist_ok=True)
        out_path = str(out_dir / f"{name}_synthetic.csv")

        try:
            govs   = ["Tunis", "Ariana", "Ben Arous", "Sousse", "Sfax", "Nabeul"]
            ptypes = ["Apartment", "House", "Villa", "Land"]
            rows   = []
            for i in range(self.MIN_SYNTHETIC_ROWS):
                rows.append({
                    "listing_id":       f"SYNTH_{name}_{i:04d}",
                    "source":           self.SYNTHETIC_SOURCE_TAG,
                    "title":            f"Synthetic listing {i+1} — pipeline fallback",
                    "property_type":    ptypes[i % len(ptypes)],
                    "transaction_type": "Sale",
                    "price":            _np.nan,
                    "currency":         "TND",
                    "surface_m2":       _np.nan,
                    "governorate":      govs[i % len(govs)],
                    "city":             _np.nan,
                    "description":      f"SYNTHETIC FALLBACK ROW — scraper {name} unavailable.",
                    "image_count":      0,
                    "country":          "Tunisia",
                    "is_synthetic":     True,
                })
            pd.DataFrame(rows).to_csv(out_path, index=False)
            return out_path
        except Exception as e:
            _log = getattr(self, "log", _build_logger("FallbackMixin"))
            _log.error(f"Synthetic generation failed for {name}: {e}")
            return None


print("✅ FallbackMixin defined")
print()
print("Phase 3 components ready:")
print("  ScraperHealthAgent  — pre-flight HTTP checks")
print("  FallbackMixin       — 4-rung degradation ladder")
print("  Manual override dir — Data/manual_override/{name}_manual.csv")
print("  Synthetic rows      — Data/raw_listings/synthetic/{name}_synthetic.csv")
````

<h2 style="color: #87BAC3;">🏠 Agent 2 — Listings Collection Agent</h2>

<h3>🎯 What it does</h3>
<p>Runs the three listing scrapers in sequence. Each uses a checkpoint system — re-running is always safe and only fetches new listings.</p>

<h3>📁 Scraper details</h3>
<ul>
  <li><strong>Mubawab</strong>: <code>cd C:\Mubawab-scrapper && python mubawab_scraper.py</code> → <code>data/mubawab_listings.csv</code></li>
  <li><strong>Tayara</strong>: <code>cd C:\tayara-scraper && python main.py --max-pages 10</code> → <code>data/tayara_scrape/processed/tayara_listings_*.csv</code> (timestamped)</li>
  <li><strong>Tunisie Annonces</strong>: <code>cd C:\Tunisie-Annonces-scrapper && python main.py</code> → <code>output/tunisie_annonce_listings.csv</code></li>
</ul>

<h3>🧠 Decision logic (deterministic)</h3>
<ul>
  <li>Check long-term memory: ≥3 recent failures → skip that scraper</li>
  <li>Check existing output rows ≥ min threshold → skip re-scraping</li>
  <li>Tayara output is timestamped → find latest file in <code>data/tayara_scrape/processed/</code></li>
  <li>Exponential backoff retry: 10s → 20s → 40s</li>
</ul>

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 115 — ListingsAgent  (Phase 2 + Phase 3 combined — full replacement)
# ─────────────────────────────────────────────────────────────────────────────

class ListingsAgent(BaseAgent, FallbackMixin):
    """
    Runs Mubawab, Tayara, and Tunisie Annonces scrapers.

    Phase 2: ReAct loop — LLM decides run/skip order with reasoning.
    Phase 3: FallbackMixin — 4-rung degradation ladder when scraper fails.
             ScraperHealthAgent results inform the LLM and fallback choices.
    """

    SCRAPERS = {
        "mubawab": {
            "cwd"        : r"C:\Mubawab-scrapper",
            "script"     : "mubawab_scraper.py",
            "args"       : [],
            "output_csv" : r"C:\Mubawab-scrapper\data\mubawab_listings.csv",
            "min_rows"   : 100,
            "timeout"    : 7200,
        },
        "tayara": {
            "cwd"        : r"C:\tayara-scraper",
            "script"     : "main.py",
            "args"       : ["--max-pages", "13"],
            "output_dir" : r"C:\tayara-scraper\data\tayara_scrape\processed",
            "output_glob": "tayara_listings_*.csv",
            "min_rows"   : 100,
            "timeout"    : 3600,
        },
        "tunisie_annonce": {
            "cwd"        : r"C:\Tunisie-Annonces-scrapper",
            "script"     : "main.py",
            "args"       : [],
            "output_csv" : r"C:\Tunisie-Annonces-scrapper\output\tunisie_annonce_listings.csv",
            "min_rows"   : 100,
            "timeout"    : 7200,
        },
    }
    MAX_OUTPUT_AGE_HOURS = 6
    MAX_RECENT_FAILURES  = 3
    MAX_RETRIES          = 3
    BASE_BACKOFF         = 10
    MAX_REACT_STEPS      = 14

    def __init__(self):
        super().__init__("ListingsAgent")

    # ── Public entry point ────────────────────────────────────────────────────
    def run(self, state: Dict[str, Any]) -> Dict[str, Any]:
        self.log.info("=" * 60)
        self.log.info("ListingsAgent — starting (ReAct + Fallback)")
        self.log.info("=" * 60)
        self.memory.clear()

        # Pull health results from ScraperHealthAgent (may be absent if skipped)
        health: Dict = state.get("scraper_health", {})
        if health:
            self.log.info(f"Health context: {state.get('scraper_health_summary','')}")
        else:
            self.log.info("No health data available — running without pre-flight context")

        collected, results = self._react_loop(health)

        # Strip synthetic rows from collected paths — they must not reach
        # StandardizationAgent unless every real source failed
        real_collected   = [p for p in collected if self.SYNTHETIC_SOURCE_TAG not in p]
        synth_collected  = [p for p in collected if self.SYNTHETIC_SOURCE_TAG in p]

        if real_collected:
            final_collected = real_collected
            if synth_collected:
                self.log.warning(
                    f"Dropping {len(synth_collected)} synthetic file(s) — real data available"
                )
        else:
            # All real scrapers failed — pass synthetic through so pipeline can continue
            final_collected = synth_collected
            if synth_collected:
                self.log.warning(
                    "ALL real scrapers failed — passing synthetic placeholder data downstream. "
                    "Final model CSV will exclude synthetic rows."
                )

        total_rows = sum(self._count_rows(p) for p in final_collected)
        state = self.update_state(state, "listings_raw_paths",   final_collected)
        state = self.update_state(state, "listings_results",     results)
        state = self.update_state(state, "listings_total_rows",  total_rows)
        state = self.update_state(state, "listings_has_synthetic",
                                  len(synth_collected) > 0 and not real_collected)

        self.log.info(f"\nListingsAgent done — {len(final_collected)} sources, {total_rows:,} total rows")
        for name, r in results.items():
            rung = r.get("fallback_rung", "")
            rung_str = f" [{rung}]" if rung else ""
            self.log.info(f"  {name:<22} → {r['status']}{rung_str}")
        return state

    # ── ReAct loop ────────────────────────────────────────────────────────────
    def _react_loop(self, health: Dict) -> Tuple[List[str], Dict]:
        collected: List[str] = []
        results:   Dict      = {}
        handled:   set       = set()

        scraper_summary = {
            name: {
                **self._inspect_scraper(name, cfg),
                "site_health": health.get(name, {}).get("status", "unknown"),
                "site_detail": health.get(name, {}).get("detail", ""),
            }
            for name, cfg in self.SCRAPERS.items()
        }

        system_context = (
            "You are ListingsAgent for LensEstate, a Tunisian real estate platform.\n"
            "Manage three listing scrapers: mubawab, tayara, tunisie_annonce.\n\n"
            "Available tools (call ONE per step):\n"
            "  inspect_scraper(name)           — full state: age, rows, failures, site health\n"
            "  run_scraper(name)               — execute the scraper subprocess\n"
            "  skip_scraper(name, reason)      — skip and use best available cache\n"
            "  activate_fallback(name, reason) — site blocked/down: trigger degradation ladder\n"
            "  finish()                        — all scrapers handled\n\n"
            "Decision rules:\n"
            "  SKIP if output fresh (< 6h) AND rows >= min_rows\n"
            "  SKIP if recent_failures >= 3 AND no valid output\n"
            "  RUN  if output stale/missing AND site_health is 'healthy'\n"
            "  ACTIVATE_FALLBACK if site_health is 'blocked' or 'down'\n"
            "  Think about order — start with the most reliable scraper\n\n"
            f"Current states:\n{json.dumps(scraper_summary, indent=2)}\n\n"
            'Reply ONLY with JSON: {"thought": "reasoning with specific numbers", '
            '"tool": "tool_name", "args": {"key": "value"}}'
        )

        messages = [{"role": "user", "content": system_context}]

        for step in range(self.MAX_REACT_STEPS):
            self.log.info(f"\n  [ReAct step {step+1}/{self.MAX_REACT_STEPS}]")

            recent = messages[-4:] if len(messages) > 4 else messages
            prompt = "\n---\n".join(m["content"] for m in recent)

            raw    = call_llm(prompt, max_tokens=512)
            parsed = parse_llm_json(raw)

            thought = parsed.get("thought", "")
            tool    = parsed.get("tool", "finish")
            args    = parsed.get("args", {})

            self.log.info(f"  thought : {thought[:220]}")
            self.log.info(f"  tool    : {tool}  args={args}")

            # ── Tool dispatch ─────────────────────────────────────────────
            if tool == "inspect_scraper":
                name        = args.get("name", "")
                result_data = {
                    **self._inspect_scraper(name, self.SCRAPERS.get(name, {})),
                    "site_health": health.get(name, {}).get("status", "unknown"),
                }
                tool_result = json.dumps(result_data)
                self.log.info(f"  → {tool_result[:280]}")

            elif tool == "run_scraper":
                name = args.get("name", "")
                cfg  = self.SCRAPERS.get(name)
                if not cfg:
                    tool_result = f"Unknown scraper: {name}"
                else:
                    success, path = self._run_with_retry(name, cfg)
                    if success and path:
                        rows = self._count_rows(path)
                        collected.append(path)
                        results[name] = {"status": "success", "path": path, "rows": rows}
                        self.memory.remember("successes", {"scraper": name, "rows": rows})
                        tool_result = f"Success — {rows:,} rows"
                        self.log.info(f"  ✅ {name} → {rows:,} rows")
                    else:
                        # Scraper failed — immediately try fallback
                        self.log.warning(f"  {name} failed — activating fallback ladder")
                        h_status = health.get(name, {}).get("status", "unknown")
                        rung, fpath, explanation = self.resolve_fallback(
                            name, cfg, h_status,
                            find_output_fn=self._find_output,
                            count_rows_fn=self._count_rows,
                        )
                        if fpath:
                            collected.append(fpath)
                            rows = self._count_rows(fpath)
                            results[name] = {
                                "status":        "fallback",
                                "fallback_rung": rung,
                                "path":          fpath,
                                "rows":          rows,
                                "explanation":   explanation,
                            }
                            self.memory.remember("failures", {
                                "scraper": name, "reason": "run_failed_fallback_activated",
                                "rung": rung,
                            })
                            tool_result = f"Failed → fallback {rung}: {explanation}"
                            self.log.warning(f"  ⚠️  {name} → {tool_result}")
                        else:
                            results[name] = {"status": "failed"}
                            self.memory.remember("failures", {
                                "scraper": name, "reason": "retries_exhausted_no_fallback",
                            })
                            tool_result = f"Failed — all retries exhausted, no fallback available"
                            self.log.error(f"  ❌ {name} → {tool_result}")
                    handled.add(name)

            elif tool == "activate_fallback":
                name   = args.get("name", "")
                reason = args.get("reason", "LLM requested fallback")
                cfg    = self.SCRAPERS.get(name, {})
                h_status = health.get(name, {}).get("status", "unknown")
                rung, fpath, explanation = self.resolve_fallback(
                    name, cfg, h_status,
                    find_output_fn=self._find_output,
                    count_rows_fn=self._count_rows,
                )
                if fpath:
                    collected.append(fpath)
                    rows = self._count_rows(fpath)
                    results[name] = {
                        "status":        "fallback",
                        "fallback_rung": rung,
                        "path":          fpath,
                        "rows":          rows,
                        "explanation":   explanation,
                        "llm_reason":    reason,
                    }
                    tool_result = f"Fallback {rung} activated: {explanation}"
                    self.log.warning(f"  ⚠️  {name} → {tool_result}")
                else:
                    results[name] = {"status": "failed", "reason": "no_fallback_available"}
                    tool_result = "No fallback available — source will be missing"
                    self.log.error(f"  ❌ {name} → {tool_result}")
                handled.add(name)

            elif tool == "skip_scraper":
                name   = args.get("name", "")
                reason = args.get("reason", "no reason given")
                cfg    = self.SCRAPERS.get(name, {})
                existing = self._find_output(cfg)
                if existing:
                    collected.append(existing)
                    rows = self._count_rows(existing)
                    results[name] = {"status": "skipped", "reason": reason,
                                     "path": existing, "rows": rows}
                    tool_result = f"Skipped — using cache: {rows:,} rows"
                else:
                    results[name] = {"status": "skipped", "reason": reason}
                    tool_result = f"Skipped — no cache"
                handled.add(name)
                self.log.info(f"  ⏭  {name}: {tool_result}")

            elif tool == "finish":
                self.log.info("  LLM called finish() — exiting ReAct loop")
                break

            else:
                tool_result = f"Unknown tool '{tool}'"
                self.log.warning(f"  ⚠  {tool_result}")

            messages.append({"role": "assistant", "content": raw})
            messages.append({"role": "user",      "content": f"Tool result: {tool_result}"})

            if handled >= set(self.SCRAPERS.keys()):
                self.log.info("  All scrapers handled — exiting ReAct loop")
                break

        # Safety net
        for name, cfg in self.SCRAPERS.items():
            if name not in handled and name not in results:
                self.log.warning(f"  Safety net: handling missed scraper {name}")
                h_status = health.get(name, {}).get("status", "unknown")
                skip, reason = self._should_skip_fallback(name, cfg)
                if skip:
                    existing = self._find_output(cfg)
                    if existing:
                        collected.append(existing)
                    results[name] = {"status": "skipped", "reason": reason}
                else:
                    success, path = self._run_with_retry(name, cfg)
                    if success and path:
                        rows = self._count_rows(path)
                        collected.append(path)
                        results[name] = {"status": "success", "path": path, "rows": rows}
                        self.memory.remember("successes", {"scraper": name, "rows": rows})
                    else:
                        rung, fpath, explanation = self.resolve_fallback(
                            name, cfg, h_status,
                            find_output_fn=self._find_output,
                            count_rows_fn=self._count_rows,
                        )
                        if fpath:
                            collected.append(fpath)
                            results[name] = {"status": "fallback", "fallback_rung": rung,
                                             "path": fpath, "explanation": explanation}
                        else:
                            results[name] = {"status": "failed"}
                        self.memory.remember("failures", {"scraper": name, "reason": "safety_net"})

        return collected, results

    # ── Tool implementations (same as Phase 2) ────────────────────────────────
    def _inspect_scraper(self, name: str, cfg: Dict) -> Dict:
        existing = self._find_output(cfg)
        min_rows = cfg.get("min_rows", 100)

        output_info = {"exists": False, "rows": 0, "age_hours": None, "fresh": False}
        if existing and Path(existing).exists():
            age_h = (time.time() - os.path.getmtime(existing)) / 3600
            rows  = self._count_rows(existing)
            output_info = {
                "exists":    True,
                "path":      existing,
                "rows":      rows,
                "age_hours": round(age_h, 1),
                "fresh":     age_h < self.MAX_OUTPUT_AGE_HOURS and rows >= min_rows,
            }

        recent_failures  = [f for f in self.memory.recall("failures",  20) if f.get("scraper") == name]
        recent_successes = [s for s in self.memory.recall("successes", 20) if s.get("scraper") == name]

        # Check manual override
        manual = self._find_manual(name)

        return {
            "scraper":          name,
            "min_rows":         min_rows,
            "max_age_hours":    self.MAX_OUTPUT_AGE_HOURS,
            "output":           output_info,
            "recent_failures":  len(recent_failures),
            "recent_successes": len(recent_successes),
            "last_failure":     recent_failures[-1]  if recent_failures  else None,
            "last_success":     recent_successes[-1] if recent_successes else None,
            "manual_override":  manual is not None,
        }

    def _should_skip_fallback(self, name: str, cfg: Dict) -> Tuple[bool, str]:
        existing = self._find_output(cfg)
        if existing and Path(existing).exists():
            age_h = (time.time() - os.path.getmtime(existing)) / 3600
            rows  = self._count_rows(existing)
            if age_h < self.MAX_OUTPUT_AGE_HOURS and rows >= cfg["min_rows"]:
                return True, f"Fresh output ({age_h:.1f}h, {rows:,} rows)"
        recent_fail = [f for f in self.memory.recall("failures", 20) if f.get("scraper") == name]
        if len(recent_fail) >= self.MAX_RECENT_FAILURES:
            return True, f"Failed {len(recent_fail)}x recently"
        return False, "Output missing or stale"

    def _run_with_retry(self, name: str, cfg: Dict) -> Tuple[bool, Optional[str]]:
        backoff = self.BASE_BACKOFF
        python  = self._resolve_python(cfg["cwd"])
        self.log.info(f"  python : {python}")
        cmd = [python, cfg["script"]] + cfg.get("args", [])
        for attempt in range(1, self.MAX_RETRIES + 1):
            self.log.info(f"  Attempt {attempt}/{self.MAX_RETRIES}")
            ok, _, _ = self._run_subprocess(cmd, cfg["cwd"], cfg.get("timeout", 3600))
            if ok:
                path = self._find_output(cfg)
                if path:
                    rows = self._count_rows(path)
                    if rows >= cfg["min_rows"]:
                        return True, path
                    self.log.warning(f"  Too few rows: {rows} < {cfg['min_rows']}")
                else:
                    self.log.warning("  No output file found after scrape")
            if attempt < self.MAX_RETRIES:
                self.log.info(f"  Retrying in {backoff}s...")
                time.sleep(backoff)
                backoff *= 2
        return False, None

    def _find_output(self, cfg: Dict) -> Optional[str]:
        if "output_csv" in cfg and Path(cfg["output_csv"]).exists():
            return cfg["output_csv"]
        if "output_dir" in cfg and "output_glob" in cfg:
            files = _glob.glob(os.path.join(cfg["output_dir"], cfg["output_glob"]))
            if files:
                return max(files, key=os.path.getmtime)
        return None

print("✅ ListingsAgent defined (Phase 2 ReAct + Phase 3 Fallback)")
````

<h2 style="color: #87BAC3;">📐 Agent 3 — Standardization Agent</h2>

<h3>🎯 What it does</h3>
<p>Reads raw CSV paths from shared state, applies per-source schema normalization, injects missing columns with type-aware defaults, and merges into <code>merged_listings.csv</code>.</p>

<h3>🧠 Decision logic (deterministic — no LLM)</h3>
<p>Routes each path to the correct standardizer based on keywords in the path (<code>mubawab</code>, <code>tayara</code>, <code>tunisie</code>). Records before/after row counts and source distribution in long-term memory for the Orchestrator to compare.</p>

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 116 — Schema constants + StandardizationAgent  (full replacement)
# ─────────────────────────────────────────────────────────────────────────────

# ── Shared schema (unchanged) ─────────────────────────────────────────────────
LISTINGS_SCHEMA = {
    "listing_id":(object,None),"listing_url":(object,None),"source":(object,None),
    "title":(object,None),"property_type":(object,None),"transaction_type":(object,None),
    "price":(float,np.nan),"currency":(object,"TND"),"surface_m2":(object,None),
    "rooms":(float,np.nan),"total_rooms":(float,np.nan),"bedrooms":(object,None),
    "bathrooms":(float,np.nan),"floor":(float,np.nan),"total_floors":(float,np.nan),
    "country":(object,"Tunisia"),"governorate":(object,None),"city":(object,None),
    "description":(object,None),"image_count":(int,0),"image_urls":(object,None),
    "date_posted":(object,None),"last_updated":(float,np.nan),"agency_name":(object,None),
    "seller_type":(object,None),"scrape_timestamp":(object,None),
    "has_elevator":(object,None),"has_basement":(bool,False),"has_heating":(object,None),
    "has_air_conditioning":(object,None),"has_garden":(bool,False),
    "has_furniture":(bool,False),"has_parking":(object,None),"has_pool":(bool,False),
    "has_security":(object,None),"has_standing":(bool,False),
    "has_terrace":(bool,False),"has_sea_view":(bool,False),
}
TX_MAP = {"Sale":"Sale","sale":"Sale","À Vendre":"Sale","Vente":"Sale",
          "rent":"Rent","Rent":"Rent","À Louer":"Rent","Location":"Rent"}
MUBAWAB_RENAME = {
    "has_ascenseur":"has_elevator","has_cave":"has_basement",
    "has_chauffage":"has_heating","has_climatisation":"has_air_conditioning",
    "has_jardin":"has_garden","has_meuble":"has_furniture",
    "has_piscine":"has_pool","has_gardiennage":"has_security",
    "has_terrasse":"has_terrace","has_vue_mer":"has_sea_view",
}
TAYARA_RENAME = {
    "surface":"surface_m2","criteria_ascenseur":"has_elevator",
    "criteria_climatisation":"has_air_conditioning","criteria_chauffage":"has_heating",
    "criteria_parking":"has_parking","criteria_gardiennage":"has_security",
}

def _align(df: pd.DataFrame) -> pd.DataFrame:
    for col, (dtype, default) in LISTINGS_SCHEMA.items():
        if col not in df.columns:
            df[col] = (False if dtype == bool
                       else np.nan if dtype == float
                       else 0 if dtype == int
                       else default)
    return df[
        [c for c in LISTINGS_SCHEMA] +
        [c for c in df.columns if c not in LISTINGS_SCHEMA]
    ]


class StandardizationAgent(BaseAgent):
    """
    Aligns raw CSVs from each scraper to the shared LISTINGS_SCHEMA.

    New in Phase 2: LLM schema repair.
    When a file's columns don't match known patterns, the LLM inspects
    a sample and proposes the correct rename mapping — so the pipeline
    adapts to minor scraper schema drift without manual code changes.
    """

    def __init__(self):
        super().__init__("StandardizationAgent")

    # ── Public entry point ────────────────────────────────────────────────────
    def run(self, state: Dict[str, Any]) -> Dict[str, Any]:
        self.log.info("=" * 60)
        self.log.info("StandardizationAgent — starting")
        self.log.info("=" * 60)
        self.memory.clear()

        paths = state.get("listings_raw_paths", [])
        self.log.info(f"Input paths ({len(paths)}):")
        for p in paths:
            self.log.info(f"  → {p}")

        dfs = []
        for path in paths:
            self.log.info(f"\nStandardizing: {path}")
            df = self._standardize(path)
            if df is not None and not df.empty:
                dfs.append(df)
                source = df["source"].iloc[0] if "source" in df.columns else "?"
                self.log.info(f"  ✅ {len(df):,} rows | source={source}")
            else:
                self.log.warning("  ⚠️  Empty result — skipping")

        if not dfs:
            self.log.error("No data to merge — all sources failed standardization")
            return self.update_state(state, "standardization_status", "failed")

        merged = pd.concat(dfs, ignore_index=True)
        if "id" in merged.columns:
            merged = merged.drop(columns=["id"])
        merged.insert(0, "id", range(1, len(merged) + 1))

        out = "Data/raw_listings/merged_listings.csv"
        Path(out).parent.mkdir(parents=True, exist_ok=True)
        merged.to_csv(out, index=False)

        src_dist = merged["source"].value_counts().to_dict() if "source" in merged.columns else {}
        self.log.info(f"\n{'─'*50}")
        self.log.info(f"Merged: {len(merged):,} rows, {merged.shape[1]} cols → {out}")
        for src, cnt in src_dist.items():
            self.log.info(f"  {src:<25} {cnt:>7,}")
        self.log.info(f"{'─'*50}")

        self.memory.remember("quality", {"rows": len(merged), "cols": merged.shape[1], "sources": src_dist})
        state = self.update_state(state, "merged_listings_path",   out)
        state = self.update_state(state, "standardization_status", "success")
        state = self.update_state(state, "standardization_report", {"rows": len(merged), "sources": src_dist})
        return state

    # ── Per-file standardization ───────────────────────────────────────────────
    def _standardize(self, path: str) -> Optional[pd.DataFrame]:
        try:
            df = pd.read_csv(path, low_memory=False)
            pl = path.lower()
            self.log.debug(f"  Raw shape: {df.shape} | cols: {df.columns.tolist()[:8]}...")

            if "mubawab" in pl:
                df["source"]           = "mubawab"
                df["transaction_type"] = "Sale"
                drop = ["price_type", "district", "language", "characteristics", "has_bathrooms"]
                df = df.drop(columns=[c for c in drop if c in df.columns])
                df = df.rename(columns={k: v for k, v in MUBAWAB_RENAME.items() if k in df.columns})

            elif "tayara" in pl:
                df["source"]  = "tayara"
                df["country"] = "Tunisia"
                drop = ["price_raw", "location_raw", "category",
                        "criteria_celliers", "criteria_parking_sous_sol"]
                df = df.drop(columns=[c for c in drop if c in df.columns])
                df = df.rename(columns={k: v for k, v in TAYARA_RENAME.items() if k in df.columns})
                if "has_bathrooms" in df.columns:
                    df["bathrooms"] = df["has_bathrooms"]
                    df = df.drop(columns=["has_bathrooms"])

            elif "tunisie" in pl:
                df["source"] = "tunisie_annonce"
                if "type_transaction" in df.columns:
                    df["transaction_type"] = df["type_transaction"].map(TX_MAP).fillna(df["type_transaction"])
                    df = df.drop(columns=["type_transaction"])
                if "region" in df.columns:
                    split = df["region"].str.split(" - ", n=1, expand=True)
                    df["governorate"] = split[0] if 0 in split.columns else None
                    df["city"]        = split[1] if 1 in split.columns else None
                    df = df.drop(columns=["region"])

            else:
                # ── Unknown source: ask LLM to propose column mapping ─────
                self.log.info("  Unknown source pattern — invoking LLM schema repair")
                df = self._llm_schema_repair(path, df)

            return _align(df)

        except Exception as e:
            self.log.error(f"Standardization error [{path}]: {e}")
            return None

    # ── LLM schema repair (new) ───────────────────────────────────────────────
    def _llm_schema_repair(self, path: str, df: pd.DataFrame) -> pd.DataFrame:
        """
        Called when a CSV doesn't match any known scraper pattern.
        The LLM sees the column names + 3 sample rows and proposes:
          1. A rename mapping  (raw_col → schema_col)
          2. A source label
          3. Any columns to drop

        This lets the pipeline adapt to new scrapers or renamed columns
        without requiring a code change.
        """
        sample_rows = df.head(3).to_dict(orient="records")
        known_cols  = list(LISTINGS_SCHEMA.keys())

        prompt = (
            f"A new real estate CSV from Tunisia has been loaded with unknown column names.\n"
            f"File: {path}\n\n"
            f"Raw columns: {df.columns.tolist()}\n\n"
            f"Sample rows (first 3):\n{json.dumps(sample_rows, ensure_ascii=False, default=str)}\n\n"
            f"Target schema columns: {known_cols}\n\n"
            "Map the raw columns to the target schema where possible.\n"
            "Return ONLY this JSON:\n"
            "{\n"
            '  "thought": "explain your mapping reasoning",\n'
            '  "source_label": "name for this scraper (e.g. new_source)",\n'
            '  "rename": {"raw_col": "schema_col", ...},\n'
            '  "drop": ["col_to_drop", ...]\n'
            "}"
        )

        result = self.decide(prompt)

        thought      = result.get("thought", "")
        source_label = result.get("source_label", "unknown_source")
        rename_map   = result.get("rename", {})
        drop_cols    = result.get("drop", [])

        self.log.info(f"  LLM schema repair thought: {thought[:200]}")
        self.log.info(f"  source_label : {source_label}")
        self.log.info(f"  rename map   : {rename_map}")
        self.log.info(f"  drop cols    : {drop_cols}")

        # Persist the inferred mapping for human review
        self.memory.remember("schema_repairs", {
            "path":         path,
            "source_label": source_label,
            "rename":       rename_map,
            "drop":         drop_cols,
            "thought":      thought,
        })

        df = df.drop(columns=[c for c in drop_cols if c in df.columns])
        df = df.rename(columns={k: v for k, v in rename_map.items() if k in df.columns})
        df["source"] = source_label
        return df

print("✅ StandardizationAgent defined (LLM schema repair)")
````

<h2 style="color: #87BAC3;">🧹 Agent 4 — Cleaning Agent</h2>

<h3>🎯 What it does</h3>
<p>Runs the 15-step deterministic cleaning pipeline, then uses <strong>Qwen2.5-0.5B</strong> to flag suspicious listings. Suspicious listings are tagged <code>llm_suspicious=True</code> but never auto-dropped — they are flagged for human review.</p>

<h3>🧠 LLM usage</h3>
<p>Samples 30 listings and asks the LLM: <em>"Which of these look suspicious?"</em> — placeholder text, clearly wrong prices, copy-paste spam. The LLM returns <code>{"suspicious_ids": [...]}</code>. This detects issues that deterministic rules miss.</p>

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 117 — Geography constants + CleaningAgent  (full replacement)
# ─────────────────────────────────────────────────────────────────────────────

CITY_TO_GOV = {
    "Tunis":"Tunis","La Marsa":"Tunis","Gammarth":"Tunis","Ain Zaghouan":"Tunis",
    "La Soukra":"Tunis","Ennasr":"Tunis","El Kram":"Tunis","El Ghazela":"Tunis",
    "Ariana":"Ariana","Raoued":"Ariana","Mnihla":"Ariana","Ettadhamen":"Ariana",
    "Ben Arous":"Ben Arous","Rades":"Ben Arous","Boumhel":"Ben Arous",
    "El Mourouj":"Ben Arous","Mégrine":"Ben Arous","Hammam Lif":"Ben Arous",
    "Manouba":"Manouba","Oued Ellil":"Manouba","Douar Hicher":"Manouba",
    "Nabeul":"Nabeul","Hammamet":"Nabeul","Kelibia":"Nabeul","Korba":"Nabeul",
    "Sousse":"Sousse","Hammam Sousse":"Sousse","Sahloul":"Sousse","Akouda":"Sousse",
    "Monastir":"Monastir","Sfax":"Sfax","Gremda":"Sfax","Sakiet Ezzit":"Sfax",
    "Mahdia":"Mahdia","Kairouan":"Kairouan","Bizerte":"Bizerte",
    "Djerba":"Médenine","Zarzis":"Médenine","Midoun":"Médenine",
    "Gabès":"Gabès","Gafsa":"Gafsa","Tozeur":"Tozeur","Tataouine":"Tataouine",
}
PT_MAP = {
    "Appartement":"Apartment","appartement":"Apartment",
    "Maison":"House","maison":"House",
    "Terrain":"Land","terrain":"Land",
    "others":"Other","Autre":"Other",
    "Local Commercial":"Commercial","Commerce":"Commercial",
    "Villa":"Villa","villa":"Villa",
    "Bureau":"Office","bureau":"Office",
    "Immeuble":"Building","Ferme":"Farm",
    "Duplex":"Duplex","Studio":"Studio",
}
SORTED_CITIES = sorted(CITY_TO_GOV.keys(), key=len, reverse=True)


class CleaningAgent(BaseAgent):
    """
    Cleans and prepares the merged listings for modelling.

    Phase 2 upgrade: the LLM anomaly detection is now more robust —
    it runs in batches, produces structured flags with reasons per listing,
    and persists its findings to memory for audit.
    """

    DROP_COLS = ["total_floors", "last_updated", "floor",
                 "scrape_timestamp", "date_posted", "agency_name"]

    SAMPLE_SIZE       = 40   # rows sent to LLM per batch
    MAX_BATCHES       = 3    # cap total LLM calls for anomaly detection
    SUSPICIOUS_PROMPT = (
        "You are a data quality agent reviewing real estate listings from Tunisia.\n"
        "Identify listings that are SUSPICIOUS — meaning any of:\n"
        "  - Placeholder or test data (description says 'test', 'lorem ipsum', etc.)\n"
        "  - Price is clearly wrong: < 5,000 TND for any property, or > 50,000,000 TND\n"
        "  - Surface is impossible: < 10 m² or > 50,000 m²\n"
        "  - Description is in a language unrelated to Tunisian real estate (e.g. pure English ad copy)\n"
        "  - Description is copy-pasted noise or irrelevant content\n\n"
        "For each suspicious listing give a brief reason.\n\n"
        "Return ONLY this JSON:\n"
        "{\n"
        '  "thought": "overall observations about this batch",\n'
        '  "flagged": [\n'
        '    {"id": 123, "reason": "price=1 TND, clearly placeholder"},\n'
        '    ...\n'
        '  ]\n'
        "}\n"
        "Empty list if nothing suspicious.\n\n"
        "LISTINGS:\n"
    )

    def __init__(self):
        super().__init__("CleaningAgent")

    # ── Public entry point ────────────────────────────────────────────────────
    def run(self, state: Dict[str, Any]) -> Dict[str, Any]:
        self.log.info("=" * 60)
        self.log.info("CleaningAgent — starting")
        self.log.info("=" * 60)
        self.memory.clear()

        path = state.get("merged_listings_path", "")
        if not path or not Path(path).exists():
            self.log.error(f"Merged listings not found: {path}")
            return self.update_state(state, "cleaning_status", "failed")

        df = pd.read_csv(path, low_memory=False)
        self.log.info(f"Loaded: {len(df):,} rows × {df.shape[1]} cols")
        initial = len(df)

        df = self._pipeline(df)

        # ── LLM anomaly detection (upgraded) ─────────────────────────────
        self.log.info("\nRunning LLM anomaly detection...")
        flagged = self._flag_suspicious(df)
        # Build lookup: id → reason
        flagged_map = {item["id"]: item["reason"] for item in flagged}

        df["llm_suspicious"]        = df["id"].isin(flagged_map.keys()) if "id" in df.columns else False
        df["llm_suspicious_reason"] = df["id"].map(flagged_map).fillna("") if "id" in df.columns else ""

        n_flagged = df["llm_suspicious"].sum()
        self.log.info(f"LLM flagged {n_flagged} suspicious listings (tagged, not dropped)")
        if flagged:
            for item in flagged[:5]:
                self.log.info(f"  ID {item['id']}: {item['reason']}")
            if len(flagged) > 5:
                self.log.info(f"  ... and {len(flagged)-5} more (see logs/pipeline.log)")

        # Persist flagged list for audit
        self.memory.remember("suspicious_flags", {
            "total_flagged": n_flagged,
            "flagged_ids":   [f["id"] for f in flagged],
            "details":       flagged,
        })

        # ── Save ──────────────────────────────────────────────────────────
        out = "Data/raw_listings/merged_listings_model_ready.csv"
        Path(out).parent.mkdir(parents=True, exist_ok=True)
        df.to_csv(out, index=False)

        null_rates = {
            col: round(df[col].isna().mean(), 3)
            for col in ["price", "surface_m2", "governorate", "property_type"]
            if col in df.columns
        }
        report = {
            "initial":          initial,
            "final":            len(df),
            "removed":          initial - len(df),
            "suspicious":       int(n_flagged),
            "null_rate":        null_rates,
            "duplicates_removed": initial - len(df),   # proxy
        }

        self.log.info(f"\n{'─'*50}")
        self.log.info(f"Cleaning summary:")
        self.log.info(f"  Initial rows : {initial:,}")
        self.log.info(f"  Final rows   : {len(df):,}")
        self.log.info(f"  Removed      : {initial-len(df):,}")
        self.log.info(f"  Suspicious   : {n_flagged}")
        self.log.info(f"  Null rates   : {null_rates}")
        self.log.info(f"  Output       : {out}")
        self.log.info(f"{'─'*50}")

        self.memory.remember("quality", report)
        state = self.update_state(state, "model_ready_path", out)
        state = self.update_state(state, "cleaning_status",  "success")
        state = self.update_state(state, "cleaning_report",  report)
        state = self.update_state(state, "final_row_count",  len(df))
        return state

    # ── Cleaning pipeline (unchanged logic, same steps) ───────────────────────
    def _pipeline(self, df: pd.DataFrame) -> pd.DataFrame:
        steps = []
        def _step(name, n):
            steps.append((name, n))
            self.log.info(f"  [{name}] {n:,} rows")

        df["source"]           = df["source"].str.lower().str.strip()
        df["transaction_type"] = df["transaction_type"].map(TX_MAP).fillna(df["transaction_type"])
        df["property_type"]    = df["property_type"].map(PT_MAP).fillna(df["property_type"])
        self.log.info("\nCleaning pipeline:")
        _step("normalize", len(df))

        def fill_gov(row):
            if pd.isna(row.get("governorate")) and pd.notna(row.get("city")):
                return CITY_TO_GOV.get(str(row["city"]).strip(), row.get("governorate"))
            return row.get("governorate")
        df["governorate"] = df.apply(fill_gov, axis=1)

        for idx, row in df[df["governorate"].isna() & df["city"].isna()].iterrows():
            text = f"{row.get('title','')} {row.get('description','')}".lower()
            for city in SORTED_CITIES:
                if re.search(r'\b' + re.escape(city.lower()) + r'\b', text):
                    df.at[idx, "city"]        = city
                    df.at[idx, "governorate"] = CITY_TO_GOV[city]
                    break
        _step("gov_recovery", len(df))

        df = df[df["transaction_type"] == "Sale"].reset_index(drop=True)
        _step("remove_rent", len(df))

        df = df.drop_duplicates(subset=["description"], keep="first").reset_index(drop=True)
        _step("dedup_description", len(df))

        df = df.drop(columns=[c for c in self.DROP_COLS if c in df.columns])
        df = df[df["description"].notna()].reset_index(drop=True)
        _step("drop_null_desc", len(df))

        df["description_word_count"] = df["description"].str.split().str.len()
        df["has_images"]  = (df["image_count"] > 0) if "image_count" in df.columns else False
        df["image_urls"]  = df["image_urls"].fillna("") if "image_urls" in df.columns else ""

        for col in ["has_elevator","has_heating","has_air_conditioning","has_parking","has_security"]:
            if col in df.columns:
                df[col] = df[col].fillna(False).map(
                    lambda x: False if str(x).lower() in ("false","0","nan","") else bool(x))

        def enc(v):
            if isinstance(v, str):
                v = v.lower()
                if any(k in v for k in ["agence","agency","pro"]): return 1
                if any(k in v for k in ["particulier","individual"]): return 0
            return -1
        df["is_agency"] = df["seller_type"].apply(enc) if "seller_type" in df.columns else -1

        df["price"] = pd.to_numeric(df["price"], errors="coerce")
        df["price"] = df["price"].where(df["price"] > 0)
        df["is_price_imputed"] = df["price"].isna()
        m_tg = df.groupby(["property_type","governorate"])["price"].median()
        m_t  = df.groupby("property_type")["price"].median()
        gm   = df["price"].median()
        def ip(r):
            if pd.notna(r["price"]): return r["price"]
            k = (r["property_type"], r["governorate"])
            if k in m_tg.index and pd.notna(m_tg[k]): return m_tg[k]
            if r["property_type"] in m_t.index and pd.notna(m_t[r["property_type"]]): return m_t[r["property_type"]]
            return gm
        df["price"] = df.apply(ip, axis=1)
        _step("price_imputed", int(df["is_price_imputed"].sum()))

        df["surface_m2"] = pd.to_numeric(df["surface_m2"], errors="coerce")
        df["surface_m2"] = df["surface_m2"].where(df["surface_m2"] > 0)
        df["has_surface"] = df["surface_m2"].notna()
        sm_t = df.groupby("property_type")["surface_m2"].median()
        gsm  = df["surface_m2"].median()
        df["surface_m2"] = df.apply(
            lambda r: r["surface_m2"] if pd.notna(r["surface_m2"])
                      else sm_t.get(r["property_type"], gsm), axis=1)

        for col in ["rooms","bathrooms","bedrooms"]:
            if col in df.columns: df[col] = pd.to_numeric(df[col], errors="coerce")
        df["has_rooms_info"]     = df["rooms"].notna()     if "rooms"     in df.columns else False
        df["has_bathrooms_info"] = df["bathrooms"].notna() if "bathrooms" in df.columns else False
        df["has_bedrooms_info"]  = df["bedrooms"].notna()  if "bedrooms"  in df.columns else False
        if "total_rooms" in df.columns: df = df.drop(columns=["total_rooms"])

        df = df[df["governorate"].notna()].reset_index(drop=True)
        _step("drop_null_gov", len(df))

        return df

    # ── LLM anomaly detection (upgraded) ─────────────────────────────────────
    def _flag_suspicious(self, df: pd.DataFrame) -> List[Dict]:
        """
        Sample the cleaned dataset in batches and ask the LLM to flag
        suspicious listings with per-row reasons.

        Returns a list of {"id": int, "reason": str} dicts.
        """
        if "id" not in df.columns or len(df) == 0:
            return []

        all_flagged: List[Dict] = []
        sample_df = df[["id","title","price","surface_m2","property_type",
                         "governorate","source","description"]] \
                    .dropna(subset=["description"]) \
                    .sample(min(self.SAMPLE_SIZE * self.MAX_BATCHES, len(df)), random_state=42)

        batches = [
            sample_df.iloc[i : i + self.SAMPLE_SIZE]
            for i in range(0, len(sample_df), self.SAMPLE_SIZE)
        ]

        for batch_num, batch in enumerate(batches[:self.MAX_BATCHES], 1):
            self.log.info(f"  Anomaly detection batch {batch_num}/{min(len(batches), self.MAX_BATCHES)} "
                          f"({len(batch)} rows)")

            lines = [
                f"ID={r['id']} | {r.get('property_type','?')} | {r.get('governorate','?')} | "
                f"Source={r.get('source','?')} | "
                f"Price={r.get('price','?')} TND | Surface={r.get('surface_m2','?')} m² | "
                f"Desc: {str(r.get('description',''))[:200]}"
                for _, r in batch.iterrows()
            ]

            prompt = self.SUSPICIOUS_PROMPT + "\n".join(lines)

            try:
                result  = self.decide(prompt)
                thought = result.get("thought", "")
                flagged = result.get("flagged", [])

                self.log.info(f"  LLM thought: {thought[:180]}")
                self.log.info(f"  Flagged in batch: {len(flagged)}")

                for item in flagged:
                    try:
                        all_flagged.append({
                            "id":     int(item["id"]),
                            "reason": str(item.get("reason", "no reason given")),
                        })
                    except (KeyError, ValueError, TypeError):
                        continue

            except Exception as e:
                self.log.warning(f"  LLM anomaly detection batch {batch_num} failed: {e}")
                continue

        # Deduplicate by id (same row might appear across batches in edge cases)
        seen = set()
        deduped = []
        for item in all_flagged:
            if item["id"] not in seen:
                seen.add(item["id"])
                deduped.append(item)

        self.log.debug(f"Total suspicious flags: {len(deduped)}")
        return deduped

print("✅ CleaningAgent defined (upgraded LLM anomaly detection)")
````

<h2 style="color: #87BAC3;">🎯 Agent 5 — Orchestrator Agent</h2>

<h3>🎯 What it does</h3>
<p>The Orchestrator is the only agent with visibility into the <strong>full pipeline state</strong>. It sequences agents in the correct order (Articles → Listings → Standardization → Cleaning → Validation), uses <strong>Qwen2.5-0.5B</strong> to make stage-transition decisions, and saves a full JSON run report.</p>

<h3>🧠 LLM decision logic</h3>
<p>After each stage, the Orchestrator sends a compact context to the LLM and asks it to choose: <code>continue / retry / abort / accept_degraded</code>. The LLM receives row counts, source statuses, and quality metrics — information that would require complex nested if-else logic to handle deterministically.</p>

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 118 — OrchestratorAgent  (Phase 1 + Phase 3 combined — full replacement)
# ─────────────────────────────────────────────────────────────────────────────

class OrchestratorAgent(BaseAgent):
    """
    Master controller — sequences all agents, uses LLM for stage gate decisions.

    Phase 3 addition: ScraperHealthAgent runs as Stage 0 (pre-flight)
    before any scraper is touched. Its results flow into ListingsAgent's
    ReAct loop and inform the LLM's routing decisions.
    """

    MIN_LISTINGS = 500

    DECISION_MODEL_ROUTINE = LLM_PRIMARY
    DECISION_MODEL_COMPLEX = LLM_FALLBACK

    def __init__(self):
        super().__init__("OrchestratorAgent")
        self.health_agent          = ScraperHealthAgent()
        self.articles_agent        = ArticlesAgent()
        self.listings_agent        = ListingsAgent()
        self.standardization_agent = StandardizationAgent()
        self.cleaning_agent        = CleaningAgent()

    # ── Main pipeline ──────────────────────────────────────────────────────────
    def run(self, state: Dict[str, Any] = None) -> Dict[str, Any]:
        state = state or {
            "run_id":          datetime.utcnow().strftime("%Y%m%d_%H%M%S"),
            "started_at":      datetime.utcnow().isoformat() + "Z",
            "pipeline_status": "running",
        }

        self.log.info("=" * 60)
        self.log.info("  LENS ESTATE PIPELINE — ORCHESTRATOR")
        self.log.info(f"  Run ID  : {state['run_id']}")
        self.log.info(f"  Started : {state['started_at']}")
        self.log.info("=" * 60)

        # ── STAGE 0: Pre-flight health checks (Phase 3) ───────────────────────
        self.log.info("\n" + "█" * 60)
        self.log.info("  STAGE 0 — Scraper Health Checks")
        self.log.info("█" * 60)
        state = self.health_agent.run(state)
        health = state.get("scraper_health", {})
        blocked = [n for n, h in health.items() if h.get("status") == "blocked"]
        down    = [n for n, h in health.items() if h.get("status") == "down"]
        if blocked:
            self.log.warning(f"  Sites BLOCKED (IP block/rate limit): {blocked}")
        if down:
            self.log.warning(f"  Sites DOWN: {down}")

        # ── STAGE 1: Articles ─────────────────────────────────────────────────
        self.log.info("\n" + "█" * 60)
        self.log.info("  STAGE 1 — Articles Collection")
        self.log.info("█" * 60)
        state     = self.articles_agent.run(state)
        art_paths = state.get("articles_raw_paths", [])
        self.log.info(f"Articles stage complete: {len(art_paths)} sources")

        # ── STAGE 2: Listings ─────────────────────────────────────────────────
        self.log.info("\n" + "█" * 60)
        self.log.info("  STAGE 2 — Listings Collection")
        self.log.info("█" * 60)
        state = self.listings_agent.run(state)
        dec   = self._decide("listings_collection", state)
        self.log_decision(dec, "after_listings")
        if dec["action"] == "abort":
            return self._abort(state, dec["reason"])

        # ── STAGE 3: Standardization ──────────────────────────────────────────
        self.log.info("\n" + "█" * 60)
        self.log.info("  STAGE 3 — Standardization")
        self.log.info("█" * 60)
        state = self.standardization_agent.run(state)
        dec   = self._decide("standardization", state)
        self.log_decision(dec, "after_standardization")
        if dec["action"] == "abort":
            return self._abort(state, dec["reason"])
        if dec["action"] == "retry":
            self.log.info("Orchestrator ordered RETRY on standardization")
            state = self.standardization_agent.run(state)

        # ── STAGE 4: Cleaning ─────────────────────────────────────────────────
        self.log.info("\n" + "█" * 60)
        self.log.info("  STAGE 4 — Cleaning")
        self.log.info("█" * 60)
        state = self.cleaning_agent.run(state)
        dec   = self._decide("cleaning", state)
        self.log_decision(dec, "after_cleaning")
        if dec["action"] == "abort":
            return self._abort(state, dec["reason"])

        # ── STAGE 5: Final Validation ─────────────────────────────────────────
        self.log.info("\n" + "█" * 60)
        self.log.info("  STAGE 5 — Final Validation")
        self.log.info("█" * 60)
        rows       = state.get("final_row_count", 0)
        has_synth  = state.get("listings_has_synthetic", False)

        if rows >= self.MIN_LISTINGS and not has_synth:
            state["pipeline_status"] = "success"
            self.log.info(f"✅ PIPELINE SUCCESS — {rows:,} model-ready rows")
        elif rows >= self.MIN_LISTINGS and has_synth:
            state["pipeline_status"] = "degraded_synthetic"
            self.log.warning(f"⚠️  PIPELINE DEGRADED (synthetic data) — {rows:,} rows")
        elif rows > 0:
            state["pipeline_status"] = "degraded"
            self.log.warning(f"⚠️  PIPELINE DEGRADED — {rows:,} rows (min {self.MIN_LISTINGS:,})")
        else:
            state["pipeline_status"] = "failed"
            self.log.error("❌ PIPELINE FAILED — 0 rows in final output")

        self._save_report(state)
        self.log.info("\n" + "=" * 60)
        self.log.info(f"  PIPELINE COMPLETE — {state['pipeline_status'].upper()}")
        self.log.info(f"  Final rows : {rows:,}")
        self.log.info(f"  Output     : {state.get('model_ready_path', 'N/A')}")
        self.log.info(f"  Report     : logs/run_{state['run_id']}.json")
        self.log.info("=" * 60)
        return state

    # ── Stage gate decision ────────────────────────────────────────────────────
    def _decide(self, stage: str, state: Dict) -> Dict:
        ctx: Dict[str, Any] = {
            "stage":           stage,
            "run_id":          state.get("run_id"),
            "min_rows_target": self.MIN_LISTINGS,
        }

        if stage == "listings_collection":
            results = state.get("listings_results", {})
            ctx.update({
                "total_rows_collected":  state.get("listings_total_rows", 0),
                "has_synthetic_data":    state.get("listings_has_synthetic", False),
                "scraper_health":        state.get("scraper_health", {}),
                "health_summary":        state.get("scraper_health_summary", ""),
                "sources": {
                    name: {
                        "status":        r.get("status"),
                        "rows":          r.get("rows", 0),
                        "fallback_rung": r.get("fallback_rung", ""),
                    }
                    for name, r in results.items()
                },
                "sources_succeeded": sum(1 for r in results.values() if r.get("status") == "success"),
                "sources_fallback":  sum(1 for r in results.values() if r.get("status") == "fallback"),
                "sources_failed":    sum(1 for r in results.values() if r.get("status") == "failed"),
                "recent_failures":   self.listings_agent.memory.recall("failures", 10),
            })

        elif stage == "standardization":
            ctx.update({
                "standardization_status": state.get("standardization_status"),
                "standardization_report": state.get("standardization_report", {}),
                "listings_total_rows":    state.get("listings_total_rows", 0),
                "has_synthetic_data":     state.get("listings_has_synthetic", False),
            })

        elif stage == "cleaning":
            cleaning_report = state.get("cleaning_report", {})
            ctx.update({
                "cleaning_status":      state.get("cleaning_status"),
                "final_row_count":      state.get("final_row_count", 0),
                "cleaning_report":      cleaning_report,
                "suspicious_count":     cleaning_report.get("suspicious", 0),
                "duplicates_removed":   cleaning_report.get("duplicates_removed", 0),
                "null_rate":            cleaning_report.get("null_rate", {}),
                "has_synthetic_data":   state.get("listings_has_synthetic", False),
            })

        rows_ok    = ctx.get("total_rows_collected", ctx.get("final_row_count", 1)) > 0
        all_failed = ctx.get("sources_failed", 0) >= len(ctx.get("sources", {"x": 1}))
        model      = self.DECISION_MODEL_COMPLEX if (not rows_ok or all_failed) \
                     else self.DECISION_MODEL_ROUTINE

        prompt = self._build_decision_prompt(stage, ctx)
        self.log.info(f"\nAsking LLM for routing decision after: {stage} (model={model})")
        result = self.decide(prompt, context=ctx, model=model)

        valid_actions = {"continue", "retry", "abort", "accept_degraded"}
        if result.get("action") not in valid_actions:
            self.log.warning(f"Invalid LLM action '{result.get('action')}' — defaulting to continue")
            result = {
                "thought":    "Invalid action returned — defaulting to continue safely.",
                "action":     "continue",
                "reason":     "invalid_llm_output",
                "confidence": 0.3,
            }
        return result

    def _build_decision_prompt(self, stage: str, ctx: Dict) -> str:
        schema = (
            'Return ONLY this JSON:\n'
            '{\n'
            '  "thought": "<step-by-step reasoning referencing specific numbers>",\n'
            '  "action":  "<continue | retry | abort | accept_degraded>",\n'
            '  "reason":  "<one sentence>",\n'
            '  "confidence": <0.0–1.0>\n'
            '}'
        )
        action_guide = (
            "Action rules:\n"
            "  continue        — data sufficient, proceed\n"
            "  retry           — borderline, worth one more attempt\n"
            "  abort           — unrecoverable (0 real rows, critical error)\n"
            "  accept_degraded — below target but usable\n"
            f"  Minimum acceptable rows: {self.MIN_LISTINGS:,}\n"
            "  NOTE: if has_synthetic_data=true, treat row count with caution — "
            "synthetic rows will be stripped before the model CSV.\n"
        )

        if stage == "listings_collection":
            return (
                f"Stage 'listings_collection' finished.\n\n{action_guide}\n"
                f"Guidance:\n"
                f"  - rows >= {self.MIN_LISTINGS} from real scrapers → continue\n"
                f"  - rows 200–{self.MIN_LISTINGS-1} with >= 1 real source → accept_degraded\n"
                f"  - all sources failed/fallback and 0 real rows → abort\n"
                f"  - blocked sites: note in thought but don't abort if cache provides data\n\n"
                f"{schema}"
            )
        elif stage == "standardization":
            return (
                f"Stage 'standardization' finished.\n\n{action_guide}\n"
                f"Guidance:\n"
                f"  - status=success and rows > {self.MIN_LISTINGS} → continue\n"
                f"  - status=failed → retry once, then abort\n"
                f"  - significant row drop → note but continue unless < 200 remain\n\n"
                f"{schema}"
            )
        elif stage == "cleaning":
            return (
                f"Stage 'cleaning' finished.\n\n{action_guide}\n"
                f"Guidance:\n"
                f"  - final_row_count >= {self.MIN_LISTINGS} → continue\n"
                f"  - 200–{self.MIN_LISTINGS-1} → accept_degraded\n"
                f"  - < 200 → abort\n"
                f"  - suspicious > 20% of rows → note but continue (they're tagged, not deleted)\n"
                f"  - high null_rate on price or surface_m2 (>40%) → note in thought\n\n"
                f"{schema}"
            )
        return f"Stage '{stage}' finished. Decide next action.\n\n{action_guide}\n{schema}"

    def _abort(self, state: Dict, reason: str) -> Dict:
        self.log.error(f"PIPELINE ABORTED: {reason}")
        state["pipeline_status"] = "aborted"
        state["abort_reason"]    = reason
        self.memory.remember("failures", {"reason": reason, "ts": _now()})
        self._save_report(state)
        return state

    def _save_report(self, state: Dict):
        report = {
            "run_id":           state.get("run_id"),
            "started_at":       state.get("started_at"),
            "completed_at":     _now(),
            "pipeline_status":  state.get("pipeline_status"),
            "final_row_count":  state.get("final_row_count", 0),
            "has_synthetic":    state.get("listings_has_synthetic", False),
            "model_ready":      state.get("model_ready_path", ""),
            "scraper_health":   state.get("scraper_health", {}),
            "health_summary":   state.get("scraper_health_summary", ""),
            "articles":         state.get("articles_results", {}),
            "listings":         state.get("listings_results", {}),
            "std_report":       state.get("standardization_report", {}),
            "cleaning_report":  state.get("cleaning_report", {}),
            "llm_decisions":    self.memory.recall("decisions", 20),
        }
        path = f"logs/run_{state.get('run_id', 'unknown')}.json"
        with open(path, "w", encoding="utf-8") as f:
            json.dump(report, f, indent=2, ensure_ascii=False, default=str)
        self.log.info(f"Run report saved → {path}")

print("✅ OrchestratorAgent defined (Phase 1 + Phase 3)")
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 119b — Phase 4: Procedural Memory
# INSERT after Cell 119 (OrchestratorAgent), before Cell 120 (run cell)
#
# Adds:
#   HeuristicsStore  — reads/writes heuristics.json, injected into every LLM prompt
#   run_weekly_review() — call this manually after 20+ runs to let the LLM
#                         distil patterns from run history into new heuristics
# ─────────────────────────────────────────────────────────────────────────────

import glob as _glob_mod

HEURISTICS_PATH = "memory/heuristics.json"

# ── Heuristics store ──────────────────────────────────────────────────────────

class HeuristicsStore:
    """
    Procedural memory — durable rules the LLM has learned from past runs.

    Stored in memory/heuristics.json as a list of structured entries:
      {
        "id":         "h_001",
        "scope":      "mubawab",          # scraper name, agent name, or "global"
        "rule":       "If mubawab fails twice in a row, wait 45 min before retry.",
        "confidence": 0.85,
        "source":     "review_20250401",  # which review session created it
        "created_at": "2025-04-01T...",
        "applies_to": ["ListingsAgent"]
      }

    Usage:
      hs = HeuristicsStore()
      hs.get_for("ListingsAgent")   → list of rule strings relevant to that agent
      hs.get_all()                  → full list of entries
      hs.add(scope, rule, ...)      → manually inject a rule
    """

    def __init__(self, path: str = HEURISTICS_PATH):
        self._path = Path(path)
        self._path.parent.mkdir(parents=True, exist_ok=True)
        self._log  = _build_logger("HeuristicsStore")
        self._data : List[Dict] = self._load()

    def _load(self) -> List[Dict]:
        if self._path.exists():
            try:
                with open(self._path, encoding="utf-8") as f:
                    return json.load(f)
            except Exception as e:
                self._log.warning(f"Could not load heuristics: {e}")
        return []

    def _save(self):
        tmp = self._path.with_suffix(".tmp")
        with open(tmp, "w", encoding="utf-8") as f:
            json.dump(self._data, f, indent=2, ensure_ascii=False)
        os.replace(tmp, self._path)

    def get_all(self) -> List[Dict]:
        return list(self._data)

    def get_for(self, agent_name: str, scope: str = None) -> List[str]:
        """Return rule strings relevant to agent_name (and optionally a scraper scope)."""
        rules = []
        for h in self._data:
            applies = h.get("applies_to", [])
            h_scope = h.get("scope", "global")
            if agent_name in applies or "global" in applies:
                if scope is None or h_scope in (scope, "global"):
                    rules.append(h["rule"])
        return rules

    def add(
        self,
        scope:       str,
        rule:        str,
        confidence:  float      = 0.75,
        applies_to:  List[str]  = None,
        source:      str        = "manual",
    ) -> Dict:
        entry = {
            "id":         f"h_{len(self._data)+1:03d}",
            "scope":      scope,
            "rule":       rule,
            "confidence": round(confidence, 2),
            "source":     source,
            "created_at": _now(),
            "applies_to": applies_to or ["global"],
        }
        self._data.append(entry)
        self._save()
        self._log.info(f"Heuristic added [{entry['id']}] scope={scope}: {rule[:80]}")
        return entry

    def remove(self, heuristic_id: str) -> bool:
        before = len(self._data)
        self._data = [h for h in self._data if h["id"] != heuristic_id]
        if len(self._data) < before:
            self._save()
            self._log.info(f"Heuristic {heuristic_id} removed")
            return True
        return False

    def format_for_prompt(self, agent_name: str, scope: str = None) -> str:
        """Return a formatted string ready to inject into a ReAct prompt."""
        rules = self.get_for(agent_name, scope)
        if not rules:
            return ""
        lines = "\n".join(f"  - {r}" for r in rules)
        return f"Learned heuristics (apply these when reasoning):\n{lines}"

    def summary(self) -> str:
        return (
            f"{len(self._data)} heuristics across "
            f"{len(set(h.get('scope','global') for h in self._data))} scopes"
        )


# Global singleton — imported by agents
HEURISTICS = HeuristicsStore()
print(f"✅ HeuristicsStore ready — {HEURISTICS.summary()}")


# ── Weekly review job ─────────────────────────────────────────────────────────

def run_weekly_review(min_runs: int = 5, max_runs_to_review: int = 30,
                      dry_run: bool = False) -> List[Dict]:
    """
    Read the last N run reports from logs/run_*.json and ask the LLM to
    distil patterns into new heuristics.

    Call this manually after accumulating enough run history:
        new_rules = run_weekly_review()

    Args:
        min_runs          — minimum run reports needed to proceed
        max_runs_to_review — cap on how many reports to feed the LLM
        dry_run           — if True, print proposed rules but don't save them

    Returns list of newly added heuristic dicts.
    """
    log = _build_logger("WeeklyReview")
    log.info("=" * 60)
    log.info("  WEEKLY HEURISTIC REVIEW")
    log.info("=" * 60)

    # ── Collect run reports ────────────────────────────────────────────────
    report_files = sorted(
        _glob_mod.glob("logs/run_*.json"),
        key=os.path.getmtime,
        reverse=True,
    )[:max_runs_to_review]

    if len(report_files) < min_runs:
        log.warning(
            f"Only {len(report_files)} run reports found — need {min_runs} minimum. "
            f"Run the pipeline more times before calling run_weekly_review()."
        )
        return []

    log.info(f"Reviewing {len(report_files)} run reports...")

    reports = []
    for path in report_files:
        try:
            with open(path, encoding="utf-8") as f:
                reports.append(json.load(f))
        except Exception as e:
            log.warning(f"Could not read {path}: {e}")

    # ── Build compact summary for LLM ─────────────────────────────────────
    summaries = []
    for r in reports:
        run_id = r.get("run_id", "?")
        status = r.get("pipeline_status", "?")
        rows   = r.get("final_row_count", 0)
        synth  = r.get("has_synthetic", False)

        listings = r.get("listings", {})
        src_summary = {
            name: {
                "status":        v.get("status"),
                "rows":          v.get("rows", 0),
                "fallback_rung": v.get("fallback_rung", ""),
            }
            for name, v in listings.items()
        }

        health = r.get("scraper_health", {})
        health_summary = {name: h.get("status") for name, h in health.items()}

        decisions = r.get("llm_decisions", [])
        decision_actions = [d.get("decision", {}).get("action") for d in decisions if d.get("decision")]

        summaries.append({
            "run_id":       run_id,
            "status":       status,
            "final_rows":   rows,
            "has_synthetic": synth,
            "listings":     src_summary,
            "health":       health_summary,
            "decisions":    decision_actions,
        })

    existing_rules = [h["rule"] for h in HEURISTICS.get_all()]

    prompt = (
        f"You are reviewing {len(summaries)} pipeline run reports for LensEstate, "
        f"a Tunisian real estate scraping pipeline.\n\n"
        f"The pipeline scrapes listings from: mubawab, tayara, tunisie_annonce.\n"
        f"It also scrapes articles from: bnb, houni.\n\n"
        f"RUN HISTORY (most recent first):\n"
        f"{json.dumps(summaries, indent=2)}\n\n"
        f"EXISTING HEURISTICS (do not duplicate):\n"
        f"{json.dumps(existing_rules, indent=2)}\n\n"
        f"Your task: analyse the run history and identify 3–6 NEW durable patterns "
        f"that should guide future agent decisions. Focus on:\n"
        f"  - Which scrapers fail most often and why (blocked, timeout, empty output)\n"
        f"  - Which fallback rungs were used and whether they helped\n"
        f"  - Patterns in health check results (e.g. a site that is always blocked on weekends)\n"
        f"  - Row count patterns (normal range vs anomaly)\n"
        f"  - Decision patterns (when did the LLM abort/retry vs continue)\n\n"
        f"Return ONLY this JSON:\n"
        f"{{\n"
        f'  "thought": "your overall analysis of the run history",\n'
        f'  "heuristics": [\n'
        f'    {{\n'
        f'      "scope": "mubawab | tayara | tunisie_annonce | bnb | houni | global",\n'
        f'      "rule": "concrete actionable rule in plain English",\n'
        f'      "confidence": 0.0-1.0,\n'
        f'      "applies_to": ["ListingsAgent", "ArticlesAgent", "OrchestratorAgent", ...]\n'
        f'    }}\n'
        f'  ]\n'
        f"}}"
    )

    log.info("Calling LLM for heuristic distillation...")
    raw    = call_llm(prompt, model=LLM_FALLBACK, max_tokens=1024)
    result = parse_llm_json(raw)

    thought    = result.get("thought", "")
    heuristics = result.get("heuristics", [])

    log.info(f"\nLLM analysis:\n  {thought[:400]}")
    log.info(f"\nProposed heuristics: {len(heuristics)}")

    added = []
    for h in heuristics:
        scope      = h.get("scope", "global")
        rule       = h.get("rule", "").strip()
        confidence = float(h.get("confidence", 0.75))
        applies_to = h.get("applies_to", ["global"])

        if not rule:
            continue

        log.info(f"\n  scope={scope}  confidence={confidence:.2f}")
        log.info(f"  rule: {rule}")
        log.info(f"  applies_to: {applies_to}")

        if dry_run:
            log.info("  [DRY RUN — not saved]")
            added.append({"scope": scope, "rule": rule, "confidence": confidence,
                          "applies_to": applies_to, "dry_run": True})
        else:
            entry = HEURISTICS.add(
                scope=scope, rule=rule, confidence=confidence,
                applies_to=applies_to,
                source=f"review_{datetime.utcnow().strftime('%Y%m%d')}",
            )
            added.append(entry)

    if not dry_run and added:
        log.info(f"\n✅ {len(added)} new heuristics saved to {HEURISTICS_PATH}")
        log.info(f"   Total heuristics: {len(HEURISTICS.get_all())}")
    elif dry_run:
        log.info(f"\n[DRY RUN] Would have added {len(added)} heuristics")

    return added


print()
print("Phase 4 — Procedural memory ready:")
print(f"  HeuristicsStore → {HEURISTICS_PATH}")
print("  run_weekly_review() → call after 5+ pipeline runs")
print()
print("Usage examples:")
print("  HEURISTICS.add('mubawab', 'Tends to 403 on Sunday mornings', 0.8, ['ListingsAgent'])")
print("  HEURISTICS.get_for('ListingsAgent', scope='mubawab')")
print("  run_weekly_review(dry_run=True)   # preview without saving")
print("  run_weekly_review()               # save new heuristics")
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 119c — Phase 5: LangGraph Pipeline Graph
# INSERT after Cell 119b (Phase 4 / HeuristicsStore)
#
# Replaces OrchestratorAgent.run() with a LangGraph StateGraph that adds:
#   - Crash recovery: SqliteSaver checkpoints every stage to disk.
#     If the pipeline crashes mid-run, the next call resumes from the
#     exact stage that failed — no re-scraping after a 90-minute run.
#   - Human-in-the-loop: quality_guard node pauses before overwriting
#     the model CSV when the LLM detects something suspicious.
#   - Typed state: PipelineState TypedDict catches key-name typos at
#     definition time instead of silently returning None at runtime.
#   - Visual graph: call graph.get_graph().draw_mermaid() to see the
#     current routing as a Mermaid diagram.
#
# Install once:  pip install langgraph
# ─────────────────────────────────────────────────────────────────────────────

try:
    from langgraph.graph       import StateGraph, END
    from langgraph.checkpoint.sqlite import SqliteSaver
    from typing import TypedDict, Literal
    LANGGRAPH_AVAILABLE = True
    print("✅ LangGraph available")
except ImportError:
    LANGGRAPH_AVAILABLE = False
    print("⚠️  LangGraph not installed — run: pip install langgraph")
    print("   Pipeline will fall back to OrchestratorAgent.run() until installed.")


# ── Typed pipeline state ───────────────────────────────────────────────────────

if LANGGRAPH_AVAILABLE:

    class PipelineState(TypedDict, total=False):
        # Identity
        run_id:                   str
        started_at:               str
        pipeline_status:          str
        abort_reason:             str

        # Stage 0 — health
        scraper_health:           dict
        scraper_health_summary:   str

        # Stage 1 — articles
        articles_raw_paths:       list
        articles_results:         dict
        merged_articles_path:     str

        # Stage 2 — listings
        listings_raw_paths:       list
        listings_results:         dict
        listings_total_rows:      int
        listings_has_synthetic:   bool

        # Stage 3 — standardization
        merged_listings_path:     str
        standardization_status:   str
        standardization_report:   dict
        standardization_retried:  bool

        # Stage 4 — cleaning
        model_ready_path:         str
        cleaning_status:          str
        cleaning_report:          dict
        final_row_count:          int

        # Stage 5 — quality guard
        quality_approved:         bool
        quality_score:            float
        quality_notes:            str

        # LangGraph routing
        _next_stage:              str


    # ── Agent singletons used by all nodes ────────────────────────────────────

    _health_agent          = ScraperHealthAgent()
    _articles_agent        = ArticlesAgent()
    _listings_agent        = ListingsAgent()
    _std_agent             = StandardizationAgent()
    _cleaning_agent        = CleaningAgent()
    _orchestrator          = OrchestratorAgent()   # reuse _decide logic


    # ── Graph nodes ───────────────────────────────────────────────────────────

    def node_health(state: PipelineState) -> PipelineState:
        """Stage 0 — pre-flight health checks."""
        _log = _build_logger("Graph.health")
        _log.info("█" * 60)
        _log.info("  NODE: health")
        return dict(_health_agent.run(dict(state)))


    def node_articles(state: PipelineState) -> PipelineState:
        """Stage 1 — article scrapers (non-blocking)."""
        _log = _build_logger("Graph.articles")
        _log.info("█" * 60)
        _log.info("  NODE: articles")
        return dict(_articles_agent.run(dict(state)))


    def node_listings(state: PipelineState) -> PipelineState:
        """Stage 2 — listing scrapers with ReAct + fallback."""
        _log = _build_logger("Graph.listings")
        _log.info("█" * 60)
        _log.info("  NODE: listings")
        return dict(_listings_agent.run(dict(state)))


    def node_route_listings(state: PipelineState) -> Literal["standardization", "abort"]:
        """
        Conditional edge after listings.
        LLM decides: continue → standardization, abort → abort node.
        """
        dec = _orchestrator._decide("listings_collection", dict(state))
        _orchestrator.log_decision(dec, "after_listings")
        if dec["action"] == "abort":
            return "abort"
        return "standardization"


    def node_standardization(state: PipelineState) -> PipelineState:
        """Stage 3 — schema alignment + LLM schema repair."""
        _log = _build_logger("Graph.standardization")
        _log.info("█" * 60)
        _log.info("  NODE: standardization")
        return dict(_std_agent.run(dict(state)))


    def node_route_standardization(
        state: PipelineState,
    ) -> Literal["cleaning", "standardization", "abort"]:
        """
        Conditional edge after standardization.
        Supports retry: if action==retry and not already retried → loop back once.
        """
        dec = _orchestrator._decide("standardization", dict(state))
        _orchestrator.log_decision(dec, "after_standardization")

        if dec["action"] == "abort":
            return "abort"
        if dec["action"] == "retry" and not state.get("standardization_retried", False):
            return "standardization"      # loop back once
        return "cleaning"


    def node_standardization_mark_retry(state: PipelineState) -> PipelineState:
        """Thin node that marks the retry flag so we don't loop infinitely."""
        return {**state, "standardization_retried": True}


    def node_cleaning(state: PipelineState) -> PipelineState:
        """Stage 4 — cleaning, imputation, LLM anomaly detection."""
        _log = _build_logger("Graph.cleaning")
        _log.info("█" * 60)
        _log.info("  NODE: cleaning")
        return dict(_cleaning_agent.run(dict(state)))


    def node_route_cleaning(
        state: PipelineState,
    ) -> Literal["quality_guard", "abort"]:
        """Conditional edge after cleaning."""
        dec = _orchestrator._decide("cleaning", dict(state))
        _orchestrator.log_decision(dec, "after_cleaning")
        if dec["action"] == "abort":
            return "abort"
        return "quality_guard"


    def node_quality_guard(state: PipelineState) -> PipelineState:
        """
        Stage 5 — quality scoring gate.

        Computes a 0–100 score and asks the LLM for a narrative assessment.
        If score < 60 AND the graph was compiled with interrupt_before=['quality_guard'],
        LangGraph will pause HERE and wait for a human to call .invoke(None, config)
        to resume — giving you a chance to inspect the data first.
        """
        _log = _build_logger("Graph.quality_guard")
        _log.info("█" * 60)
        _log.info("  NODE: quality_guard")

        cleaning_report = state.get("cleaning_report", {})
        final_rows      = state.get("final_row_count", 0)
        min_listings    = OrchestratorAgent.MIN_LISTINGS
        has_synthetic   = state.get("listings_has_synthetic", False)

        # ── Compute quality score 0-100 ───────────────────────────────────
        score = 100.0

        # Row count penalty
        if final_rows < min_listings:
            score -= 40 * (1 - final_rows / max(min_listings, 1))

        # Synthetic data penalty
        if has_synthetic:
            score -= 15

        # Null rate penalties (price + surface most important)
        null_rate = cleaning_report.get("null_rate", {})
        for col, weight in [("price", 15), ("surface_m2", 10), ("governorate", 8)]:
            rate = null_rate.get(col, 0)
            if rate > 0.4:
                score -= weight * (rate - 0.4) / 0.6

        # Suspicious row penalty
        suspicious = cleaning_report.get("suspicious", 0)
        if final_rows > 0:
            susp_rate = suspicious / final_rows
            if susp_rate > 0.05:
                score -= 10 * min(susp_rate / 0.2, 1.0)

        score = max(0.0, min(100.0, round(score, 1)))

        # ── LLM narrative ──────────────────────────────────────────────────
        prompt = (
            f"You are the QualityGuard for LensEstate pipeline run {state.get('run_id')}.\n"
            f"Quality score: {score}/100\n"
            f"Final rows: {final_rows:,} (min target: {min_listings:,})\n"
            f"Has synthetic data: {has_synthetic}\n"
            f"Null rates: {null_rate}\n"
            f"Suspicious rows: {suspicious}\n\n"
            f"Write a 2-sentence quality assessment and decide whether to approve.\n"
            f"Approve if score >= 60 and at least some real data exists.\n"
            'Return ONLY JSON: {"thought": "...", "approved": true/false, '
            '"notes": "2-sentence assessment"}'
        )

        result   = _orchestrator.decide(prompt)
        approved = result.get("approved", score >= 60)
        notes    = result.get("notes", "")

        _log.info(f"  Quality score : {score}/100")
        _log.info(f"  Approved      : {approved}")
        _log.info(f"  Notes         : {notes}")

        # Determine final pipeline status
        if approved and not has_synthetic:
            pipeline_status = "success"
        elif approved and has_synthetic:
            pipeline_status = "degraded_synthetic"
        elif final_rows > 0:
            pipeline_status = "degraded"
        else:
            pipeline_status = "failed"

        _orchestrator._save_report({
            **dict(state),
            "pipeline_status":  pipeline_status,
            "quality_score":    score,
            "quality_notes":    notes,
            "quality_approved": approved,
        })

        _log.info(f"  Pipeline status: {pipeline_status.upper()}")

        return {
            **state,
            "quality_approved": approved,
            "quality_score":    score,
            "quality_notes":    notes,
            "pipeline_status":  pipeline_status,
        }


    def node_abort(state: PipelineState) -> PipelineState:
        """Terminal abort node — logs and saves report."""
        _log = _build_logger("Graph.abort")
        _log.error("PIPELINE ABORTED by LangGraph routing decision")
        result = dict(state)
        result["pipeline_status"] = "aborted"
        _orchestrator._save_report(result)
        return result


    # ── Build the graph ───────────────────────────────────────────────────────

    def build_pipeline_graph(
        checkpoint_db:  str  = "memory/pipeline_runs.db",
        hitl_on_quality: bool = True,
    ) -> tuple:
        """
        Build and compile the LangGraph pipeline.

        Args:
            checkpoint_db   — SQLite file for crash-recovery checkpoints
            hitl_on_quality — if True, graph pauses at quality_guard when
                              score < 60 for human review before continuing

        Returns (app, checkpointer) — call app.invoke(initial_state, config={"configurable": {"thread_id": run_id}})
        """
        graph = StateGraph(PipelineState)

        # ── Register nodes ────────────────────────────────────────────────
        graph.add_node("health",                  node_health)
        graph.add_node("articles",                node_articles)
        graph.add_node("listings",                node_listings)
        graph.add_node("standardization",         node_standardization)
        graph.add_node("standardization_retry",   node_standardization_mark_retry)
        graph.add_node("cleaning",                node_cleaning)
        graph.add_node("quality_guard",           node_quality_guard)
        graph.add_node("abort",                   node_abort)

        # ── Entry point ───────────────────────────────────────────────────
        graph.set_entry_point("health")

        # ── Unconditional edges ───────────────────────────────────────────
        graph.add_edge("health",   "articles")
        graph.add_edge("articles", "listings")

        # ── Conditional edges ─────────────────────────────────────────────
        graph.add_conditional_edges(
            "listings",
            node_route_listings,
            {"standardization": "standardization", "abort": "abort"},
        )

        graph.add_conditional_edges(
            "standardization",
            node_route_standardization,
            {
                "cleaning":        "cleaning",
                "standardization": "standardization_retry",
                "abort":           "abort",
            },
        )
        graph.add_edge("standardization_retry", "standardization")

        graph.add_conditional_edges(
            "cleaning",
            node_route_cleaning,
            {"quality_guard": "quality_guard", "abort": "abort"},
        )

        # Terminal nodes
        graph.add_edge("quality_guard", END)
        graph.add_edge("abort",         END)

        # ── Checkpointer (crash recovery) ─────────────────────────────────
        Path(checkpoint_db).parent.mkdir(parents=True, exist_ok=True)
        checkpointer = SqliteSaver.from_conn_string(checkpoint_db)

        # ── Human-in-the-loop ─────────────────────────────────────────────
        interrupt_before = ["quality_guard"] if hitl_on_quality else []

        app = graph.compile(
            checkpointer=checkpointer,
            interrupt_before=interrupt_before,
        )

        _build_logger("Graph").info(
            f"Pipeline graph compiled — "
            f"checkpointer={checkpoint_db}  "
            f"hitl={'quality_guard' if hitl_on_quality else 'disabled'}"
        )
        return app, checkpointer


    # ── Runner function ───────────────────────────────────────────────────────

    def run_pipeline_graph(
        hitl_on_quality: bool = False,
        resume_thread_id: Optional[str] = None,
    ) -> dict:
        """
        Run the full pipeline via LangGraph.

        Args:
            hitl_on_quality — pause at quality_guard for human approval
                              (set to True in production, False for automated runs)
            resume_thread_id — if provided, resume an interrupted run by its run_id

        Usage:
            # Normal run
            result = run_pipeline_graph()

            # Production run with human quality gate
            result = run_pipeline_graph(hitl_on_quality=True)

            # Resume a crashed run
            result = run_pipeline_graph(resume_thread_id="20250401_143022")
        """
        log = _build_logger("PipelineRunner")

        app, checkpointer = build_pipeline_graph(hitl_on_quality=hitl_on_quality)

        if resume_thread_id:
            # Resume from checkpoint — pass None as input to continue from last state
            thread_id = resume_thread_id
            log.info(f"Resuming run {thread_id} from checkpoint...")
            config   = {"configurable": {"thread_id": thread_id}}
            result   = app.invoke(None, config=config)
        else:
            # Fresh run
            run_id   = datetime.utcnow().strftime("%Y%m%d_%H%M%S")
            thread_id = run_id
            initial_state: PipelineState = {
                "run_id":          run_id,
                "started_at":      datetime.utcnow().isoformat() + "Z",
                "pipeline_status": "running",
            }
            log.info(f"Starting fresh run {run_id}...")
            config = {"configurable": {"thread_id": thread_id}}
            result = app.invoke(initial_state, config=config)

        # Handle HITL pause (graph stopped at quality_guard)
        if result is None:
            log.info(
                "\n" + "=" * 60 + "\n"
                "  PIPELINE PAUSED — human review required\n"
                f"  Quality score: {result.get('quality_score', '?') if result else '?'}/100\n"
                f"  To approve and continue:\n"
                f"      run_pipeline_graph(resume_thread_id='{thread_id}')\n"
                "=" * 60
            )
            return {"pipeline_status": "paused_hitl", "thread_id": thread_id}

        # Print summary
        sep = "=" * 60
        print()
        print(sep)
        print("FINAL PIPELINE SUMMARY  (LangGraph)")
        print(sep)
        print("Status       :", result.get("pipeline_status", "?").upper())
        print("Run ID       :", result.get("run_id", "?"))
        print("Final rows   :", result.get("final_row_count", 0))
        print("Quality score:", result.get("quality_score", "N/A"))
        print("Approved     :", result.get("quality_approved", "N/A"))
        if result.get("quality_notes"):
            print("Notes        :", result["quality_notes"])
        print("Output       :", result.get("model_ready_path", "N/A"))
        print(sep)

        return result


print()
print("Phase 5 — LangGraph pipeline ready:")
print("  run_pipeline_graph()                           — standard automated run")
print("  run_pipeline_graph(hitl_on_quality=True)       — pause for human review")
print("  run_pipeline_graph(resume_thread_id='...')     — resume after crash/pause")
print()
print("Heuristics reminder (Phase 4):")
print("  run_weekly_review(dry_run=True)  — preview new heuristics")
print("  run_weekly_review()              — save after 5+ runs")
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 119d — Phase 6: QualityGuardAgent
# INSERT after Cell 121 (Phase 5 / LangGraph), before Cell 122 (run cell)
#
# Adds:
#   QualityGuardAgent  — standalone agent with deep quality analysis, LLM
#                        narrative, configurable approval threshold, and an
#                        HTML quality report saved to logs/quality_{run_id}.html
#
# The Phase 5 node_quality_guard is replaced by a thin wrapper that delegates
# to QualityGuardAgent.run() — so both the LangGraph path and the legacy
# OrchestratorAgent path share the same quality logic.
# ─────────────────────────────────────────────────────────────────────────────


class QualityGuardAgent(BaseAgent):
    """
    Dedicated quality scoring agent — the final gate before a pipeline run
    is marked successful.

    Responsibilities:
      1. Compute a structured 0–100 quality score across 5 dimensions.
      2. Ask the LLM for a narrative assessment and an approval decision.
      3. Persist the full quality report as:
           - state["quality_report"]  (dict, available downstream)
           - logs/quality_{run_id}.html  (human-readable HTML report)
           - memory "quality" category  (for weekly review / heuristics)

    Score dimensions (each 0–20, total 0–100):
      D1  Row volume      — how close to MIN_LISTINGS
      D2  Data freshness  — real data vs fallback/synthetic
      D3  Completeness    — null rates on critical columns
      D4  Consistency     — suspicious row rate from CleaningAgent
      D5  Source balance  — no single source dominates > 90%

    Approval threshold: score >= APPROVE_THRESHOLD (default 60).
    Can be overridden per-run via state["quality_threshold"].
    """

    APPROVE_THRESHOLD = 60     # score below this triggers a warning / HITL pause
    MIN_LISTINGS      = 500    # mirrors OrchestratorAgent.MIN_LISTINGS

    # Dimension weights (must sum to 100)
    WEIGHTS = {
        "row_volume":    25,
        "data_freshness": 20,
        "completeness":  25,
        "consistency":   15,
        "source_balance": 15,
    }

    def __init__(self):
        super().__init__("QualityGuardAgent")

    # ── Public entry point ─────────────────────────────────────────────────────
    def run(self, state: Dict[str, Any]) -> Dict[str, Any]:
        self.log.info("=" * 60)
        self.log.info("QualityGuardAgent — quality assessment")
        self.log.info("=" * 60)
        self.memory.clear()

        run_id    = state.get("run_id", "unknown")
        threshold = state.get("quality_threshold", self.APPROVE_THRESHOLD)

        # ── 1. Compute structured score ───────────────────────────────────
        scores, details = self._compute_scores(state)
        total = round(sum(scores.values()), 1)

        self.log.info(f"\nQuality scores:")
        for dim, s in scores.items():
            max_s = self.WEIGHTS[dim]
            bar   = "█" * int(s / max_s * 10) + "░" * (10 - int(s / max_s * 10))
            self.log.info(f"  {dim:<18} [{bar}] {s:.1f}/{max_s}")
        self.log.info(f"  {'TOTAL':<18}  {total:.1f}/100")

        # ── 2. LLM narrative + approval ───────────────────────────────────
        narrative, approved, llm_thought = self._llm_assess(
            state, scores, total, threshold
        )
        self.log.info(f"\nLLM thought   : {llm_thought[:200]}")
        self.log.info(f"LLM narrative : {narrative[:200]}")
        self.log.info(f"Approved      : {approved}  (threshold={threshold})")

        # ── 3. Determine pipeline status ──────────────────────────────────
        has_synth = state.get("listings_has_synthetic", False)
        final_rows = state.get("final_row_count", 0)

        if approved and not has_synth:
            pipeline_status = "success"
        elif approved and has_synth:
            pipeline_status = "degraded_synthetic"
        elif final_rows > 0:
            pipeline_status = "degraded"
        else:
            pipeline_status = "failed"

        # ── 4. Assemble full quality report dict ──────────────────────────
        quality_report = {
            "run_id":             run_id,
            "total_score":        total,
            "threshold":          threshold,
            "approved":           approved,
            "pipeline_status":    pipeline_status,
            "dimension_scores":   scores,
            "dimension_details":  details,
            "narrative":          narrative,
            "llm_thought":        llm_thought,
            "final_row_count":    final_rows,
            "has_synthetic":      has_synth,
            "assessed_at":        _now(),
        }

        # ── 5. Save HTML report ───────────────────────────────────────────
        html_path = self._save_html_report(quality_report)
        quality_report["html_report"] = html_path
        self.log.info(f"\nHTML report   : {html_path}")

        # ── 6. Persist to memory ──────────────────────────────────────────
        self.memory.remember("quality", {
            "run_id": run_id,
            "score":  total,
            "approved": approved,
            "status": pipeline_status,
        })

        # ── 7. Update state ───────────────────────────────────────────────
        state = self.update_state(state, "quality_report",    quality_report)
        state = self.update_state(state, "quality_score",     total)
        state = self.update_state(state, "quality_approved",  approved)
        state = self.update_state(state, "quality_notes",     narrative)
        state = self.update_state(state, "pipeline_status",   pipeline_status)

        self.log.info(f"\n{'─'*60}")
        self.log.info(f"QualityGuardAgent done — {pipeline_status.upper()}  score={total}/100")
        self.log.info(f"{'─'*60}")
        return state

    # ── Score computation ──────────────────────────────────────────────────────
    def _compute_scores(
        self, state: Dict
    ) -> Tuple[Dict[str, float], Dict[str, Any]]:
        """
        Returns (scores dict, details dict).
        Each score is 0–weight[dim], summing to 0–100.
        """
        cleaning_report  = state.get("cleaning_report", {})
        listings_results = state.get("listings_results", {})
        final_rows       = state.get("final_row_count", 0)
        has_synth        = state.get("listings_has_synthetic", False)
        null_rate        = cleaning_report.get("null_rate", {})
        suspicious       = cleaning_report.get("suspicious", 0)

        scores  = {}
        details = {}

        # ── D1: Row volume (0-25) ─────────────────────────────────────────
        w = self.WEIGHTS["row_volume"]
        if final_rows >= self.MIN_LISTINGS:
            d1 = w
        elif final_rows > 0:
            d1 = round(w * (final_rows / self.MIN_LISTINGS) ** 0.7, 2)
        else:
            d1 = 0.0
        scores["row_volume"]  = d1
        details["row_volume"] = {
            "final_rows":    final_rows,
            "target":        self.MIN_LISTINGS,
            "score":         d1,
            "max":           w,
        }

        # ── D2: Data freshness (0-20) ─────────────────────────────────────
        w  = self.WEIGHTS["data_freshness"]
        n_success  = sum(1 for r in listings_results.values() if r.get("status") == "success")
        n_fallback = sum(1 for r in listings_results.values() if r.get("status") == "fallback")
        n_total    = max(len(listings_results), 1)

        freshness_ratio = n_success / n_total
        d2 = round(w * freshness_ratio, 2)
        if has_synth:
            d2 = round(d2 * 0.5, 2)   # heavy penalty for synthetic data

        # Check fallback rungs — cache_soft is better than cache_stale
        rungs = [r.get("fallback_rung", "") for r in listings_results.values()
                 if r.get("status") == "fallback"]
        rung_penalty = sum({"cache_stale": 3, "synthetic": 8}.get(rg, 0) for rg in rungs)
        d2 = max(0.0, round(d2 - rung_penalty, 2))

        scores["data_freshness"]  = d2
        details["data_freshness"] = {
            "sources_succeeded": n_success,
            "sources_fallback":  n_fallback,
            "has_synthetic":     has_synth,
            "fallback_rungs":    rungs,
            "score":             d2,
            "max":               w,
        }

        # ── D3: Completeness (0-25) ───────────────────────────────────────
        w = self.WEIGHTS["completeness"]
        critical_cols = {
            "price":        0.40,   # penalty threshold
            "surface_m2":   0.45,
            "governorate":  0.20,
            "property_type": 0.25,
        }
        col_scores = []
        col_details = {}
        for col, threshold in critical_cols.items():
            rate = null_rate.get(col, 0.0)
            if rate <= threshold:
                col_s = 1.0
            else:
                col_s = max(0.0, 1.0 - (rate - threshold) / (1.0 - threshold))
            col_scores.append(col_s)
            col_details[col] = {"null_rate": round(rate, 3), "score_ratio": round(col_s, 3)}

        d3 = round(w * (sum(col_scores) / len(col_scores)), 2) if col_scores else w
        scores["completeness"]  = d3
        details["completeness"] = {"columns": col_details, "score": d3, "max": w}

        # ── D4: Consistency (0-15) ────────────────────────────────────────
        w = self.WEIGHTS["consistency"]
        susp_rate = suspicious / max(final_rows, 1)
        if susp_rate <= 0.02:
            d4 = w
        elif susp_rate <= 0.10:
            d4 = round(w * (1 - (susp_rate - 0.02) / 0.08), 2)
        else:
            d4 = 0.0
        scores["consistency"]  = d4
        details["consistency"] = {
            "suspicious_rows":  suspicious,
            "suspicious_rate":  round(susp_rate, 4),
            "score":            d4,
            "max":              w,
        }

        # ── D5: Source balance (0-15) ─────────────────────────────────────
        w = self.WEIGHTS["source_balance"]
        row_counts = [r.get("rows", 0) for r in listings_results.values()
                      if r.get("rows", 0) > 0]
        if len(row_counts) >= 2:
            total_r   = sum(row_counts)
            max_share = max(row_counts) / total_r if total_r > 0 else 1.0
            # Penalty only if one source > 90% of all rows
            d5 = round(w * max(0.0, 1.0 - max(0.0, max_share - 0.9) / 0.1), 2)
        elif len(row_counts) == 1:
            d5 = round(w * 0.6, 2)   # single source — partial credit
        else:
            d5 = 0.0
        scores["source_balance"]  = d5
        details["source_balance"] = {
            "source_row_counts": row_counts,
            "score":             d5,
            "max":               w,
        }

        return scores, details

    # ── LLM assessment ─────────────────────────────────────────────────────────
    def _llm_assess(
        self,
        state:     Dict,
        scores:    Dict[str, float],
        total:     float,
        threshold: int,
    ) -> Tuple[str, bool, str]:
        """Ask the LLM for a narrative + approval decision."""

        cleaning_report = state.get("cleaning_report", {})
        listings_results = state.get("listings_results", {})

        source_summary = {
            name: {
                "status":        r.get("status"),
                "rows":          r.get("rows", 0),
                "fallback_rung": r.get("fallback_rung", ""),
            }
            for name, r in listings_results.items()
        }

        prompt = (
            f"You are the QualityGuardAgent for LensEstate, run {state.get('run_id')}.\n\n"
            f"Quality score: {total}/100  (approval threshold: {threshold}/100)\n\n"
            f"Dimension scores:\n"
            + "\n".join(
                f"  {dim:<18}: {s:.1f}/{self.WEIGHTS[dim]}"
                for dim, s in scores.items()
            )
            + f"\n\nData context:\n"
            f"  Final rows     : {state.get('final_row_count', 0):,}\n"
            f"  Has synthetic  : {state.get('listings_has_synthetic', False)}\n"
            f"  Null rates     : {cleaning_report.get('null_rate', {})}\n"
            f"  Suspicious rows: {cleaning_report.get('suspicious', 0)}\n"
            f"  Sources        : {json.dumps(source_summary)}\n\n"
            f"Write a 3-sentence quality assessment covering:\n"
            f"  1. What the data quality looks like overall\n"
            f"  2. The biggest weakness in this run\n"
            f"  3. Whether this dataset is fit for model training\n\n"
            f"Approve if total >= {threshold} AND at least some real (non-synthetic) data exists.\n\n"
            f'Return ONLY JSON:\n'
            f'{{"thought": "...", "approved": true/false, "narrative": "3-sentence assessment"}}'
        )

        result    = self.decide(prompt, model=LLM_FALLBACK)
        thought   = result.get("thought", "")
        narrative = result.get("narrative", f"Quality score {total}/100.")
        approved  = result.get("approved", total >= threshold)

        return narrative, bool(approved), thought

    # ── HTML report ────────────────────────────────────────────────────────────
    def _save_html_report(self, report: Dict) -> str:
        """Generate and save a human-readable HTML quality report."""

        run_id  = report.get("run_id", "unknown")
        total   = report.get("total_score", 0)
        status  = report.get("pipeline_status", "?").upper()
        scores  = report.get("dimension_scores", {})
        details = report.get("dimension_details", {})

        status_color = {
            "SUCCESS":             "#1D9E75",
            "DEGRADED_SYNTHETIC":  "#BA7517",
            "DEGRADED":            "#BA7517",
            "FAILED":              "#A32D2D",
        }.get(status, "#888780")

        score_color = "#1D9E75" if total >= 80 else "#BA7517" if total >= 60 else "#A32D2D"

        def dim_row(dim, score):
            w    = self.WEIGHTS.get(dim, 20)
            pct  = int(score / w * 100)
            bar_color = "#1D9E75" if pct >= 80 else "#BA7517" if pct >= 50 else "#A32D2D"
            label = dim.replace("_", " ").title()
            return (
                f"<tr><td style='padding:8px 12px;font-size:13px'>{label}</td>"
                f"<td style='padding:8px 12px'>"
                f"<div style='background:#f0f0f0;border-radius:4px;height:16px;width:200px'>"
                f"<div style='background:{bar_color};width:{pct}%;height:16px;border-radius:4px'></div>"
                f"</div></td>"
                f"<td style='padding:8px 12px;font-size:13px;font-weight:500'>{score:.1f}/{w}</td>"
                f"<td style='padding:8px 12px;font-size:12px;color:#666'>{self._detail_str(dim, details.get(dim, {}))}</td>"
                f"</tr>"
            )

        rows_html = "\n".join(dim_row(d, s) for d, s in scores.items())

        narrative = report.get("narrative", "").replace("\n", "<br>")
        sources   = report.get("dimension_details", {}).get("data_freshness", {})
        null_info = report.get("dimension_details", {}).get("completeness", {}).get("columns", {})
        null_rows = "\n".join(
            f"<tr><td style='padding:4px 12px;font-size:12px'>{col}</td>"
            f"<td style='padding:4px 12px;font-size:12px'>{info.get('null_rate',0)*100:.1f}%</td></tr>"
            for col, info in null_info.items()
        )

        html = f"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<title>LensEstate Quality Report — {run_id}</title>
<style>
  body {{ font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', sans-serif;
          max-width: 900px; margin: 40px auto; padding: 0 24px; color: #1a1a1a; }}
  h1   {{ font-size: 22px; font-weight: 600; margin-bottom: 4px; }}
  h2   {{ font-size: 16px; font-weight: 500; margin: 28px 0 12px; color: #333; }}
  .meta {{ font-size: 13px; color: #666; margin-bottom: 28px; }}
  .badge {{ display: inline-block; padding: 4px 12px; border-radius: 20px;
             font-size: 13px; font-weight: 500; color: #fff;
             background: {status_color}; margin-left: 10px; vertical-align: middle; }}
  .score-big {{ font-size: 52px; font-weight: 700; color: {score_color};
                 line-height: 1; display: inline-block; }}
  .score-label {{ font-size: 14px; color: #666; vertical-align: bottom;
                   padding-bottom: 8px; margin-left: 6px; }}
  .card {{ background: #f8f8f8; border-radius: 10px; padding: 18px 22px; margin-bottom: 18px; }}
  table {{ border-collapse: collapse; width: 100%; }}
  tr:nth-child(even) {{ background: #fafafa; }}
  th {{ text-align: left; padding: 8px 12px; font-size: 12px;
        font-weight: 500; color: #888; border-bottom: 1px solid #e8e8e8; }}
  .narrative {{ font-size: 14px; line-height: 1.7; color: #333; }}
  .footer {{ font-size: 12px; color: #aaa; margin-top: 40px; border-top: 1px solid #eee;
              padding-top: 16px; }}
</style>
</head>
<body>
<h1>LensEstate Pipeline Quality Report <span class="badge">{status}</span></h1>
<p class="meta">Run ID: <strong>{run_id}</strong> &nbsp;·&nbsp; Assessed: {report.get("assessed_at","")[:19]}</p>

<div class="card" style="display:flex;align-items:flex-end;gap:32px">
  <div>
    <div class="score-big">{total}</div>
    <span class="score-label">/100 &nbsp; (threshold: {report.get("threshold",60)})</span>
  </div>
  <div style="font-size:13px;color:#555;max-width:400px">
    <strong>Final rows:</strong> {report.get("final_row_count",0):,} &nbsp;
    <strong>Synthetic:</strong> {"yes" if report.get("has_synthetic") else "no"}<br>
    <strong>Approved:</strong> {"✅ yes" if report.get("approved") else "❌ no"}
  </div>
</div>

<h2>Quality dimensions</h2>
<table>
  <tr><th>Dimension</th><th>Score</th><th>Points</th><th>Detail</th></tr>
  {rows_html}
</table>

<h2>LLM assessment</h2>
<div class="card narrative">{narrative}</div>

<h2>Null rates (critical columns)</h2>
<table>
  <tr><th>Column</th><th>Null rate</th></tr>
  {null_rows}
</table>

<div class="footer">
  Generated by QualityGuardAgent · LensEstate pipeline · {report.get("assessed_at","")[:10]}
</div>
</body>
</html>"""

        out_dir = Path("logs")
        out_dir.mkdir(exist_ok=True)
        out_path = str(out_dir / f"quality_{run_id}.html")
        with open(out_path, "w", encoding="utf-8") as f:
            f.write(html)
        return out_path

    def _detail_str(self, dim: str, detail: Dict) -> str:
        """One-line detail string per dimension for the HTML table."""
        if dim == "row_volume":
            return f"{detail.get('final_rows',0):,} rows / {detail.get('target',0):,} target"
        if dim == "data_freshness":
            return (
                f"{detail.get('sources_succeeded',0)} fresh, "
                f"{detail.get('sources_fallback',0)} fallback"
                + (" + synthetic" if detail.get("has_synthetic") else "")
            )
        if dim == "completeness":
            worst = max(detail.get("columns", {}).items(),
                        key=lambda x: x[1].get("null_rate", 0),
                        default=("?", {"null_rate": 0}))
            return f"worst: {worst[0]} {worst[1].get('null_rate',0)*100:.1f}% null"
        if dim == "consistency":
            return f"{detail.get('suspicious_rows',0)} suspicious ({detail.get('suspicious_rate',0)*100:.1f}%)"
        if dim == "source_balance":
            counts = detail.get("source_row_counts", [])
            return f"{len(counts)} active sources"
        return ""


# ── Global singleton ──────────────────────────────────────────────────────────
_quality_agent = QualityGuardAgent()

# ── Patch node_quality_guard to delegate to QualityGuardAgent ────────────────
# This replaces the inline scoring logic in the Phase 5 LangGraph cell
# so both the LangGraph path and the legacy path share the same code.

if LANGGRAPH_AVAILABLE:
    def node_quality_guard(state: PipelineState) -> PipelineState:
        """
        Phase 6 upgrade: delegates entirely to QualityGuardAgent.
        Same signature as before — LangGraph graph definition unchanged.
        """
        _log = _build_logger("Graph.quality_guard")
        _log.info("█" * 60)
        _log.info("  NODE: quality_guard  (QualityGuardAgent)")

        # Apply HITL threshold from state if set, else use default
        result = _quality_agent.run(dict(state))
        return result


print("✅ QualityGuardAgent defined (Phase 6)")
print()
print("Outputs per run:")
print("  state['quality_report']           — full structured dict")
print("  state['quality_score']            — float 0-100")
print("  state['quality_approved']         — bool")
print("  logs/quality_{run_id}.html        — human-readable HTML report")
print("  memory 'quality' category         — persisted for weekly review")
print()
print("To set a custom approval threshold for one run:")
print("  state['quality_threshold'] = 70   # before calling run_pipeline_graph()")
````

<h2 style="color: #87BAC3;">🚀 Run the Full Pipeline</h2>
<p>Instantiate the Orchestrator and run. Detailed logs appear in the cell output (INFO level) and in <code>logs/pipeline.log</code> (DEBUG level — includes every subprocess call, LLM response, row count, and state update).</p>
<p><strong>Note on scrapers:</strong> if the scraper paths don't exist on this machine, the agents will log the FileNotFoundError clearly and continue — existing CSVs will be used as fallback if available.</p>

````python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 120 — Run the pipeline  (Phases 1-6)
# Prerequisite: set OPENROUTER_API_KEY before this cell.
# Quick sanity check: test_llm_connection()
#
# Two execution paths:
#   Path A (recommended) — LangGraph with crash recovery + QualityGuardAgent
#   Path B (legacy)      — OrchestratorAgent.run() + QualityGuardAgent
# ─────────────────────────────────────────────────────────────────────────────

def _print_summary(result: dict, mode: str = "LangGraph"):
    sep = "=" * 60
    print()
    print(sep)
    print(f"FINAL PIPELINE SUMMARY  ({mode})")
    print(sep)
    print("Status        :", result.get("pipeline_status", "?").upper())
    print("Run ID        :", result.get("run_id", "?"))
    print("Final rows    :", result.get("final_row_count", 0))
    print("Output        :", result.get("model_ready_path", "N/A"))
    print("Has synthetic :", result.get("listings_has_synthetic", False))

    qr = result.get("quality_report", {})
    if qr:
        print()
        print(f"Quality score : {qr.get('total_score', '?')}/100  "
              f"(threshold={qr.get('threshold', 60)})  "
              f"approved={'yes' if qr.get('approved') else 'no'}")
        print("Dimensions    :", {
            k: f"{v:.1f}" for k, v in qr.get("dimension_scores", {}).items()
        })
        if qr.get("narrative"):
            print("Narrative     :", qr["narrative"][:180])
        if qr.get("html_report"):
            print("HTML report   :", qr["html_report"])

    health = result.get("scraper_health", {})
    if health:
        print()
        print("Scraper health:")
        for sname, h in health.items():
            print("  " + sname.ljust(22) + "-> " + h.get("status", "?") +
                  "  (" + h.get("detail", "") + ")")

    listings = result.get("listings_results", {})
    if listings:
        print()
        print("Listings results:")
        for sname, r in listings.items():
            rung     = r.get("fallback_rung", "")
            rung_str = " [fallback:" + rung + "]" if rung else ""
            print("  " + sname.ljust(22) + "-> " + r.get("status", "?") +
                  rung_str + "  rows=" + str(r.get("rows", 0)))

    run_id = result.get("run_id", "?")
    print()
    print("Log file      : logs/pipeline.log")
    print("Run report    : logs/run_" + run_id + ".json")
    if qr.get("html_report"):
        print("Quality HTML  : " + qr["html_report"])
    print(sep)


# ── Path A: LangGraph (recommended) ──────────────────────────────────────────
if LANGGRAPH_AVAILABLE:

    # Standard automated run — QualityGuardAgent runs as node_quality_guard
    result = run_pipeline_graph(hitl_on_quality=False)

    # Production run — pause at quality_guard for human approval before finalising:
    # result = run_pipeline_graph(hitl_on_quality=True)

    # Resume a crashed or paused run by its run_id:
    # result = run_pipeline_graph(resume_thread_id="20250401_143022")

    # Override approval threshold for this run only:
    # result = run_pipeline_graph()  # after setting state["quality_threshold"] = 70

    _print_summary(result, mode="LangGraph")

# ── Path B: Legacy OrchestratorAgent (fallback if LangGraph not installed) ───
else:
    print("LangGraph not available — using OrchestratorAgent + QualityGuardAgent directly.")
    print("Install with: pip install langgraph\n")

    orchestrator = OrchestratorAgent()
    result       = orchestrator.run()

    # Phase 6: run QualityGuardAgent explicitly on the legacy path
    result = _quality_agent.run(result)

    _print_summary(result, mode="Legacy")
````

<h2 style="color: #87BAC3;">🔄 Reset Stale Failure Memory (Run When Needed)</h2>

<h3>🎯 Purpose</h3>
<p>The agent memory persists failure history across runs. If scrapers failed during early testing (e.g. because virtual environments weren't activated or paths weren't set up yet), those failures get recorded and the agent will permanently skip those scrapers even after the issue is resolved — because it hit the <code>MAX_RECENT_FAILURES = 3</code> threshold.</p>
<p>Run this cell to <strong>clear the failure history</strong> for specific agents. This does NOT delete quality snapshots or decision logs — only the failure entries. After resetting, the agents will attempt to run the scrapers again on the next pipeline execution.</p>

````python
# ── Memory management utilities ───────────────────────────────────────────────
# Cell 121 — Run any of these blocks individually as needed.

# ── Block A: Clear failure history so scrapers will retry ─────────────────────
# Run this when a scraper is fixed and you want the agent to attempt it again.

agents_to_reset = {
    "ListingsAgent" : ["mubawab", "tayara", "tunisie_annonce"],
    "ArticlesAgent" : ["bnb", "houni"],
}

from pathlib import Path
import json, os

state_dir = Path("memory/state/")

for agent_name, scrapers in agents_to_reset.items():
    mem_path = state_dir / f"{agent_name}_memory.json"
    if not mem_path.exists():
        print(f"⚠️  {agent_name} memory file not found at {mem_path}")
        continue
    with open(mem_path, encoding="utf-8") as f:
        mem = json.load(f)
    before = len(mem.get("failures", []))
    mem["failures"] = [
        entry for entry in mem.get("failures", [])
        if entry.get("scraper") not in scrapers
    ]
    after = len(mem["failures"])
    tmp = mem_path.with_suffix(".tmp")
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(mem, f, indent=2, ensure_ascii=False)
    os.replace(tmp, mem_path)
    print(f"✅ {agent_name}: cleared {before - after} failure entries (scrapers: {scrapers})")
    print(f"   Remaining failures: {after}")

print("\nFailure reset complete — re-run the pipeline to retry these scrapers.")
print()

# ── Block B: Inspect semantic knowledge store ─────────────────────────────────
# Shows what each agent has learned across runs.

print("=" * 55)
print("SEMANTIC KNOWLEDGE STORE")
print("=" * 55)
for agent_name in ["ListingsAgent", "ArticlesAgent", "StandardizationAgent",
                   "CleaningAgent", "OrchestratorAgent", "ScraperHealthAgent"]:
    mem_path = state_dir / f"{agent_name}_memory.json"
    if not mem_path.exists():
        continue
    with open(mem_path, encoding="utf-8") as f:
        mem = json.load(f)
    knowledge = mem.get("knowledge", {})
    schema_repairs = mem.get("schema_repairs", [])
    if knowledge or schema_repairs:
        print(f"\n  {agent_name}:")
        if knowledge:
            for k, v in knowledge.items():
                if k != "_last_updated":
                    print(f"    knowledge.{k} = {v}")
            if "_last_updated" in knowledge:
                print(f"    (last updated: {knowledge['_last_updated']})")
        if schema_repairs:
            print(f"    schema_repairs: {len(schema_repairs)} inferred mappings")
            for repair in schema_repairs[-2:]:
                print(f"      [{repair.get('ts','')[:10]}] {repair.get('path','')} → {repair.get('rename',{})}")
print()

# ── Block C: Write a manual knowledge entry ───────────────────────────────────
# Use this to inject facts you've observed manually into an agent's knowledge.
# Example: after noticing mubawab always returns ~1200 rows on weekdays:
#
# agent = ListingsAgent()
# agent.memory.know("mubawab.typical_rows", 1200)
# agent.memory.know("mubawab.known_issues", ["slow on sundays 2-4am", "may 403 on weekends"])
# print("Knowledge written ✅")
````

<h2 style="color: #87BAC3;">🔍 Inspect Agent Memory After Run</h2>
<p>After the pipeline runs, inspect each agent's memory to audit decisions, review quality snapshots, and debug failures.</p>

````python
# ── Full pipeline memory dump  (Phases 1-6) ──────────────────────────────────
# Run after a pipeline execution to inspect all agent memories.

all_agents = [
    _health_agent,
    _articles_agent,
    _listings_agent,
    _std_agent,
    _cleaning_agent,
    _quality_agent,
    _orchestrator,
]

CATEGORIES = [
    "successes",
    "failures",
    "decisions",
    "quality",
    "schema_repairs",
    "suspicious_flags",
]

for agent in all_agents:
    print(f"\n{'─'*60}")
    print(f"  {agent.name}")
    summary = agent.memory.summary()
    print(f"  successes={summary['successes']}  failures={summary['failures']}  "
          f"decisions={summary['decisions']}")
    if summary["knowledge"]:
        print(f"  knowledge keys: {summary['knowledge']}")
    print(f"{'─'*60}")

    for cat in CATEGORIES:
        items = agent.memory.recall(cat, 3)
        if not items:
            continue
        print(f"  {cat} (last {len(items)}):")
        for item in items:
            ts      = item.get("ts", "")[:19]
            rest    = {k: v for k, v in item.items() if k != "ts"}
            rest_str = json.dumps(rest, ensure_ascii=False)
            if len(rest_str) > 200:
                rest_str = rest_str[:197] + "..."
            print(f"    [{ts}] {rest_str}")

    knowledge = agent.memory.recall_knowledge()
    if knowledge:
        print(f"  knowledge store:")
        for k, v in knowledge.items():
            if k != "_last_updated":
                print(f"    {k}: {v}")

print(f"\n{'─'*60}")
print("Memory dump complete.")
print(f"Heuristics: {HEURISTICS.summary()}")
print(f"{'─'*60}")
print()
print("Quality reports available in logs/quality_*.html")
print("Run reports available in logs/run_*.json")
````
